```python
# ==================================================
# EXERCISE 2.1: LEARNING CURVES COMPARISON
# ==================================================

print("\n" + "=" * 80)
print("EXERCISE 2.1: Learning Curves - All Algorithms")
print("=" * 80)

print("\n📊 Creating Learning Curves Comparison:")
print("=" * 80)

# Simulate learning curves for all algorithms
# (In practice, you'd load actual training data from Days 71-76)

np.random.seed(42)
episodes = 1000

# REINFORCE: High variance, slow convergence
reinforce_curve = []
for i in range(episodes):
    base = -150 + 270 * (1 - np.exp(-i/600))
    noise = np.random.normal(0, 80)
    reinforce_curve.append(min(250, max(-250, base + noise)))

# A2C: Medium variance, moderate convergence
a2c_curve = []
for i in range(episodes):
    base = -100 + 250 * (1 - np.exp(-i/400))
    noise = np.random.normal(0, 60)
    a2c_curve.append(min(250, max(-250, base + noise)))

# PPO: Low variance, fast convergence
ppo_curve = []
for i in range(episodes):
    base = -50 + 250 * (1 - np.exp(-i/300))
    noise = np.random.normal(0, 40)
    ppo_curve.append(min(250, max(-250, base + noise)))

# Create comprehensive learning curves visualization
fig, axes = plt.subplots(2, 2, figsize=(18, 10))
fig.suptitle('Learning Curves Comparison - All Algorithms', fontsize=18, fontweight='bold')

# 1. Raw Learning Curves
ax1 = axes[0, 0]
episodes_range = range(1, episodes + 1)

ax1.plot(episodes_range, reinforce_curve, alpha=0.3, linewidth=0.5, 
        color='orange', label='REINFORCE')
ax1.plot(episodes_range, a2c_curve, alpha=0.3, linewidth=0.5,
        color='blue', label='A2C')
ax1.plot(episodes_range, ppo_curve, alpha=0.3, linewidth=0.5,
        color='green', label='PPO')

ax1.axhline(200, color='green', linestyle='--', linewidth=2, alpha=0.5, label='Solved')
ax1.axhline(0, color='gray', linestyle='-', linewidth=1, alpha=0.3)

ax1.set_xlabel('Episode', fontsize=11, fontweight='bold')
ax1.set_ylabel('Reward', fontsize=11, fontweight='bold')
ax1.set_title('Raw Learning Curves (LunarLander)', fontsize=13, fontweight='bold')
ax1.legend()
ax1.grid(True, alpha=0.3)

# 2. Smoothed Learning Curves
ax2 = axes[0, 1]

reinforce_ma = pd.Series(reinforce_curve).rolling(window=100).mean()
a2c_ma = pd.Series(a2c_curve).rolling(window=100).mean()
ppo_ma = pd.Series(ppo_curve).rolling(window=100).mean()

ax2.plot(episodes_range, reinforce_ma, linewidth=3, color='orange', label='REINFORCE')
ax2.plot(episodes_range, a2c_ma, linewidth=3, color='blue', label='A2C')
ax2.plot(episodes_range, ppo_ma, linewidth=3, color='green', label='PPO')

ax2.axhline(200, color='green', linestyle='--', linewidth=2, alpha=0.5, label='Solved')
ax2.axhline(100, color='orange', linestyle='--', linewidth=2, alpha=0.5, label='Good')
ax2.axhline(0, color='gray', linestyle='-', linewidth=1, alpha=0.3)

ax2.set_xlabel('Episode', fontsize=11, fontweight='bold')
ax2.set_ylabel('Reward (100-ep MA)', fontsize=11, fontweight='bold')
ax2.set_title('Smoothed Learning Curves', fontsize=13, fontweight='bold')
ax2.legend()
ax2.grid(True, alpha=0.3)

# Add annotations
ax2.annotate('PPO learns fastest!', xy=(300, 150), xytext=(500, 220),
            arrowprops=dict(arrowstyle='->', color='green', lw=2),
            fontsize=11, color='green', fontweight='bold')

# 3. Convergence Analysis
ax3 = axes[1, 0]

# Find convergence episodes (when > 100 avg for 50 episodes)
def find_convergence(curve, threshold=100, window=50):
    ma = pd.Series(curve).rolling(window=window).mean()
    for i, val in enumerate(ma):
        if val > threshold and i > window:
            return i
    return len(curve)

reinforce_conv = find_convergence(reinforce_curve)
a2c_conv = find_convergence(a2c_curve)
ppo_conv = find_convergence(ppo_curve)

conv_episodes = [reinforce_conv, a2c_conv, ppo_conv]
algorithms = ['REINFORCE', 'A2C', 'PPO']
colors = ['orange', 'blue', 'green']

bars = ax3.barh(algorithms, conv_episodes, color=colors, alpha=0.7,
               edgecolor='black', linewidth=2)
ax3.set_xlabel('Episodes to Convergence', fontsize=11, fontweight='bold')
ax3.set_title('Convergence Speed Comparison', fontsize=13, fontweight='bold')
ax3.grid(True, alpha=0.3, axis='x')

for bar, ep in zip(bars, conv_episodes):
    width = bar.get_width()
    ax3.text(width, bar.get_y() + bar.get_height()/2.,
            f'  {ep} eps', ha='left', va='center', fontweight='bold')

speedup_vs_reinforce = [(reinforce_conv - conv) / reinforce_conv * 100 for conv in conv_episodes]
ax3.text(0.98, 0.95, f'PPO: {speedup_vs_reinforce[2]:.0f}% faster\nthan REINFORCE!',
        transform=ax3.transAxes, ha='right', va='top',
        bbox=dict(boxstyle='round', facecolor='lightgreen', alpha=0.7),
        fontsize=11, fontweight='bold')

# 4. Variance Over Time
ax4 = axes[1, 1]

# Compute rolling variance
window = 50
reinforce_var = pd.Series(reinforce_curve).rolling(window=window).std()
a2c_var = pd.Series(a2c_curve).rolling(window=window).std()
ppo_var = pd.Series(ppo_curve).rolling(window=window).std()

ax4.plot(episodes_range, reinforce_var, linewidth=2, color='orange', label='REINFORCE')
ax4.plot(episodes_range, a2c_var, linewidth=2, color='blue', label='A2C')
ax4.plot(episodes_range, ppo_var, linewidth=2, color='green', label='PPO')

ax4.set_xlabel('Episode', fontsize=11, fontweight='bold')
ax4.set_ylabel('Standard Deviation (50-ep window)', fontsize=11, fontweight='bold')
ax4.set_title('Variance Over Time', fontsize=13, fontweight='bold')
ax4.legend()
ax4.grid(True, alpha=0.3)

# Calculate average variance
avg_vars = [
    reinforce_var[window:].mean(),
    a2c_var[window:].mean(),
    ppo_var[window:].mean()
]

ax4.text(0.98, 0.95, 
        f'Avg Variance:\n'
        f'REINFORCE: {avg_vars[0]:.1f}\n'
        f'A2C: {avg_vars[1]:.1f}\n'
        f'PPO: {avg_vars[2]:.1f}',
        transform=ax4.transAxes, ha='right', va='top',
        bbox=dict(boxstyle='round', facecolor='lightyellow', alpha=0.7),
        fontsize=10, fontfamily='monospace')

plt.tight_layout()
plt.savefig('week11_learning_curves.png', dpi=300, bbox_inches='tight')
print("✅ Learning curves saved!")
plt.show()

print("\n💡 Key Observations:")
print("=" * 80)
print(f"""
Convergence Speed:
━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━
REINFORCE: ~{reinforce_conv} episodes
A2C: ~{a2c_conv} episodes ({(reinforce_conv-a2c_conv)/reinforce_conv*100:.0f}% faster)
PPO: ~{ppo_conv} episodes ({(reinforce_conv-ppo_conv)/reinforce_conv*100:.0f}% faster)

Variance:
━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━
REINFORCE: {avg_vars[0]:.1f} (highest)
A2C: {avg_vars[1]:.1f} (medium)
PPO: {avg_vars[2]:.1f} (lowest)

PPO shows:
✓ Fastest convergence
✓ Lowest variance
✓ Most stable learning
✓ Highest final performance

Clear winner! 🏆
""")

print("\n✅ Exercise 2.1 Complete!")
print("=" * 80)
```
