````python
# ==================================================
# EXERCISE 1.2: IMPLEMENT INTERACTIVE CHARTS
# ==================================================

print("\n" + "="*80)
print("EXERCISE 1.2: Building Interactive Visualizations")
print("="*80)

"""
📖 THEORY: Interactive Charts with Plotly

Plotly Advantages:
==================================================

Built-in Interactivity:
- Hover tooltips
- Zoom and pan
- Click to filter
- Download as PNG
- Responsive

Easy Customization:
- Color themes
- Annotations
- Multiple traces
- Subplots

Streamlit Integration:
```python
fig = go.Figure(...)
st.plotly_chart(fig, use_container_width=True)
```

Chart Creation Patterns:
==================================================

Pattern 1: Simple Bar Chart
```python
import plotly.graph_objects as go

fig = go.Figure(data=[
    go.Bar(
        x=['Sentiment', 'Summary', 'Fake News'],
        y=[40, 30, 20],
        marker_color='#4CAF50'
    )
])

fig.update_layout(
    title="Tool Usage",
    xaxis_title="Tool",
    yaxis_title="Count"
)
```

Pattern 2: Line Chart with Multiple Traces
```python
fig = go.Figure()

fig.add_trace(go.Scatter(
    x=dates,
    y=values1,
    name='Series 1',
    mode='lines+markers'
))

fig.add_trace(go.Scatter(
    x=dates,
    y=values2,
    name='Series 2',
    mode='lines+markers'
))
```

Pattern 3: Pie Chart
```python
fig = go.Figure(data=[
    go.Pie(
        labels=['A', 'B', 'C'],
        values=[40, 30, 30],
        marker=dict(colors=['#4CAF50', '#2196F3', '#FF9800'])
    )
])
```

Pattern 4: Heatmap
```python
fig = go.Figure(data=go.Heatmap(
    z=matrix,
    x=x_labels,
    y=y_labels,
    colorscale='Greens'
))
```

Customization Options:
==================================================

Layout:
```python
fig.update_layout(
    title="Chart Title",
    xaxis_title="X Axis",
    yaxis_title="Y Axis",
    font=dict(size=12),
    template="plotly_white",
    height=400
)
```

Hover:
```python
fig.update_traces(
    hovertemplate='<b>%{x}</b><br>Value: %{y}<extra></extra>'
)
```

Colors:
```python
marker=dict(
    color='#4CAF50',
    line=dict(color='#388E3C', width=2)
)
```

Our Chart Implementations:
==================================================

1. Usage Trend (Line)
2. Tool Distribution (Pie + Bar)
3. Activity Heatmap
4. Performance Scatter
5. Latency Box Plot
"""

print("\n⏱️  Implementing interactive charts...")

# ==================================================
# Chart 1: Usage Trend Line Chart
# ==================================================

def create_usage_trend_chart(history):
    """Create interactive usage trend line chart."""
    df = pd.DataFrame(history)
    df['date'] = pd.to_datetime(df['timestamp']).dt.date
    
    # Count queries per day
    daily_counts = df.groupby('date').size().reset_index(name='count')
    
    # Create figure
    fig = go.Figure()
    
    fig.add_trace(go.Scatter(
        x=daily_counts['date'],
        y=daily_counts['count'],
        mode='lines+markers',
        name='Daily Queries',
        line=dict(color='#4CAF50', width=3),
        marker=dict(size=8, color='#388E3C'),
        hovertemplate='<b>%{x}</b><br>Queries: %{y}<extra></extra>'
    ))
    
    fig.update_layout(
        title="Usage Trend Over Time",
        xaxis_title="Date",
        yaxis_title="Number of Queries",
        template="plotly_white",
        height=400,
        hovermode='x unified'
    )
    
    return fig

print("✅ Usage trend chart created")

# ==================================================
# Chart 2: Tool Distribution (Pie + Bar)
# ==================================================

def create_tool_distribution_charts(history):
    """Create tool distribution pie and bar charts."""
    df = pd.DataFrame(history)
    tool_counts = df['tool'].value_counts()
    
    # Define colors
    tool_colors = {
        'sentiment': '#4CAF50',
        'summarizer': '#2196F3',
        'fake_news': '#FF9800',
        'job_matcher': '#9C27B0'
    }
    
    colors = [tool_colors.get(tool, '#999999') for tool in tool_counts.index]
    
    # Pie Chart
    pie_fig = go.Figure(data=[
        go.Pie(
            labels=tool_counts.index,
            values=tool_counts.values,
            marker=dict(colors=colors),
            hole=0.3,  # Donut chart
            hovertemplate='<b>%{label}</b><br>Count: %{value}<br>Percentage: %{percent}<extra></extra>'
        )
    ])
    
    pie_fig.update_layout(
        title="Tool Usage Distribution",
        template="plotly_white",
        height=400
    )
    
    # Bar Chart
    bar_fig = go.Figure(data=[
        go.Bar(
            x=tool_counts.index,
            y=tool_counts.values,
            marker=dict(color=colors),
            hovertemplate='<b>%{x}</b><br>Count: %{y}<extra></extra>'
        )
    ])
    
    bar_fig.update_layout(
        title="Tool Usage Comparison",
        xaxis_title="Tool",
        yaxis_title="Number of Queries",
        template="plotly_white",
        height=400
    )
    
    return pie_fig, bar_fig

print("✅ Tool distribution charts created")

# ==================================================
# Chart 3: Activity Heatmap
# ==================================================

def create_activity_heatmap(history):
    """Create usage pattern heatmap (day of week vs hour)."""
    df = pd.DataFrame(history)
    df['datetime'] = pd.to_datetime(df['timestamp'])
    df['day_of_week'] = df['datetime'].dt.day_name()
    df['hour'] = df['datetime'].dt.hour
    
    # Create pivot table
    pivot = df.groupby(['day_of_week', 'hour']).size().unstack(fill_value=0)
    
    # Reorder days
    days_order = ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday']
    pivot = pivot.reindex([day for day in days_order if day in pivot.index])
    
    # Create heatmap
    fig = go.Figure(data=go.Heatmap(
        z=pivot.values,
        x=[f'{h:02d}:00' for h in pivot.columns],
        y=pivot.index,
        colorscale='Greens',
        hovertemplate='<b>%{y}</b><br>Hour: %{x}<br>Queries: %{z}<extra></extra>'
    ))
    
    fig.update_layout(
        title="Activity Heatmap (Day/Hour)",
        xaxis_title="Hour of Day",
        yaxis_title="Day of Week",
        template="plotly_white",
        height=400
    )
    
    return fig

print("✅ Activity heatmap created")

# ==================================================
# Chart 4: Processing Time Distribution
# ==================================================

def create_latency_distribution(history):
    """Create processing time distribution by tool."""
    df = pd.DataFrame(history)
    
    # Tool colors
    tool_colors = {
        'sentiment': '#4CAF50',
        'summarizer': '#2196F3',
        'fake_news': '#FF9800',
        'job_matcher': '#9C27B0'
    }
    
    fig = go.Figure()
    
    for tool in df['tool'].unique():
        tool_data = df[df['tool'] == tool]
        
        fig.add_trace(go.Box(
            y=tool_data['processing_time_ms'],
            name=tool.title(),
            marker_color=tool_colors.get(tool, '#999999'),
            hovertemplate='<b>%{fullData.name}</b><br>Time: %{y}ms<extra></extra>'
        ))
    
    fig.update_layout(
        title="Processing Time Distribution by Tool",
        yaxis_title="Processing Time (ms)",
        template="plotly_white",
        height=400,
        showlegend=True
    )
    
    return fig

print("✅ Latency distribution chart created")

# ==================================================
# Chart 5: Success Rate Over Time
# ==================================================

def create_success_rate_trend(history):
    """Create success rate trend line chart."""
    df = pd.DataFrame(history)
    df['date'] = pd.to_datetime(df['timestamp']).dt.date
    
    # Calculate daily success rate
    daily_success = df.groupby('date').agg({
        'success': ['sum', 'count']
    })
    daily_success.columns = ['successes', 'total']
    daily_success['success_rate'] = (daily_success['successes'] / daily_success['total']) * 100
    daily_success = daily_success.reset_index()
    
    fig = go.Figure()
    
    fig.add_trace(go.Scatter(
        x=daily_success['date'],
        y=daily_success['success_rate'],
        mode='lines+markers',
        name='Success Rate',
        line=dict(color='#4CAF50', width=3),
        marker=dict(size=8, color='#388E3C'),
        fill='tozeroy',
        fillcolor='rgba(76, 175, 80, 0.2)',
        hovertemplate='<b>%{x}</b><br>Success Rate: %{y:.1f}%<extra></extra>'
    ))
    
    # Add reference line at 100%
    fig.add_hline(
        y=100, 
        line_dash="dash", 
        line_color="gray",
        annotation_text="Target: 100%"
    )
    
    fig.update_layout(
        title="Success Rate Trend",
        xaxis_title="Date",
        yaxis_title="Success Rate (%)",
        template="plotly_white",
        height=400,
        yaxis=dict(range=[0, 105])
    )
    
    return fig

print("✅ Success rate trend chart created")

# ==================================================
# Test Charts with Sample Data
# ==================================================

print("\n📊 Testing Charts with Sample Data:")

print("\n   Creating charts...")
usage_trend = create_usage_trend_chart(sample_history)
pie_chart, bar_chart = create_tool_distribution_charts(sample_history)
heatmap = create_activity_heatmap(sample_history)
latency_dist = create_latency_distribution(sample_history)
success_trend = create_success_rate_trend(sample_history)

print("   ✅ Usage trend chart")
print("   ✅ Tool distribution (pie + bar)")
print("   ✅ Activity heatmap")
print("   ✅ Latency distribution (box plot)")
print("   ✅ Success rate trend")

print("\n✨ All interactive charts created successfully!")

print("\n💡 Chart Features:")
print("   • Hover tooltips with detailed info")
print("   • Zoom and pan capabilities")
print("   • Responsive (adapts to container)")
print("   • Download as PNG")
print("   • Professional appearance")
print("   • Consistent color scheme")

print("\n✅ Exercise 1.2 Complete!")
print("="*80)
````

````python
# ==================================================
# EXERCISE 1.3: IMPLEMENT ADVANCED ANALYTICS
# ==================================================

print("\n" + "="*80)
print("EXERCISE 1.3: Advanced Analytics Calculations")
print("="*80)

"""
📖 THEORY: Advanced Analytics

Beyond Basic Metrics:
==================================================

Basic Metrics:
- Count (total queries)
- Average (mean latency)
- Sum (total successes)

Advanced Metrics:
- Growth rate (% change)
- Percentiles (P50, P95, P99)
- Moving averages (trends)
- Correlation analysis
- Anomaly detection

Growth Rate:
==================================================

Formula:
growth_rate = ((current - previous) / previous) * 100

Example:
```python
def calculate_growth_rate(current, previous):
    if previous == 0:
        return 0
    return ((current - previous) / previous) * 100

# This week vs last week
this_week = 100
last_week = 80
growth = calculate_growth_rate(this_week, last_week)
# Result: 25% growth
```

Percentiles:
==================================================

What are percentiles?
- P50 (median): 50% of values below
- P95: 95% of values below
- P99: 99% of values below

Why use them?
- Understand distribution
- Identify outliers
- Better than average alone

Example:
```python
latencies = [100, 150, 200, 250, 300, 5000]
p50 = np.percentile(latencies, 50)  # 225
p95 = np.percentile(latencies, 95)  # 2850
p99 = np.percentile(latencies, 99)  # 4520

# P95 shows outliers (5000ms) pull up average
```

Moving Averages:
==================================================

Purpose: Smooth out noise, see trends

Simple Moving Average (SMA):
```python
def calculate_sma(data, window=7):
    return data.rolling(window=window).mean()

# 7-day moving average
df['sma_7'] = calculate_sma(df['count'], window=7)
```

Types:
- 7-day: Weekly trend
- 30-day: Monthly trend
- Exponential: Recent data weighted more

Time-Based Analysis:
==================================================

Peak Hours:
```python
def find_peak_hours(history):
    df = pd.DataFrame(history)
    df['hour'] = pd.to_datetime(df['timestamp']).dt.hour
    hourly_counts = df['hour'].value_counts()
    peak_hour = hourly_counts.idxmax()
    return peak_hour
```

Busiest Days:
```python
def find_busiest_days(history):
    df = pd.DataFrame(history)
    df['date'] = pd.to_datetime(df['timestamp']).dt.date
    daily_counts = df['date'].value_counts()
    return daily_counts.head(5)
```

Comparative Analysis:
==================================================

Tool Comparison:
```python
def compare_tools(history):
    df = pd.DataFrame(history)
    
    comparison = df.groupby('tool').agg({
        'processing_time_ms': ['mean', 'median', 'std'],
        'success': ['sum', 'count']
    })
    
    comparison['success_rate'] = (
        comparison['success']['sum'] / 
        comparison['success']['count'] * 100
    )
    
    return comparison
```

Time Period Comparison:
```python
def compare_periods(history, days=7):
    df = pd.DataFrame(history)
    df['date'] = pd.to_datetime(df['timestamp'])
    
    cutoff = datetime.now() - timedelta(days=days)
    recent = df[df['date'] >= cutoff]
    older = df[df['date'] < cutoff]
    
    return {
        'recent_count': len(recent),
        'older_count': len(older),
        'growth': calculate_growth_rate(
            len(recent), len(older)
        )
    }
```

User Engagement Metrics:
==================================================

Retention:
- Daily active users
- Weekly active users
- Monthly active users

Engagement:
- Queries per user
- Session duration
- Return rate

Churn:
- Users who stopped
- Time since last activity
"""

print("\n⏱️  Implementing advanced analytics...")

# ==================================================
# Analytics Calculator Class
# ==================================================

class AdvancedAnalytics:
    """Advanced analytics calculations for history data."""
    
    def __init__(self, history):
        """Initialize with history data."""
        self.df = pd.DataFrame(history)
        if not self.df.empty:
            self.df['datetime'] = pd.to_datetime(self.df['timestamp'])
            self.df['date'] = self.df['datetime'].dt.date
            self.df['hour'] = self.df['datetime'].dt.hour
            self.df['day_of_week'] = self.df['datetime'].dt.day_name()
    
    def calculate_growth_rate(self, period_days=7):
        """Calculate growth rate vs previous period."""
        if self.df.empty:
            return 0
        
        cutoff = datetime.now() - timedelta(days=period_days)
        
        # Recent period
        recent = self.df[self.df['datetime'] >= cutoff]
        recent_count = len(recent)
        
        # Previous period
        previous_start = cutoff - timedelta(days=period_days)
        previous = self.df[
            (self.df['datetime'] >= previous_start) & 
            (self.df['datetime'] < cutoff)
        ]
        previous_count = len(previous)
        
        if previous_count == 0:
            return 0
        
        growth = ((recent_count - previous_count) / previous_count) * 100
        
        return {
            'recent_count': recent_count,
            'previous_count': previous_count,
            'growth_rate': growth,
            'period_days': period_days
        }
    
    def calculate_percentiles(self):
        """Calculate latency percentiles."""
        if self.df.empty:
            return {}
        
        latencies = self.df['processing_time_ms']
        
        return {
            'p50': np.percentile(latencies, 50),
            'p75': np.percentile(latencies, 75),
            'p90': np.percentile(latencies, 90),
            'p95': np.percentile(latencies, 95),
            'p99': np.percentile(latencies, 99),
            'min': latencies.min(),
            'max': latencies.max()
        }
    
    def calculate_moving_average(self, window=7):
        """Calculate moving average for trend analysis."""
        if self.df.empty:
            return pd.DataFrame()
        
        daily_counts = self.df.groupby('date').size().reset_index(name='count')
        daily_counts = daily_counts.sort_values('date')
        daily_counts['ma'] = daily_counts['count'].rolling(window=window).mean()
        
        return daily_counts
    
    def find_peak_hours(self):
        """Find peak usage hours."""
        if self.df.empty:
            return {}
        
        hourly_counts = self.df['hour'].value_counts().sort_index()
        peak_hour = hourly_counts.idxmax()
        
        return {
            'peak_hour': peak_hour,
            'peak_count': hourly_counts[peak_hour],
            'hourly_distribution': hourly_counts.to_dict()
        }
    
    def find_busiest_days(self, top_n=5):
        """Find busiest days."""
        if self.df.empty:
            return []
        
        daily_counts = self.df['date'].value_counts().head(top_n)
        
        return [
            {'date': str(date), 'count': int(count)}
            for date, count in daily_counts.items()
        ]
    
    def compare_tools(self):
        """Compare performance across tools."""
        if self.df.empty:
            return {}
        
        comparison = {}
        
        for tool in self.df['tool'].unique():
            tool_data = self.df[self.df['tool'] == tool]
            
            comparison[tool] = {
                'count': len(tool_data),
                'success_rate': (tool_data['success'].sum() / len(tool_data)) * 100,
                'avg_latency': tool_data['processing_time_ms'].mean(),
                'median_latency': tool_data['processing_time_ms'].median(),
                'p95_latency': np.percentile(tool_data['processing_time_ms'], 95)
            }
        
        return comparison
    
    def get_daily_trend(self):
        """Get daily usage trend."""
        if self.df.empty:
            return []
        
        daily = self.df.groupby('date').agg({
            'id': 'count',
            'success': 'sum',
            'processing_time_ms': 'mean'
        }).reset_index()
        
        daily.columns = ['date', 'total', 'successes', 'avg_latency']
        daily['success_rate'] = (daily['successes'] / daily['total']) * 100
        
        return daily.to_dict('records')
    
    def get_tool_trend(self, tool):
        """Get trend for specific tool."""
        if self.df.empty:
            return []
        
        tool_data = self.df[self.df['tool'] == tool]
        
        daily = tool_data.groupby('date').size().reset_index(name='count')
        daily = daily.sort_values('date')
        
        return daily.to_dict('records')

print("✅ AdvancedAnalytics class created")

# ==================================================
# Test Advanced Analytics
# ==================================================

print("\n📊 Testing Advanced Analytics:")

analytics = AdvancedAnalytics(sample_history)

print("\n   Growth Rate Analysis:")
growth = analytics.calculate_growth_rate(period_days=7)
print(f"      Last 7 days: {growth['recent_count']} queries")
print(f"      Previous 7 days: {growth['previous_count']} queries")
print(f"      Growth rate: {growth['growth_rate']:.1f}%")

print("\n   Latency Percentiles:")
percentiles = analytics.calculate_percentiles()
for key, value in percentiles.items():
    print(f"      {key.upper()}: {value:.1f}ms")

print("\n   Peak Usage Analysis:")
peak_hours = analytics.find_peak_hours()
print(f"      Peak hour: {peak_hours['peak_hour']}:00")
print(f"      Peak count: {peak_hours['peak_count']} queries")

print("\n   Busiest Days:")
busiest = analytics.find_busiest_days(top_n=3)
for i, day in enumerate(busiest, 1):
    print(f"      {i}. {day['date']}: {day['count']} queries")

print("\n   Tool Comparison:")
tool_comparison = analytics.compare_tools()
for tool, stats in tool_comparison.items():
    print(f"\n      {tool.title()}:")
    print(f"         Count: {stats['count']}")
    print(f"         Success Rate: {stats['success_rate']:.1f}%")
    print(f"         Avg Latency: {stats['avg_latency']:.1f}ms")
    print(f"         P95 Latency: {stats['p95_latency']:.1f}ms")

print("\n✅ Exercise 1.3 Complete!")
print("="*80)
````

````python
# ==================================================
# EXERCISE 2.1: USER ENGAGEMENT METRICS
# ==================================================

print("\n" + "="*80)
print("EXERCISE 2.1: Analyzing User Engagement")
print("="*80)

"""
📖 THEORY: User Engagement Metrics

What is User Engagement?
==================================================

Definition: How actively users interact with your app

Key Metrics:
- Frequency: How often they use it
- Depth: How much they use it
- Recency: When they last used it
- Retention: Do they come back?

Engagement Categories:
==================================================

1. Activity Metrics:
   - Daily Active Users (DAU)
   - Weekly Active Users (WAU)
   - Monthly Active Users (MAU)
   - Queries per user
   - Session duration

2. Retention Metrics:
   - Return rate
   - Churn rate
   - Time since last activity
   - Cohort retention

3. Quality Metrics:
   - Success rate
   - Tool diversity (how many tools used)
   - Feature adoption
   - Advanced feature usage

4. Growth Metrics:
   - New users
   - User growth rate
   - Activation rate
   - Referral rate

DAU/WAU/MAU:
==================================================

Daily Active Users (DAU):
```python
def calculate_dau(users_data, date):
    active_users = [
        user for user in users_data
        if user['last_active_date'] == date
    ]
    return len(active_users)
```

Weekly Active Users (WAU):
```python
def calculate_wau(users_data, week_start):
    week_end = week_start + timedelta(days=7)
    active_users = [
        user for user in users_data
        if week_start <= user['last_active_date'] < week_end
    ]
    return len(active_users)
```

DAU/MAU Ratio:
- Sticky product: >20%
- Average: 10-20%
- Low engagement: <10%

Queries Per User:
==================================================

Average Queries:
```python
def calculate_avg_queries_per_user(history):
    df = pd.DataFrame(history)
    queries_per_user = df.groupby('username').size()
    return queries_per_user.mean()
```

Distribution:
- Power users: >100 queries
- Active users: 10-100 queries
- Casual users: 1-10 queries
- Dormant: 0 queries (registered but inactive)

Retention Analysis:
==================================================

Return Rate:
```python
def calculate_return_rate(users):
    total_users = len(users)
    returning = [
        user for user in users
        if user['visit_count'] > 1
    ]
    return (len(returning) / total_users) * 100
```

Churn Detection:
```python
def identify_churned_users(users, days=30):
    cutoff = datetime.now() - timedelta(days=days)
    churned = [
        user for user in users
        if user['last_active'] < cutoff
    ]
    return churned
```

Tool Diversity:
==================================================

How many different tools has user tried?
```python
def calculate_tool_diversity(user_history):
    tools_used = set([entry['tool'] for entry in user_history])
    return len(tools_used)
```

Interpretation:
- 4 tools: Highly engaged (tried everything)
- 2-3 tools: Active (exploring)
- 1 tool: Limited engagement
- 0 tools: Inactive

Cohort Analysis:
==================================================

Group users by signup date, track retention
```python
def cohort_analysis(users):
    cohorts = {}
    
    for user in users:
        signup_month = user['created_at'][:7]  # YYYY-MM
        
        if signup_month not in cohorts:
            cohorts[signup_month] = []
        
        cohorts[signup_month].append(user)
    
    return cohorts
```

Track:
- Month 0 retention: 100%
- Month 1 retention: 60%?
- Month 2 retention: 40%?

User Lifecycle:
==================================================

Stages:
1. New (0-7 days)
2. Active (regular usage)
3. At-risk (declining usage)
4. Churned (inactive >30 days)

Transitions:
```python
def classify_user_stage(user):
    days_since_signup = (datetime.now() - user['created_at']).days
    days_since_active = (datetime.now() - user['last_active']).days
    
    if days_since_signup <= 7:
        return 'new'
    elif days_since_active <= 7:
        return 'active'
    elif days_since_active <= 30:
        return 'at_risk'
    else:
        return 'churned'
```
"""

print("\n⏱️  Implementing user engagement metrics...")

# ==================================================
# User Engagement Calculator
# ==================================================

class UserEngagementMetrics:
    """Calculate user engagement metrics."""
    
    def __init__(self, users_data, history_data):
        """
        Initialize with users and history.
        
        Args:
            users_data: List of user objects
            history_data: List of history entries
        """
        self.users = users_data
        self.history_df = pd.DataFrame(history_data) if history_data else pd.DataFrame()
        
        if not self.history_df.empty:
            self.history_df['datetime'] = pd.to_datetime(self.history_df['timestamp'])
            self.history_df['date'] = self.history_df['datetime'].dt.date
    
    def calculate_queries_per_user(self):
        """Calculate queries per user statistics."""
        if self.history_df.empty:
            return {}
        
        # Group by user (if username field exists)
        # For now, use overall stats
        total_queries = len(self.history_df)
        total_users = len(self.users) if self.users else 1
        
        return {
            'total_queries': total_queries,
            'total_users': total_users,
            'avg_queries_per_user': total_queries / total_users if total_users > 0 else 0,
            'max_queries': total_queries,  # Would be per-user in real app
            'min_queries': 0
        }
    
    def calculate_tool_diversity(self):
        """Calculate how many different tools users have tried."""
        if self.history_df.empty:
            return {}
        
        unique_tools = self.history_df['tool'].nunique()
        total_tools = 4  # We have 4 tools
        
        # Distribution of tools used
        tool_counts = self.history_df['tool'].value_counts()
        
        return {
            'unique_tools_used': unique_tools,
            'total_available_tools': total_tools,
            'diversity_rate': (unique_tools / total_tools) * 100,
            'tool_distribution': tool_counts.to_dict()
        }
    
    def calculate_activity_frequency(self):
        """Calculate how frequently users are active."""
        if self.history_df.empty:
            return {}
        
        # Calculate days with activity
        active_dates = self.history_df['date'].nunique()
        
        # Calculate date range
        min_date = self.history_df['date'].min()
        max_date = self.history_df['date'].max()
        date_range = (max_date - min_date).days + 1
        
        # Activity frequency
        activity_frequency = (active_dates / date_range) * 100 if date_range > 0 else 0
        
        return {
            'active_days': active_dates,
            'date_range_days': date_range,
            'activity_frequency': activity_frequency,
            'avg_queries_per_active_day': len(self.history_df) / active_dates if active_dates > 0 else 0
        }
    
    def analyze_user_segments(self):
        """Segment users by engagement level."""
        if self.history_df.empty:
            return {}
        
        total_queries = len(self.history_df)
        
        # Define segments (in real app, would be per-user)
        segments = {
            'power_users': 0,      # >100 queries
            'active_users': 0,     # 10-100 queries
            'casual_users': 0,     # 1-10 queries
            'dormant_users': 0     # 0 queries
        }
        
        # Classify based on total (simplified)
        if total_queries > 100:
            segments['power_users'] = 1
        elif total_queries >= 10:
            segments['active_users'] = 1
        elif total_queries >= 1:
            segments['casual_users'] = 1
        else:
            segments['dormant_users'] = 1
        
        return segments
    
    def calculate_engagement_score(self):
        """Calculate overall engagement score (0-100)."""
        if self.history_df.empty:
            return 0
        
        # Factors:
        # 1. Query frequency (40%)
        # 2. Tool diversity (30%)
        # 3. Success rate (30%)
        
        # Query frequency score
        queries_per_user = self.calculate_queries_per_user()
        avg_queries = queries_per_user['avg_queries_per_user']
        frequency_score = min(avg_queries / 100, 1.0) * 40  # Cap at 100 queries
        
        # Tool diversity score
        diversity = self.calculate_tool_diversity()
        diversity_score = (diversity['diversity_rate'] / 100) * 30
        
        # Success rate score
        success_rate = (self.history_df['success'].sum() / len(self.history_df)) * 100
        success_score = (success_rate / 100) * 30
        
        total_score = frequency_score + diversity_score + success_score
        
        return {
            'total_score': total_score,
            'frequency_score': frequency_score,
            'diversity_score': diversity_score,
            'success_score': success_score,
            'grade': self._get_grade(total_score)
        }
    
    def _get_grade(self, score):
        """Convert score to letter grade."""
        if score >= 90:
            return 'A+'
        elif score >= 80:
            return 'A'
        elif score >= 70:
            return 'B'
        elif score >= 60:
            return 'C'
        else:
            return 'D'

print("✅ UserEngagementMetrics class created")

# ==================================================
# Generate Sample User Data
# ==================================================

def generate_sample_users(num_users=10):
    """Generate sample user data for testing."""
    users = []
    base_date = datetime.now() - timedelta(days=60)
    
    for i in range(num_users):
        signup_offset = timedelta(days=np.random.randint(0, 60))
        signup_date = base_date + signup_offset
        
        # Last active (some recent, some not)
        if np.random.random() > 0.3:  # 70% recently active
            last_active_offset = timedelta(days=np.random.randint(0, 7))
        else:  # 30% inactive
            last_active_offset = timedelta(days=np.random.randint(30, 60))
        
        last_active = datetime.now() - last_active_offset
        
        user = {
            'username': f'user_{i+1}',
            'email': f'user{i+1}@example.com',
            'created_at': signup_date.isoformat(),
            'last_login': last_active.isoformat(),
            'usage_count': np.random.randint(5, 150)
        }
        
        users.append(user)
    
    return users

sample_users = generate_sample_users(10)

print(f"\n✅ Generated {len(sample_users)} sample users")

# ==================================================
# Test User Engagement Metrics
# ==================================================

print("\n📊 Testing User Engagement Metrics:")

engagement = UserEngagementMetrics(sample_users, sample_history)

print("\n   Queries Per User:")
queries_stats = engagement.calculate_queries_per_user()
print(f"      Total Queries: {queries_stats['total_queries']}")
print(f"      Total Users: {queries_stats['total_users']}")
print(f"      Avg Per User: {queries_stats['avg_queries_per_user']:.1f}")

print("\n   Tool Diversity:")
diversity = engagement.calculate_tool_diversity()
print(f"      Unique Tools Used: {diversity['unique_tools_used']}/{diversity['total_available_tools']}")
print(f"      Diversity Rate: {diversity['diversity_rate']:.1f}%")

print("\n   Activity Frequency:")
frequency = engagement.calculate_activity_frequency()
print(f"      Active Days: {frequency['active_days']}")
print(f"      Date Range: {frequency['date_range_days']} days")
print(f"      Activity Frequency: {frequency['activity_frequency']:.1f}%")
print(f"      Avg Queries/Day: {frequency['avg_queries_per_active_day']:.1f}")

print("\n   User Segments:")
segments = engagement.analyze_user_segments()
for segment, count in segments.items():
    print(f"      {segment.replace('_', ' ').title()}: {count}")

print("\n   Engagement Score:")
score = engagement.calculate_engagement_score()
print(f"      Total Score: {score['total_score']:.1f}/100")
print(f"      Grade: {score['grade']}")
print(f"      Breakdown:")
print(f"         Frequency: {score['frequency_score']:.1f}/40")
print(f"         Diversity: {score['diversity_score']:.1f}/30")
print(f"         Success: {score['success_score']:.1f}/30")

print("\n✅ Exercise 2.1 Complete!")
print("="*80)
````

````python
# ==================================================
# EXERCISE 2.2: USAGE PATTERNS & INSIGHTS
# ==================================================

print("\n" + "="*80)
print("EXERCISE 2.2: Discovering Usage Patterns")
print("="*80)

"""
📖 THEORY: Usage Pattern Analysis

What are Usage Patterns?
==================================================

Patterns = Recurring behaviors in user data

Types:
- Temporal: When users are active
- Sequential: What they do in order
- Frequency: How often they repeat actions
- Preference: What they prefer to use

Temporal Patterns:
==================================================

Peak Times:
- Peak hours: When most active
- Peak days: Which days busiest
- Seasonality: Monthly/yearly trends

Example insights:
- "Most queries happen 2-4 PM"
- "Mondays are busiest"
- "December sees 30% more usage"

Code:
```python
def find_peak_hours(history):
    df = pd.DataFrame(history)
    df['hour'] = pd.to_datetime(df['timestamp']).dt.hour
    
    hourly_counts = df['hour'].value_counts()
    peak_hour = hourly_counts.idxmax()
    
    return peak_hour
```

Sequential Patterns:
==================================================

Tool Sequences:
- What tool do they use after X?
- Common workflows
- Feature discovery paths

Example:
- 60% who use sentiment also try summarizer
- Users who try job_matcher often return
- Fake news → sentiment (common sequence)

Transition Matrix:
```python
def calculate_transitions(history):
    df = pd.DataFrame(history)
    df = df.sort_values('timestamp')
    
    # Get tool sequences
    sequences = df['tool'].tolist()
    
    # Count transitions
    transitions = {}
    for i in range(len(sequences)-1):
        current = sequences[i]
        next_tool = sequences[i+1]
        
        if current not in transitions:
            transitions[current] = {}
        
        if next_tool not in transitions[current]:
            transitions[current][next_tool] = 0
        
        transitions[current][next_tool] += 1
    
    return transitions
```

Preference Patterns:
==================================================

Tool Preferences:
- Which tool is favorite?
- Do preferences change over time?
- Are there user segments by preference?

Time Preferences:
- Morning vs evening users
- Weekday vs weekend users
- Bursty vs steady users

Content Preferences:
- Short vs long inputs
- Simple vs complex queries
- Batch vs single queries

Behavioral Segments:
==================================================

Segment by Behavior:
1. Explorers: Try all tools
2. Specialists: Stick to one tool
3. Batch Users: Use batch features
4. Interactive Users: Single queries only
```python
def classify_user_behavior(user_history):
    df = pd.DataFrame(user_history)
    
    # Check tool diversity
    tools_used = df['tool'].nunique()
    
    # Check batch usage (if we had batch flag)
    # has_batch = df['is_batch'].any()
    
    if tools_used >= 3:
        return 'explorer'
    elif tools_used == 1:
        return 'specialist'
    else:
        return 'casual'
```

Insights Generation:
==================================================

From Patterns to Insights:
- Pattern: "90% queries on weekdays"
- Insight: "Professional use case, not personal"
- Action: "Add work-related features"

Example Insights:
- "Sentiment most popular (40% of queries)"
  → Focus on sentiment improvements
  
- "Users churn after 1 week without success"
  → Improve onboarding, success rates
  
- "Power users use all 4 tools"
  → Encourage tool discovery for others

Anomaly Detection:
==================================================

Detect Unusual Patterns:
- Sudden spike in failures
- Unusual query times
- Abnormal latencies
- Potential abuse

Example:
```python
def detect_anomalies(history):
    df = pd.DataFrame(history)
    
    # Daily query counts
    daily = df.groupby('date').size()
    
    # Calculate mean and std
    mean = daily.mean()
    std = daily.std()
    
    # Flag anomalies (>3 std from mean)
    anomalies = daily[abs(daily - mean) > 3 * std]
    
    return anomalies
```
"""

print("\n⏱️  Analyzing usage patterns...")

# ==================================================
# Usage Pattern Analyzer
# ==================================================

class UsagePatternAnalyzer:
    """Analyze usage patterns in history data."""
    
    def __init__(self, history):
        """Initialize with history data."""
        self.df = pd.DataFrame(history)
        if not self.df.empty:
            self.df['datetime'] = pd.to_datetime(self.df['timestamp'])
            self.df['date'] = self.df['datetime'].dt.date
            self.df['hour'] = self.df['datetime'].dt.hour
            self.df['day_of_week'] = self.df['datetime'].dt.day_name()
            self.df['is_weekend'] = self.df['datetime'].dt.dayofweek >= 5
    
    def find_temporal_patterns(self):
        """Find temporal usage patterns."""
        if self.df.empty:
            return {}
        
        # Peak hour
        hourly_counts = self.df['hour'].value_counts()
        peak_hour = hourly_counts.idxmax()
        
        # Peak day
        daily_counts = self.df['day_of_week'].value_counts()
        peak_day = daily_counts.idxmax()
        
        # Weekday vs weekend
        weekday_count = len(self.df[~self.df['is_weekend']])
        weekend_count = len(self.df[self.df['is_weekend']])
        
        return {
            'peak_hour': peak_hour,
            'peak_hour_count': int(hourly_counts[peak_hour]),
            'peak_day': peak_day,
            'peak_day_count': int(daily_counts[peak_day]),
            'weekday_usage': weekday_count,
            'weekend_usage': weekend_count,
            'weekday_percentage': (weekday_count / len(self.df)) * 100
        }
    
    def analyze_tool_transitions(self):
        """Analyze tool usage sequences."""
        if self.df.empty or len(self.df) < 2:
            return {}
        
        # Sort by time
        df_sorted = self.df.sort_values('datetime')
        
        # Calculate transitions
        transitions = {}
        tools = df_sorted['tool'].tolist()
        
        for i in range(len(tools) - 1):
            current = tools[i]
            next_tool = tools[i + 1]
            
            if current not in transitions:
                transitions[current] = {}
            
            if next_tool not in transitions[current]:
                transitions[current][next_tool] = 0
            
            transitions[current][next_tool] += 1
        
        # Calculate percentages
        transition_percentages = {}
        for tool, next_tools in transitions.items():
            total = sum(next_tools.values())
            transition_percentages[tool] = {
                next_t: (count / total) * 100
                for next_t, count in next_tools.items()
            }
        
        return {
            'transition_counts': transitions,
            'transition_percentages': transition_percentages
        }
    
    def identify_user_preferences(self):
        """Identify user preferences and favorites."""
        if self.df.empty:
            return {}
        
        # Favorite tool
        tool_counts = self.df['tool'].value_counts()
        favorite_tool = tool_counts.idxmax()
        
        # Time preference
        morning = len(self.df[self.df['hour'] < 12])
        afternoon = len(self.df[(self.df['hour'] >= 12) & (self.df['hour'] < 18)])
        evening = len(self.df[self.df['hour'] >= 18])
        
        preferred_time = 'morning' if morning > afternoon and morning > evening else \
                        'afternoon' if afternoon > evening else 'evening'
        
        return {
            'favorite_tool': favorite_tool,
            'favorite_tool_percentage': (tool_counts[favorite_tool] / len(self.df)) * 100,
            'preferred_time': preferred_time,
            'morning_usage': morning,
            'afternoon_usage': afternoon,
            'evening_usage': evening
        }
    
    def generate_insights(self):
        """Generate actionable insights from patterns."""
        insights = []
        
        if self.df.empty:
            return insights
        
        # Temporal patterns
        temporal = self.find_temporal_patterns()
        
        if temporal['weekday_percentage'] > 80:
            insights.append({
                'type': 'temporal',
                'insight': f"{temporal['weekday_percentage']:.0f}% of usage on weekdays",
                'action': 'Indicates professional/work use case',
                'priority': 'medium'
            })
        
        if temporal['peak_hour'] >= 9 and temporal['peak_hour'] <= 17:
            insights.append({
                'type': 'temporal',
                'insight': f"Peak usage at {temporal['peak_hour']}:00 (business hours)",
                'action': 'Schedule maintenance outside peak hours',
                'priority': 'high'
            })
        
        # Tool preferences
        prefs = self.identify_user_preferences()
        
        if prefs['favorite_tool_percentage'] > 50:
            insights.append({
                'type': 'preference',
                'insight': f"{prefs['favorite_tool']} dominates usage ({prefs['favorite_tool_percentage']:.0f}%)",
                'action': 'Consider promoting other tools or improving favorites',
                'priority': 'medium'
            })
        
        # Success rate
        success_rate = (self.df['success'].sum() / len(self.df)) * 100
        
        if success_rate < 90:
            insights.append({
                'type': 'quality',
                'insight': f"Success rate is {success_rate:.1f}% (below 90% target)",
                'action': 'Investigate failures, improve error handling',
                'priority': 'high'
            })
        
        # Activity frequency
        active_days = self.df['date'].nunique()
        date_range = (self.df['date'].max() - self.df['date'].min()).days + 1
        frequency = (active_days / date_range) * 100
        
        if frequency < 30:
            insights.append({
                'type': 'engagement',
                'insight': f"Low activity frequency ({frequency:.0f}% of days)",
                'action': 'Improve retention, send reminders, add value',
                'priority': 'high'
            })
        
        return insights
    
    def detect_anomalies(self):
        """Detect unusual usage patterns."""
        if self.df.empty:
            return []
        
        anomalies = []
        
        # Daily query count anomalies
        daily_counts = self.df.groupby('date').size()
        
        if len(daily_counts) > 3:
            mean = daily_counts.mean()
            std = daily_counts.std()
            
            # Find days with unusual activity
            unusual_days = daily_counts[abs(daily_counts - mean) > 2 * std]
            
            for date, count in unusual_days.items():
                anomalies.append({
                    'type': 'volume',
                    'date': str(date),
                    'count': int(count),
                    'expected': f"{mean:.0f}",
                    'severity': 'high' if abs(count - mean) > 3 * std else 'medium'
                })
        
        # Latency anomalies
        avg_latency = self.df['processing_time_ms'].mean()
        std_latency = self.df['processing_time_ms'].std()
        
        slow_queries = self.df[self.df['processing_time_ms'] > avg_latency + 2 * std_latency]
        
        if len(slow_queries) > 0:
            anomalies.append({
                'type': 'performance',
                'count': len(slow_queries),
                'avg_latency': f"{slow_queries['processing_time_ms'].mean():.0f}ms",
                'expected': f"{avg_latency:.0f}ms",
                'severity': 'medium'
            })
        
        return anomalies

print("✅ UsagePatternAnalyzer class created")

# ==================================================
# Test Usage Pattern Analysis
# ==================================================

print("\n📊 Testing Usage Pattern Analysis:")

pattern_analyzer = UsagePatternAnalyzer(sample_history)

print("\n   Temporal Patterns:")
temporal = pattern_analyzer.find_temporal_patterns()
print(f"      Peak Hour: {temporal['peak_hour']}:00 ({temporal['peak_hour_count']} queries)")
print(f"      Peak Day: {temporal['peak_day']} ({temporal['peak_day_count']} queries)")
print(f"      Weekday Usage: {temporal['weekday_percentage']:.1f}%")
print(f"      Weekend Usage: {100 - temporal['weekday_percentage']:.1f}%")

print("\n   Tool Transitions:")
transitions = pattern_analyzer.analyze_tool_transitions()
if transitions.get('transition_percentages'):
    print("      Common sequences:")
    for tool, next_tools in list(transitions['transition_percentages'].items())[:2]:
        top_next = max(next_tools.items(), key=lambda x: x[1])
        print(f"         {tool} → {top_next[0]}: {top_next[1]:.1f}%")

print("\n   User Preferences:")
prefs = pattern_analyzer.identify_user_preferences()
print(f"      Favorite Tool: {prefs['favorite_tool']} ({prefs['favorite_tool_percentage']:.1f}%)")
print(f"      Preferred Time: {prefs['preferred_time'].title()}")
print(f"         Morning: {prefs['morning_usage']}")
print(f"         Afternoon: {prefs['afternoon_usage']}")
print(f"         Evening: {prefs['evening_usage']}")

print("\n   Generated Insights:")
insights = pattern_analyzer.generate_insights()
for i, insight in enumerate(insights, 1):
    print(f"\n      Insight {i} ({insight['priority'].upper()} priority):")
    print(f"         Type: {insight['type']}")
    print(f"         Finding: {insight['insight']}")
    print(f"         Action: {insight['action']}")

print("\n   Anomaly Detection:")
anomalies = pattern_analyzer.detect_anomalies()
if anomalies:
    for i, anomaly in enumerate(anomalies, 1):
        print(f"\n      Anomaly {i} ({anomaly['severity'].upper()} severity):")
        print(f"         Type: {anomaly['type']}")
        if 'date' in anomaly:
            print(f"         Date: {anomaly['date']}")
            print(f"         Count: {anomaly['count']} (expected: {anomaly['expected']})")
        else:
            print(f"         Affected: {anomaly['count']} queries")
            print(f"         Avg: {anomaly['avg_latency']} (expected: {anomaly['expected']})")
else:
    print("      No anomalies detected")

print("\n✅ Exercise 2.2 Complete!")
print("="*80)
````

````python
# ==================================================
# EXERCISE 3.1: SYSTEM-WIDE METRICS
# ==================================================

print("\n" + "="*80)
print("EXERCISE 3.1: Designing Admin Dashboard & System Metrics")
print("="*80)

"""
📖 THEORY: Admin Dashboard & System Monitoring

Admin Dashboard vs User Dashboard:
==================================================

User Dashboard:
- Personal metrics
- Individual usage
- Own history
- Self-improvement

Admin Dashboard:
- System-wide metrics
- All users aggregated
- Platform health
- Business decisions

System-Wide Metrics:
==================================================

1. User Metrics:
   - Total users (registered)
   - Active users (DAU/WAU/MAU)
   - New signups (growth)
   - Churn rate

2. Usage Metrics:
   - Total queries (all time)
   - Queries today/week/month
   - Queries per tool
   - Average queries per user

3. Performance Metrics:
   - Average latency
   - Success rate
   - Error rate
   - P95/P99 latency

4. Resource Metrics:
   - API calls
   - Storage used
   - Rate limit hits
   - System uptime

User Statistics:
==================================================

Total Users:
```python
total_users = len(users)
```

Active Users:
```python
def calculate_active_users(users, days=7):
    cutoff = datetime.now() - timedelta(days=days)
    active = [
        user for user in users
        if datetime.fromisoformat(user['last_login']) >= cutoff
    ]
    return len(active)

dau = calculate_active_users(users, days=1)
wau = calculate_active_users(users, days=7)
mau = calculate_active_users(users, days=30)
```

Growth Metrics:
```python
def calculate_user_growth(users):
    # Users by signup date
    df = pd.DataFrame(users)
    df['signup_date'] = pd.to_datetime(df['created_at']).dt.date
    
    # Count by month
    monthly = df.groupby(
        pd.to_datetime(df['signup_date']).dt.to_period('M')
    ).size()
    
    return monthly
```

Usage Statistics:
==================================================

Total Queries:
```python
total_queries = len(all_history)
```

Queries by Tool:
```python
df = pd.DataFrame(all_history)
tool_counts = df['tool'].value_counts()
```

Queries Over Time:
```python
df['date'] = pd.to_datetime(df['timestamp']).dt.date
daily_queries = df.groupby('date').size()
```

Performance Monitoring:
==================================================

Average Metrics:
```python
avg_latency = df['processing_time_ms'].mean()
success_rate = (df['success'].sum() / len(df)) * 100
error_rate = 100 - success_rate
```

Percentiles:
```python
p50 = df['processing_time_ms'].quantile(0.50)
p95 = df['processing_time_ms'].quantile(0.95)
p99 = df['processing_time_ms'].quantile(0.99)
```

Tool Performance:
```python
tool_perf = df.groupby('tool').agg({
    'processing_time_ms': ['mean', 'median', 'std'],
    'success': ['sum', 'count']
})
```

Top Users:
==================================================

Most Active Users:
```python
def get_top_users(history, n=10):
    df = pd.DataFrame(history)
    # Assuming username field exists
    top = df['username'].value_counts().head(n)
    return top
```

Power Users:
```python
power_users = [
    user for user in users
    if user.get('usage_count', 0) > 100
]
```

System Health:
==================================================

Health Score (0-100):
- Performance: Latency < targets
- Reliability: Success rate > 95%
- Availability: Uptime > 99%
- Growth: User growth > 0%
```python
def calculate_health_score(metrics):
    # Performance (40 points)
    perf_score = min(
        40 * (1000 - metrics['avg_latency']) / 1000,
        40
    )
    
    # Reliability (30 points)
    rel_score = (metrics['success_rate'] / 100) * 30
    
    # Growth (30 points)
    growth_score = min(
        metrics['user_growth_rate'] * 3,
        30
    )
    
    return perf_score + rel_score + growth_score
```

Admin Actions:
==================================================

What Admins Can Do:
- Monitor system health
- Identify bottlenecks
- Track user growth
- Spot anomalies
- Plan capacity
- Debug issues

Alerts:
- Success rate < 90%
- Latency > 2s (P95)
- Error spike detected
- No new users (7 days)
- High rate limit hits
"""

print("\n⏱️  Implementing system-wide metrics...")

# ==================================================
# System Metrics Calculator
# ==================================================

class SystemMetrics:
    """Calculate system-wide metrics for admin dashboard."""
    
    def __init__(self, users, all_history):
        """
        Initialize with all users and history.
        
        Args:
            users: List of all user objects
            all_history: Combined history from all users
        """
        self.users = users
        self.history_df = pd.DataFrame(all_history) if all_history else pd.DataFrame()
        
        if not self.history_df.empty:
            self.history_df['datetime'] = pd.to_datetime(self.history_df['timestamp'])
            self.history_df['date'] = self.history_df['datetime'].dt.date
    
    def get_user_statistics(self):
        """Get user-related statistics."""
        total_users = len(self.users)
        
        # Active users (last 7 days)
        if self.users:
            cutoff_7d = datetime.now() - timedelta(days=7)
            cutoff_30d = datetime.now() - timedelta(days=30)
            
            active_7d = sum(
                1 for user in self.users
                if datetime.fromisoformat(user['last_login']) >= cutoff_7d
            )
            
            active_30d = sum(
                1 for user in self.users
                if datetime.fromisoformat(user['last_login']) >= cutoff_30d
            )
        else:
            active_7d = 0
            active_30d = 0
        
        # New users (last 30 days)
        if self.users:
            new_users = sum(
                1 for user in self.users
                if datetime.fromisoformat(user['created_at']) >= cutoff_30d
            )
        else:
            new_users = 0
        
        return {
            'total_users': total_users,
            'active_7d': active_7d,
            'active_30d': active_30d,
            'new_users_30d': new_users,
            'activation_rate': (active_7d / total_users * 100) if total_users > 0 else 0
        }
    
    def get_usage_statistics(self):
        """Get usage-related statistics."""
        if self.history_df.empty:
            return {
                'total_queries': 0,
                'queries_today': 0,
                'queries_7d': 0,
                'queries_30d': 0,
                'avg_queries_per_user': 0
            }
        
        total_queries = len(self.history_df)
        
        # Time-based queries
        now = datetime.now()
        today = now.date()
        cutoff_7d = now - timedelta(days=7)
        cutoff_30d = now - timedelta(days=30)
        
        queries_today = len(self.history_df[self.history_df['date'] == today])
        queries_7d = len(self.history_df[self.history_df['datetime'] >= cutoff_7d])
        queries_30d = len(self.history_df[self.history_df['datetime'] >= cutoff_30d])
        
        # Per user
        avg_per_user = total_queries / len(self.users) if self.users else 0
        
        return {
            'total_queries': total_queries,
            'queries_today': queries_today,
            'queries_7d': queries_7d,
            'queries_30d': queries_30d,
            'avg_queries_per_user': avg_per_user
        }
    
    def get_performance_statistics(self):
        """Get performance-related statistics."""
        if self.history_df.empty:
            return {
                'avg_latency': 0,
                'p50_latency': 0,
                'p95_latency': 0,
                'p99_latency': 0,
                'success_rate': 0,
                'error_rate': 0
            }
        
        latencies = self.history_df['processing_time_ms']
        
        avg_latency = latencies.mean()
        p50 = latencies.quantile(0.50)
        p95 = latencies.quantile(0.95)
        p99 = latencies.quantile(0.99)
        
        success_rate = (self.history_df['success'].sum() / len(self.history_df)) * 100
        error_rate = 100 - success_rate
        
        return {
            'avg_latency': avg_latency,
            'p50_latency': p50,
            'p95_latency': p95,
            'p99_latency': p99,
            'success_rate': success_rate,
            'error_rate': error_rate
        }
    
    def get_tool_statistics(self):
        """Get tool-specific statistics."""
        if self.history_df.empty:
            return {}
        
        tool_stats = {}
        
        for tool in self.history_df['tool'].unique():
            tool_data = self.history_df[self.history_df['tool'] == tool]
            
            tool_stats[tool] = {
                'total_queries': len(tool_data),
                'success_rate': (tool_data['success'].sum() / len(tool_data)) * 100,
                'avg_latency': tool_data['processing_time_ms'].mean(),
                'p95_latency': tool_data['processing_time_ms'].quantile(0.95)
            }
        
        return tool_stats
    
    def get_growth_metrics(self):
        """Get growth-related metrics."""
        if not self.users or self.history_df.empty:
            return {
                'user_growth_30d': 0,
                'query_growth_30d': 0
            }
        
        cutoff_30d = datetime.now() - timedelta(days=30)
        cutoff_60d = datetime.now() - timedelta(days=60)
        
        # User growth
        new_users_30d = sum(
            1 for user in self.users
            if datetime.fromisoformat(user['created_at']) >= cutoff_30d
        )
        
        new_users_prev_30d = sum(
            1 for user in self.users
            if cutoff_60d <= datetime.fromisoformat(user['created_at']) < cutoff_30d
        )
        
        user_growth = ((new_users_30d - new_users_prev_30d) / new_users_prev_30d * 100) \
            if new_users_prev_30d > 0 else 0
        
        # Query growth
        queries_30d = len(self.history_df[self.history_df['datetime'] >= cutoff_30d])
        queries_prev_30d = len(self.history_df[
            (self.history_df['datetime'] >= cutoff_60d) & 
            (self.history_df['datetime'] < cutoff_30d)
        ])
        
        query_growth = ((queries_30d - queries_prev_30d) / queries_prev_30d * 100) \
            if queries_prev_30d > 0 else 0
        
        return {
            'user_growth_30d': user_growth,
            'query_growth_30d': query_growth,
            'new_users_30d': new_users_30d,
            'new_users_prev_30d': new_users_prev_30d
        }
    
    def calculate_system_health(self):
        """Calculate overall system health score (0-100)."""
        perf = self.get_performance_statistics()
        usage = self.get_usage_statistics()
        growth = self.get_growth_metrics()
        
        # Performance score (40 points)
        # Target: avg latency < 500ms
        perf_score = min(40 * (1000 - perf['avg_latency']) / 1000, 40) \
            if perf['avg_latency'] > 0 else 40
        
        # Reliability score (30 points)
        # Target: success rate > 95%
        rel_score = (perf['success_rate'] / 100) * 30
        
        # Growth score (30 points)
        # Positive growth = full points, negative = 0
        growth_score = min(max(growth['query_growth_30d'] / 10, 0), 30)
        
        total_score = perf_score + rel_score + growth_score
        
        return {
            'total_score': total_score,
            'performance_score': perf_score,
            'reliability_score': rel_score,
            'growth_score': growth_score,
            'grade': self._get_health_grade(total_score)
        }
    
    def _get_health_grade(self, score):
        """Convert health score to grade."""
        if score >= 90:
            return 'Excellent'
        elif score >= 80:
            return 'Good'
        elif score >= 70:
            return 'Fair'
        elif score >= 60:
            return 'Poor'
        else:
            return 'Critical'
    
    def get_top_users(self, n=10):
        """Get top N most active users."""
        if not self.users:
            return []
        
        # Sort by usage count
        sorted_users = sorted(
            self.users,
            key=lambda x: x.get('usage_count', 0),
            reverse=True
        )
        
        top_n = sorted_users[:n]
        
        return [
            {
                'username': user['username'],
                'usage_count': user.get('usage_count', 0),
                'created_at': user['created_at'][:10]
            }
            for user in top_n
        ]

print("✅ SystemMetrics class created")

# ==================================================
# Test System Metrics
# ==================================================

print("\n📊 Testing System-Wide Metrics:")

# Combine history for all users (simulate)
system_metrics = SystemMetrics(sample_users, sample_history)

print("\n   User Statistics:")
user_stats = system_metrics.get_user_statistics()
print(f"      Total Users: {user_stats['total_users']}")
print(f"      Active (7d): {user_stats['active_7d']}")
print(f"      Active (30d): {user_stats['active_30d']}")
print(f"      New Users (30d): {user_stats['new_users_30d']}")
print(f"      Activation Rate: {user_stats['activation_rate']:.1f}%")

print("\n   Usage Statistics:")
usage_stats = system_metrics.get_usage_statistics()
print(f"      Total Queries: {usage_stats['total_queries']}")
print(f"      Queries Today: {usage_stats['queries_today']}")
print(f"      Queries (7d): {usage_stats['queries_7d']}")
print(f"      Queries (30d): {usage_stats['queries_30d']}")
print(f"      Avg Per User: {usage_stats['avg_queries_per_user']:.1f}")

print("\n   Performance Statistics:")
perf_stats = system_metrics.get_performance_statistics()
print(f"      Avg Latency: {perf_stats['avg_latency']:.0f}ms")
print(f"      P95 Latency: {perf_stats['p95_latency']:.0f}ms")
print(f"      P99 Latency: {perf_stats['p99_latency']:.0f}ms")
print(f"      Success Rate: {perf_stats['success_rate']:.1f}%")
print(f"      Error Rate: {perf_stats['error_rate']:.1f}%")

print("\n   Tool Statistics:")
tool_stats = system_metrics.get_tool_statistics()
for tool, stats in tool_stats.items():
    print(f"\n      {tool.title()}:")
    print(f"         Queries: {stats['total_queries']}")
    print(f"         Success: {stats['success_rate']:.1f}%")
    print(f"         Avg Latency: {stats['avg_latency']:.0f}ms")

print("\n   Growth Metrics:")
growth_metrics = system_metrics.get_growth_metrics()
print(f"      User Growth (30d): {growth_metrics['user_growth_30d']:.1f}%")
print(f"      Query Growth (30d): {growth_metrics['query_growth_30d']:.1f}%")

print("\n   System Health:")
health = system_metrics.calculate_system_health()
print(f"      Health Score: {health['total_score']:.1f}/100")
print(f"      Grade: {health['grade']}")
print(f"      Breakdown:")
print(f"         Performance: {health['performance_score']:.1f}/40")
print(f"         Reliability: {health['reliability_score']:.1f}/30")
print(f"         Growth: {health['growth_score']:.1f}/30")

print("\n   Top Users:")
top_users = system_metrics.get_top_users(n=5)
for i, user in enumerate(top_users, 1):
    print(f"      {i}. {user['username']}: {user['usage_count']} queries")

print("\n✅ Exercise 3.1 Complete!")
print("="*80)
````

````python
# ==================================================
# EXERCISE 3.2: ADMIN VISUALIZATIONS
# ==================================================

print("\n" + "="*80)
print("EXERCISE 3.2: Creating Admin Dashboard Visualizations")
print("="*80)

"""
📖 THEORY: Admin Dashboard Visualizations

Admin-Specific Charts:
==================================================

1. User Growth Chart:
   - Line chart showing user growth over time
   - New signups per month
   - Active users trend

2. System Health Gauge:
   - Overall health score (0-100)
   - Visual indicator (red/yellow/green)

3. Tool Performance Comparison:
   - Multi-metric comparison
   - Success rate, latency, usage

4. User Leaderboard:
   - Top users by activity
   - Table or bar chart

5. Query Volume Trend:
   - Daily/weekly query counts
   - Compare tools

6. Performance Monitoring:
   - Latency over time
   - Success rate trend
   - Error spike detection

Chart Features:
==================================================

Annotations:
- Mark important events
- Highlight anomalies
- Show targets/goals

Multiple Y-Axes:
- Compare different scales
- E.g., queries (count) + latency (ms)

Drill-Down:
- Click to filter
- Show details on hover
- Interactive legends
"""

print("\n⏱️  Creating admin dashboard visualizations...")

# ==================================================
# Admin Dashboard Charts
# ==================================================

def create_system_health_gauge(health_data):
    """Create system health gauge chart."""
    score = health_data['total_score']
    
    fig = go.Figure(go.Indicator(
        mode="gauge+number+delta",
        value=score,
        domain={'x': [0, 1], 'y': [0, 1]},
        title={'text': "System Health Score", 'font': {'size': 24}},
        delta={'reference': 80},
        gauge={
            'axis': {'range': [None, 100], 'tickwidth': 1},
            'bar': {'color': "#2196F3"},
            'steps': [
                {'range': [0, 60], 'color': "#FFCDD2"},
                {'range': [60, 80], 'color': "#FFF9C4"},
                {'range': [80, 100], 'color': "#C8E6C9"}
            ],
            'threshold': {
                'line': {'color': "green", 'width': 4},
                'thickness': 0.75,
                'value': 90
            }
        }
    ))
    
    fig.add_annotation(
        x=0.5, y=0.2,
        text=f"Grade: {health_data['grade']}",
        showarrow=False,
        font={'size': 18, 'color': 'gray'}
    )
    
    fig.update_layout(
        template="plotly_white",
        height=350
    )
    
    return fig

def create_user_growth_chart(users):
    """Create user growth over time chart."""
    df = pd.DataFrame(users)
    df['signup_date'] = pd.to_datetime(df['created_at']).dt.date
    
    # Cumulative users by date
    df = df.sort_values('signup_date')
    df['cumulative'] = range(1, len(df) + 1)
    
    # Group by month for new signups
    df['month'] = pd.to_datetime(df['signup_date']).dt.to_period('M')
    monthly_signups = df.groupby('month').size().reset_index(name='new_users')
    monthly_signups['month'] = monthly_signups['month'].astype(str)
    
    # Create figure with secondary y-axis
    fig = make_subplots(specs=[[{"secondary_y": True}]])
    
    # Cumulative users (line)
    fig.add_trace(
        go.Scatter(
            x=df['signup_date'],
            y=df['cumulative'],
            name="Total Users",
            mode='lines',
            line=dict(color='#4CAF50', width=3),
            fill='tozeroy',
            fillcolor='rgba(76, 175, 80, 0.2)'
        ),
        secondary_y=False
    )
    
    # New signups (bars)
    fig.add_trace(
        go.Bar(
            x=monthly_signups['month'],
            y=monthly_signups['new_users'],
            name="New Signups (Monthly)",
            marker_color='#2196F3',
            opacity=0.6
        ),
        secondary_y=True
    )
    
    fig.update_xaxes(title_text="Date")
    fig.update_yaxes(title_text="Total Users", secondary_y=False)
    fig.update_yaxes(title_text="New Signups", secondary_y=True)
    
    fig.update_layout(
        title="User Growth Over Time",
        template="plotly_white",
        height=400,
        hovermode='x unified'
    )
    
    return fig

def create_tool_performance_comparison(tool_stats):
    """Create comprehensive tool performance comparison."""
    tools = list(tool_stats.keys())
    
    # Prepare data
    queries = [tool_stats[t]['total_queries'] for t in tools]
    success_rates = [tool_stats[t]['success_rate'] for t in tools]
    avg_latencies = [tool_stats[t]['avg_latency'] for t in tools]
    
    # Create subplots
    fig = make_subplots(
        rows=1, cols=3,
        subplot_titles=('Query Volume', 'Success Rate', 'Avg Latency'),
        specs=[[{'type': 'bar'}, {'type': 'bar'}, {'type': 'bar'}]]
    )
    
    # Tool colors
    colors = ['#4CAF50', '#2196F3', '#FF9800', '#9C27B0']
    
    # Query volume
    fig.add_trace(
        go.Bar(x=tools, y=queries, name='Queries', marker_color=colors, showlegend=False),
        row=1, col=1
    )
    
    # Success rate
    fig.add_trace(
        go.Bar(x=tools, y=success_rates, name='Success %', marker_color=colors, showlegend=False),
        row=1, col=2
    )
    
    # Latency
    fig.add_trace(
        go.Bar(x=tools, y=avg_latencies, name='Latency (ms)', marker_color=colors, showlegend=False),
        row=1, col=3
    )
    
    # Update axes
    fig.update_yaxes(title_text="Count", row=1, col=1)
    fig.update_yaxes(title_text="Success %", row=1, col=2)
    fig.update_yaxes(title_text="Latency (ms)", row=1, col=3)
    
    # Add target line for success rate
    fig.add_hline(y=95, line_dash="dash", line_color="green", row=1, col=2, 
                  annotation_text="Target: 95%")
    
    # Add target line for latency
    fig.add_hline(y=500, line_dash="dash", line_color="orange", row=1, col=3,
                  annotation_text="Target: 500ms")
    
    fig.update_layout(
        title_text="Tool Performance Comparison",
        template="plotly_white",
        height=400,
        showlegend=False
    )
    
    return fig

def create_query_volume_trend(history):
    """Create query volume trend over time."""
    df = pd.DataFrame(history)
    df['date'] = pd.to_datetime(df['timestamp']).dt.date
    
    # Queries by tool by date
    pivot = df.groupby(['date', 'tool']).size().unstack(fill_value=0)
    
    # Tool colors
    tool_colors = {
        'sentiment': '#4CAF50',
        'summarizer': '#2196F3',
        'fake_news': '#FF9800',
        'job_matcher': '#9C27B0'
    }
    
    fig = go.Figure()
    
    for tool in pivot.columns:
        fig.add_trace(go.Scatter(
            x=pivot.index,
            y=pivot[tool],
            name=tool.title(),
            mode='lines+markers',
            line=dict(color=tool_colors.get(tool, '#999'), width=2),
            stackgroup='one'  # Stacked area
        ))
    
    fig.update_layout(
        title="Query Volume Trend (Stacked by Tool)",
        xaxis_title="Date",
        yaxis_title="Number of Queries",
        template="plotly_white",
        height=400,
        hovermode='x unified'
    )
    
    return fig

def create_user_leaderboard_chart(top_users):
    """Create user leaderboard bar chart."""
    usernames = [user['username'] for user in top_users]
    counts = [user['usage_count'] for user in top_users]
    
    # Color gradient (top users darker)
    colors = [f'rgb({int(76 + i*10)}, {int(175 - i*10)}, {int(80 - i*5)})' 
              for i in range(len(usernames))]
    
    fig = go.Figure(data=[
        go.Bar(
            x=counts,
            y=usernames,
            orientation='h',
            marker=dict(
                color=colors,
                line=dict(color='#388E3C', width=1)
            ),
            text=counts,
            textposition='auto'
        )
    ])
    
    fig.update_layout(
        title="Top 10 Most Active Users",
        xaxis_title="Total Queries",
        yaxis_title="Username",
        template="plotly_white",
        height=400,
        yaxis={'categoryorder': 'total ascending'}
    )
    
    return fig

def create_performance_monitoring_chart(history):
    """Create performance monitoring dual-axis chart."""
    df = pd.DataFrame(history)
    df['date'] = pd.to_datetime(df['timestamp']).dt.date
    
    # Daily aggregates
    daily = df.groupby('date').agg({
        'processing_time_ms': 'mean',
        'success': lambda x: (x.sum() / len(x)) * 100
    }).reset_index()
    
    daily.columns = ['date', 'avg_latency', 'success_rate']
    
    # Create figure with secondary y-axis
    fig = make_subplots(specs=[[{"secondary_y": True}]])
    
    # Latency (line)
    fig.add_trace(
        go.Scatter(
            x=daily['date'],
            y=daily['avg_latency'],
            name="Avg Latency",
            mode='lines+markers',
            line=dict(color='#FF9800', width=2),
            marker=dict(size=6)
        ),
        secondary_y=False
    )
    
    # Success rate (line)
    fig.add_trace(
        go.Scatter(
            x=daily['date'],
            y=daily['success_rate'],
            name="Success Rate",
            mode='lines+markers',
            line=dict(color='#4CAF50', width=2),
            marker=dict(size=6)
        ),
        secondary_y=True
    )
    
    # Target lines
    fig.add_hline(y=500, line_dash="dash", line_color="orange", 
                  annotation_text="Latency Target: 500ms", secondary_y=False)
    fig.add_hline(y=95, line_dash="dash", line_color="green",
                  annotation_text="Success Target: 95%", secondary_y=True)
    
    fig.update_xaxes(title_text="Date")
    fig.update_yaxes(title_text="Avg Latency (ms)", secondary_y=False)
    fig.update_yaxes(title_text="Success Rate (%)", secondary_y=True, range=[0, 105])
    
    fig.update_layout(
        title="Performance Monitoring (Latency & Success Rate)",
        template="plotly_white",
        height=400,
        hovermode='x unified'
    )
    
    return fig

print("✅ Admin dashboard charts created")

# ==================================================
# Test Admin Charts
# ==================================================

print("\n📊 Creating Admin Dashboard Charts:")

# Get metrics
system_metrics = SystemMetrics(sample_users, sample_history)
health = system_metrics.calculate_system_health()
tool_stats = system_metrics.get_tool_statistics()
top_users = system_metrics.get_top_users(n=10)

# Create charts
print("\n   Creating visualizations...")

health_gauge = create_system_health_gauge(health)
print("   ✅ System health gauge")

user_growth = create_user_growth_chart(sample_users)
print("   ✅ User growth chart")

tool_comparison = create_tool_performance_comparison(tool_stats)
print("   ✅ Tool performance comparison")

query_trend = create_query_volume_trend(sample_history)
print("   ✅ Query volume trend")

leaderboard = create_user_leaderboard_chart(top_users)
print("   ✅ User leaderboard")

perf_monitoring = create_performance_monitoring_chart(sample_history)
print("   ✅ Performance monitoring chart")

print("\n✨ Admin Dashboard Charts:")
print("   • System health gauge (0-100 with grade)")
print("   • User growth (cumulative + monthly signups)")
print("   • Tool performance comparison (3 metrics)")
print("   • Query volume trend (stacked area)")
print("   • User leaderboard (top 10)")
print("   • Performance monitoring (dual-axis)")

print("\n💡 Chart Features:")
print("   • Dual-axis for multiple metrics")
print("   • Target lines for goals")
print("   • Stacked areas for composition")
print("   • Interactive tooltips")
print("   • Professional appearance")

print("\n✅ Exercise 3.2 Complete!")
print("="*80)
````

````python
# ==================================================
# EXERCISE 3.3: ALERTS & MONITORING
# ==================================================

print("\n" + "="*80)
print("EXERCISE 3.3: System Alerts & Monitoring")
print("="*80)

"""
📖 THEORY: Alerting & Monitoring

Why Alerts?
==================================================

Purpose:
- Proactive problem detection
- Quick response to issues
- Prevent small problems becoming big
- Maintain SLA/quality

Alert Types:
==================================================

1. Threshold Alerts:
   - Metric crosses threshold
   - E.g., success rate < 90%

2. Trend Alerts:
   - Metric trending wrong direction
   - E.g., latency increasing 3 days

3. Anomaly Alerts:
   - Unusual patterns detected
   - E.g., query spike

4. Health Alerts:
   - Overall system health degraded
   - E.g., health score < 70

Alert Severity:
==================================================

Critical:
- System down
- Success rate < 80%
- Data loss

High:
- Success rate < 90%
- Latency > 2s (P95)
- Error spike

Medium:
- Success rate < 95%
- Latency > 1s (P95)
- Slow growth

Low:
- Minor issues
- Warnings
- Recommendations

Alert Rules:
==================================================

Performance:
```python
if avg_latency > 2000:
    alert("HIGH", "Latency exceeds 2s")

if p95_latency > 3000:
    alert("HIGH", "P95 latency exceeds 3s")
```

Reliability:
```python
if success_rate < 80:
    alert("CRITICAL", "Success rate below 80%")

if success_rate < 90:
    alert("HIGH", "Success rate below 90%")
```

Growth:
```python
if new_users_7d == 0:
    alert("MEDIUM", "No new users in 7 days")

if query_growth < -50:
    alert("HIGH", "Queries down 50%")
```

Health:
```python
if health_score < 60:
    alert("CRITICAL", "System health critical")

if health_score < 70:
    alert("HIGH", "System health poor")
```

Alert Actions:
==================================================

Notification:
- Email admin
- Slack message
- SMS (critical)
- Dashboard badge

Auto-remediation:
- Restart service
- Scale resources
- Enable fallback
- Rate limit users

Logging:
- Record alert
- Track resolution
- Post-mortem

Suppression:
- Avoid alert spam
- Throttle similar alerts
- Acknowledge alerts
"""

print("\n⏱️  Implementing alerting system...")

# ==================================================
# Alert Manager
# ==================================================

class AlertManager:
    """Manage system alerts and monitoring."""
    
    SEVERITY_LEVELS = ['CRITICAL', 'HIGH', 'MEDIUM', 'LOW']
    
    def __init__(self):
        """Initialize alert manager."""
        self.alerts = []
    
    def check_performance_alerts(self, perf_stats):
        """Check for performance-related alerts."""
        alerts = []
        
        # Latency alerts
        if perf_stats['avg_latency'] > 2000:
            alerts.append({
                'severity': 'HIGH',
                'category': 'performance',
                'metric': 'avg_latency',
                'value': f"{perf_stats['avg_latency']:.0f}ms",
                'threshold': '2000ms',
                'message': 'Average latency exceeds 2 seconds',
                'action': 'Investigate slow queries, optimize models'
            })
        
        if perf_stats['p95_latency'] > 3000:
            alerts.append({
                'severity': 'HIGH',
                'category': 'performance',
                'metric': 'p95_latency',
                'value': f"{perf_stats['p95_latency']:.0f}ms",
                'threshold': '3000ms',
                'message': 'P95 latency exceeds 3 seconds',
                'action': 'Check for outliers, increase timeout'
            })
        
        return alerts
    
    def check_reliability_alerts(self, perf_stats):
        """Check for reliability-related alerts."""
        alerts = []
        
        success_rate = perf_stats['success_rate']
        
        if success_rate < 80:
            alerts.append({
                'severity': 'CRITICAL',
                'category': 'reliability',
                'metric': 'success_rate',
                'value': f"{success_rate:.1f}%",
                'threshold': '80%',
                'message': 'Success rate critically low',
                'action': 'Immediate investigation required, check error logs'
            })
        elif success_rate < 90:
            alerts.append({
                'severity': 'HIGH',
                'category': 'reliability',
                'metric': 'success_rate',
                'value': f"{success_rate:.1f}%",
                'threshold': '90%',
                'message': 'Success rate below target',
                'action': 'Review recent failures, improve error handling'
            })
        elif success_rate < 95:
            alerts.append({
                'severity': 'MEDIUM',
                'category': 'reliability',
                'metric': 'success_rate',
                'value': f"{success_rate:.1f}%",
                'threshold': '95%',
                'message': 'Success rate below optimal level',
                'action': 'Monitor for trends, investigate common errors'
            })
        
        return alerts
    
    def check_growth_alerts(self, user_stats, growth_metrics):
        """Check for growth-related alerts."""
        alerts = []
        
        # New user alerts
        if user_stats['new_users_30d'] == 0:
            alerts.append({
                'severity': 'MEDIUM',
                'category': 'growth',
                'metric': 'new_users',
                'value': '0',
                'threshold': '>0',
                'message': 'No new users in 30 days',
                'action': 'Review marketing, improve onboarding'
            })
        
        # Negative growth
        if growth_metrics['query_growth_30d'] < -50:
            alerts.append({
                'severity': 'HIGH',
                'category': 'growth',
                'metric': 'query_growth',
                'value': f"{growth_metrics['query_growth_30d']:.1f}%",
                'threshold': '>-50%',
                'message': 'Queries down significantly',
                'action': 'Investigate churn, user feedback, feature issues'
            })
        
        return alerts
    
    def check_health_alerts(self, health_data):
        """Check for overall health alerts."""
        alerts = []
        
        score = health_data['total_score']
        
        if score < 60:
            alerts.append({
                'severity': 'CRITICAL',
                'category': 'health',
                'metric': 'health_score',
                'value': f"{score:.0f}/100",
                'threshold': '60/100',
                'message': f'System health critical ({health_data["grade"]})',
                'action': 'Immediate attention required across all areas'
            })
        elif score < 70:
            alerts.append({
                'severity': 'HIGH',
                'category': 'health',
                'metric': 'health_score',
                'value': f"{score:.0f}/100",
                'threshold': '70/100',
                'message': f'System health poor ({health_data["grade"]})',
                'action': 'Review performance, reliability, and growth metrics'
            })
        
        return alerts
    
    def check_tool_alerts(self, tool_stats):
        """Check for tool-specific alerts."""
        alerts = []
        
        for tool, stats in tool_stats.items():
            # Tool success rate
            if stats['success_rate'] < 85:
                alerts.append({
                    'severity': 'HIGH',
                    'category': 'tool',
                    'metric': f'{tool}_success_rate',
                    'value': f"{stats['success_rate']:.1f}%",
                    'threshold': '85%',
                    'message': f'{tool.title()} has low success rate',
                    'action': f'Debug {tool} failures, check model health'
                })
            
            # Tool latency
            if stats['avg_latency'] > 2000:
                alerts.append({
                    'severity': 'MEDIUM',
                    'category': 'tool',
                    'metric': f'{tool}_latency',
                    'value': f"{stats['avg_latency']:.0f}ms",
                    'threshold': '2000ms',
                    'message': f'{tool.title()} is slow',
                    'action': f'Optimize {tool} inference, check resources'
                })
        
        return alerts
    
    def check_all_alerts(self, system_metrics):
        """Run all alert checks."""
        all_alerts = []
        
        # Get metrics
        perf_stats = system_metrics.get_performance_statistics()
        user_stats = system_metrics.get_user_statistics()
        growth_metrics = system_metrics.get_growth_metrics()
        health_data = system_metrics.calculate_system_health()
        tool_stats = system_metrics.get_tool_statistics()
        
        # Check all categories
        all_alerts.extend(self.check_performance_alerts(perf_stats))
        all_alerts.extend(self.check_reliability_alerts(perf_stats))
        all_alerts.extend(self.check_growth_alerts(user_stats, growth_metrics))
        all_alerts.extend(self.check_health_alerts(health_data))
        all_alerts.extend(self.check_tool_alerts(tool_stats))
        
        # Sort by severity
        severity_order = {level: i for i, level in enumerate(self.SEVERITY_LEVELS)}
        all_alerts.sort(key=lambda x: severity_order[x['severity']])
        
        self.alerts = all_alerts
        return all_alerts
    
    def get_alerts_by_severity(self, severity):
        """Get alerts of specific severity."""
        return [a for a in self.alerts if a['severity'] == severity]
    
    def get_alert_summary(self):
        """Get summary of alerts by severity."""
        summary = {level: 0 for level in self.SEVERITY_LEVELS}
        
        for alert in self.alerts:
            summary[alert['severity']] += 1
        
        return summary

print("✅ AlertManager class created")

# ==================================================
# Test Alert System
# ==================================================

print("\n🚨 Testing Alert System:")

alert_manager = AlertManager()
system_metrics = SystemMetrics(sample_users, sample_history)

# Check all alerts
all_alerts = alert_manager.check_all_alerts(system_metrics)

print(f"\n   Total Alerts: {len(all_alerts)}")

# Summary by severity
summary = alert_manager.get_alert_summary()
for severity, count in summary.items():
    if count > 0:
        emoji = '🔴' if severity == 'CRITICAL' else '🟠' if severity == 'HIGH' else '🟡' if severity == 'MEDIUM' else '🔵'
        print(f"   {emoji} {severity}: {count}")

# Display alerts
if all_alerts:
    print("\n   Active Alerts:")
    for i, alert in enumerate(all_alerts[:5], 1):  # Show first 5
        print(f"\n   {i}. [{alert['severity']}] {alert['category'].upper()}")
        print(f"      Metric: {alert['metric']}")
        print(f"      Value: {alert['value']} (threshold: {alert['threshold']})")
        print(f"      Message: {alert['message']}")
        print(f"      Action: {alert['action']}")
else:
    print("\n   ✅ No alerts - system healthy!")

print("\n✅ Exercise 3.3 Complete!")
print("="*80)
````
