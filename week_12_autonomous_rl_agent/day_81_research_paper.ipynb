```python
# ==================================================
# EXERCISE 1.4: RELATED WORK & BACKGROUND
# ==================================================

print("\n" + "=" * 80)
print("EXERCISE 1.4: Related Work & Background")
print("=" * 80)

"""
📖 THEORY: Related Work Section

Should cover:
1. Historical context
2. Algorithm evolution
3. Related comparisons
4. Theoretical foundations

Target: 2 pages
"""

print("\n⏱️ Creating related work section...")

related_work = """
2. RELATED WORK & BACKGROUND
━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━

2.1 Policy Gradient Methods
━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━

Policy gradient methods directly optimize the policy parameters θ by
performing gradient ascent on the expected return:

    ∇_θ J(θ) = E_τ[∑_t ∇_θ log π_θ(a_t|s_t) R_t]

where τ represents trajectories, π_θ is the policy, and R_t is the return
from time t. This approach offers several advantages over value-based methods:

1. Direct policy optimization without requiring a value function
2. Natural handling of continuous action spaces
3. Guaranteed convergence to local optima under appropriate conditions
4. Ability to learn stochastic policies

However, vanilla policy gradients suffer from high variance, requiring
sophisticated variance reduction techniques for practical application.

2.2 REINFORCE Algorithm
━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━

Williams (1992) introduced REINFORCE as the foundational Monte Carlo policy
gradient algorithm. REINFORCE uses the complete trajectory return as an
unbiased but high-variance estimate of the advantage:

    ∇_θ J(θ) = E_τ[∑_t ∇_θ log π_θ(a_t|s_t) G_t]

where G_t = ∑_{k=t}^T γ^{k-t} r_k is the Monte Carlo return.

Key characteristics:
- Unbiased gradient estimates
- High variance (requires many samples)
- Simple implementation
- Inefficient sample usage (no experience replay)

Despite its limitations, REINFORCE established the foundation for modern
policy gradient methods and remains pedagogically important for understanding
the field.

2.3 Actor-Critic Methods
━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━

Actor-critic methods reduce variance by replacing Monte Carlo returns with
learned value function estimates. The actor (policy) and critic (value
function) are trained simultaneously:

    Actor update:  ∇_θ J(θ) = E[∑_t ∇_θ log π_θ(a_t|s_t) A_t]
    Critic update: minimize E[(V_φ(s_t) - G_t)²]

where A_t = G_t - V_φ(s_t) is the advantage function.

Advantage Actor-Critic (A2C) is a synchronous variant introduced by OpenAI
(2017) as a simplified version of A3C [6]. A2C performs synchronous updates
across multiple parallel workers, offering:

- Lower variance than REINFORCE
- Better sample efficiency through bootstrapping
- Stable training with multiple workers
- Simpler implementation than asynchronous variants

However, A2C still suffers from data inefficiency due to single-epoch updates.

2.4 Proximal Policy Optimization (PPO)
━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━

Schulman et al. (2017) introduced PPO as a more practical alternative to
Trust Region Policy Optimization (TRPO) [7]. PPO constrains policy updates
using a clipped surrogate objective:

    L^{CLIP}(θ) = E[min(r_t(θ)Â_t, clip(r_t(θ), 1-ε, 1+ε)Â_t)]

where r_t(θ) = π_θ(a_t|s_t) / π_θ_old(a_t|s_t) is the probability ratio and
ε is the clipping parameter.

Key innovations:
- Clipped objective prevents destructively large policy updates
- Multiple epochs of minibatch updates improve sample efficiency
- Simpler implementation than TRPO (no second-order optimization)
- Robust performance across diverse tasks

PPO has become the default algorithm for many RL applications due to its
excellent balance of performance, simplicity, and reliability.

2.5 Generalized Advantage Estimation (GAE)
━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━

Schulman et al. (2016) introduced GAE [8] to balance bias and variance in
advantage estimation:

    Â_t^{GAE(γ,λ)} = ∑_{l=0}^∞ (γλ)^l δ_{t+l}

where δ_t = r_t + γV(s_{t+1}) - V(s_t) is the TD error.

The parameter λ controls the bias-variance tradeoff:
- λ = 0: Low variance, high bias (pure TD)
- λ = 1: High variance, low bias (Monte Carlo)
- λ = 0.95: Empirically optimal for many tasks

I use GAE with λ=0.95 across all algorithms to ensure fair comparison.

2.6 Previous Comparative Studies
━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━

Several prior works have compared policy gradient algorithms:

Henderson et al. (2018) [9] highlighted reproducibility issues in deep RL,
demonstrating that implementation details and random seeds significantly
impact reported performance. This motivated my rigorous statistical testing.

Engstrom et al. (2020) [10] showed that "implementation matters" in deep RL,
with seemingly minor details causing large performance differences. I address
this by implementing all algorithms from scratch with identical components.

Ilyas et al. (2018) [11] compared policy gradient algorithms on continuous
control tasks, finding PPO to be most robust. My work extends this to
discrete action spaces with comprehensive hyperparameter analysis.

My work differs from these studies by:
1. Implementing all algorithms from scratch for controlled comparison
2. Conducting rigorous statistical testing (100+ episodes per environment)
3. Validating hyperparameter transfer across environments
4. Providing detailed failure case analysis
5. Offering practical guidance for algorithm selection

2.7 Theoretical Foundations
━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━

All three algorithms I study are grounded in the policy gradient theorem:

    ∇_θ J(θ) = E_τ[∑_t ∇_θ log π_θ(a_t|s_t) Q^π(s_t, a_t)]

The algorithms differ primarily in:
1. How they estimate Q^π (Monte Carlo vs bootstrapping)
2. How they reduce variance (baseline subtraction, advantage estimation)
3. How they ensure stable updates (trust regions, clipping)

REINFORCE uses Monte Carlo returns (unbiased, high variance).
A2C uses bootstrapped returns with baseline (lower variance, small bias).
PPO adds trust region constraints (controlled updates, stable training).

This progression represents the evolution of policy gradient methods toward
more sample-efficient and stable algorithms.

━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━
Word Count: ~700 words

References (Related Work):
[6] Mnih et al., "Asynchronous Methods for Deep Reinforcement Learning,"
    ICML, 2016
[7] Schulman et al., "Trust Region Policy Optimization," ICML, 2015
[8] Schulman et al., "High-Dimensional Continuous Control Using Generalized
    Advantage Estimation," ICLR, 2016
[9] Henderson et al., "Deep Reinforcement Learning that Matters,"
    AAAI, 2018
[10] Engstrom et al., "Implementation Matters in Deep RL: A Case Study on
     PPO and TRPO," ICLR, 2020
[11] Ilyas et al., "A Closer Look at Deep Policy Gradients," ICLR, 2018
"""

print(related_work)

# Save related work
with open('results/day81/related_work.txt', 'w') as f:
    f.write(related_work)

print("\n✅ Related work saved: results/day81/related_work.txt")

print("\n✅ Exercise 1.4 Complete!")
print("=" * 80)
```
