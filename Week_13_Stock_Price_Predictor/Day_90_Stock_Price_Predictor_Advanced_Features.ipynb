````python
# ==================================================
# EXERCISE 1.1: ENSEMBLE METHODS THEORY
# ==================================================

print("\n" + "=" * 80)
print("EXERCISE 1.1: Model Ensemble Concepts")
print("=" * 80)

print("""
📖 THEORY: Ensemble Methods for Stock Prediction

What is Ensemble Learning?
Combining multiple models to make better predictions than any single model.

Why Ensemble Works:
✓ Reduces overfitting (no single model dominates)
✓ Captures different patterns (diversity)
✓ More robust to market changes
✓ Better confidence estimation

Types of Ensembles:

1. Averaging Ensemble
   - Simple average: (Model1 + Model2 + Model3) / 3
   - Weighted average: (w1×M1 + w2×M2 + w3×M3)
   - Best when: Models have similar accuracy

2. Voting Ensemble
   - Each model votes (UP, DOWN, HOLD)
   - Majority wins
   - Best when: Binary predictions

3. Stacking Ensemble
   - Train meta-model on predictions
   - Learns how to combine models
   - Best when: Complex relationships

4. Boosting Ensemble
   - Train models sequentially
   - Each model corrects previous errors
   - Best when: Lots of training data

For Stock Prediction:
We'll use WEIGHTED AVERAGING because:
✓ Simple to implement
✓ Easy to interpret
✓ Works well with LSTM predictions
✓ Can assign weights based on validation performance

Our Approach:
1. Train 3 LSTM models with different configs
2. Assign weights based on validation MAPE
3. Weighted average of predictions
4. Compare vs single best model

Example:
Model A: MAPE 3.2% → Weight 0.40
Model B: MAPE 4.1% → Weight 0.35
Model C: MAPE 5.0% → Weight 0.25

Final Prediction = 0.40×A + 0.35×B + 0.25×C
""")

print("\n💡 Benefits of Ensemble:")
print("   • Lower variance (more stable)")
print("   • Better generalization")
print("   • Confidence intervals (std of predictions)")
print("   • Robust to individual model failures")

print("\n✅ Exercise 1.1 Complete!")
print("=" * 80)
````

````python
# ==================================================
# EXERCISE 1.2: ENSEMBLE PREDICTION IMPLEMENTATION
# ==================================================

print("\n" + "=" * 80)
print("EXERCISE 1.2: Building Ensemble Predictor")
print("=" * 80)

print("\n⏱️ Creating EnsemblePredictor class...")

class EnsemblePredictor:
    """
    Simple ensemble predictor for stock prices.
    Combines multiple LSTM models with weighted averaging.
    """

    def __init__(self, configs):
        """
        Args:
            configs: List of model configurations
                    Each config = {'units': [128,64,32], 'dropout': 0.2, 'lr': 0.001}
        """
        self.configs = configs
        self.models = []
        self.weights = []

    def train_models(self, X_train, y_train, X_val, y_val, epochs=50):
        """Train all models in the ensemble"""
        print(f"\n⏱️ Training {len(self.configs)} models...")

        val_scores = []

        for i, config in enumerate(self.configs):
            print(f"\n{'─'*60}")
            print(f"Training Model {i+1}/{len(self.configs)}")
            print(f"Config: {config}")
            print(f"{'─'*60}")

            # Build model
            model = Sequential([
                Input(shape=(X_train.shape[1], X_train.shape[2])),
                LSTM(config['units'][0], return_sequences=True),
                Dropout(config['dropout']),
                LSTM(config['units'][1], return_sequences=True),
                Dropout(config['dropout']),
                LSTM(config['units'][2], return_sequences=False),
                Dropout(config['dropout']),
                Dense(16, activation='relu'),
                Dense(1, activation='linear')
            ])

            model.compile(
                optimizer=tf.keras.optimizers.Adam(learning_rate=config['lr']),
                loss='mse',
                metrics=['mae']
            )

            # Train
            history = model.fit(
                X_train, y_train,
                validation_data=(X_val, y_val),
                epochs=epochs,
                batch_size=32,
                callbacks=[
                    tf.keras.callbacks.EarlyStopping(
                        monitor='val_loss',
                        patience=5,
                        restore_best_weights=True
                    )
                ],
                verbose=0
            )

            # Evaluate
            val_pred = model.predict(X_val, verbose=0)
            val_mape = np.mean(np.abs((y_val - val_pred.flatten()) / y_val)) * 100

            print(f"   Epochs trained: {len(history.history['loss'])}")
            print(f"   Val MAPE: {val_mape:.3f}%")
            print(f"   ✅ Model {i+1} complete")

            # Store
            self.models.append(model)
            val_scores.append(val_mape)

        # Calculate weights (inverse of MAPE)
        # Lower MAPE = Higher weight
        inv_scores = [1/score for score in val_scores]
        total = sum(inv_scores)
        self.weights = [inv/total for inv in inv_scores]

        print(f"\n{'='*60}")
        print("📊 Ensemble Weights (based on validation MAPE):")
        for i, (score, weight) in enumerate(zip(val_scores, self.weights)):
            print(f"   Model {i+1}: MAPE={score:.3f}% → Weight={weight:.3f}")
        print(f"{'='*60}")

    def predict(self, X):
        """Make ensemble prediction"""
        predictions = []

        for model in self.models:
            pred = model.predict(X, verbose=0)
            predictions.append(pred.flatten())

        # Weighted average
        ensemble_pred = np.zeros(len(predictions[0]))
        for pred, weight in zip(predictions, self.weights):
            ensemble_pred += pred * weight

        return ensemble_pred

    def predict_with_confidence(self, X):
        """
        Make ensemble prediction with confidence interval.

        Returns:
            mean_pred: Weighted average prediction
            std_pred: Standard deviation (uncertainty)
            individual_preds: Predictions from each model
        """
        predictions = []

        for model in self.models:
            pred = model.predict(X, verbose=0)
            predictions.append(pred.flatten())

        # Stack predictions
        pred_array = np.array(predictions)

        # Weighted average
        mean_pred = np.zeros(pred_array.shape[1])
        for pred, weight in zip(predictions, self.weights):
            mean_pred += pred * weight

        # Standard deviation (confidence)
        std_pred = np.std(pred_array, axis=0)

        return mean_pred, std_pred, predictions

print("\n✅ EnsemblePredictor class created!")

print("\n📊 Key Features:")
print("   • Trains multiple LSTM models")
print("   • Calculates optimal weights (inverse MAPE)")
print("   • Makes weighted ensemble predictions")
print("   • Provides confidence intervals")

print("\n💡 Usage Example:")
print("""
configs = [
    {'units': [128, 64, 32], 'dropout': 0.2, 'lr': 0.001},
    {'units': [256, 128, 64], 'dropout': 0.15, 'lr': 0.001},
    {'units': [64, 32, 16], 'dropout': 0.25, 'lr': 0.0005}
]

ensemble = EnsemblePredictor(configs)
ensemble.train_models(X_train, y_train, X_val, y_val)

# Predict
mean, std, individual = ensemble.predict_with_confidence(X_test)
print(f"Prediction: {mean[0]:.2f} ± {std[0]:.2f}")
""")

print("\n✅ Exercise 1.2 Complete!")
print("=" * 80)
````

````python
# ==================================================
# EXERCISE 2.3: ALERT SYSTEM (DEMO)
# ==================================================

print("\n" + "=" * 80)
print("EXERCISE 2.3: Alert System Implementation")
print("=" * 80)

print("""
📖 THEORY: Trading Alert System

Types of Alerts:

1. Price Alerts
   - Target price reached
   - Stop loss triggered
   - Support/resistance broken

2. Prediction Alerts
   - High confidence buy signal
   - High confidence sell signal
   - Model confidence drop

3. Volatility Alerts
   - Unusual price movement (>5% in day)
   - Volume spike (>2x average)
   - Gap up/down (>3% overnight)

4. Portfolio Alerts
   - Portfolio value threshold
   - Max drawdown warning
   - Rebalancing needed

Alert Delivery:
- In-app notifications (Streamlit)
- Email (via SMTP)
- SMS (via Twilio)
- Webhook (Slack, Discord)
""")

class AlertSystem:
    """Simple alert system for trading signals"""

    def __init__(self):
        self.alerts = []
        self.alert_history = []

    def check_price_alert(self, current_price, target_price, direction='above'):
        """Check if price target is reached"""
        if direction == 'above' and current_price >= target_price:
            alert = {
                'type': 'PRICE',
                'message': f"Price ${current_price:.2f} reached target ${target_price:.2f}",
                'level': 'HIGH',
                'timestamp': datetime.now()
            }
            self.alerts.append(alert)
            return True
        elif direction == 'below' and current_price <= target_price:
            alert = {
                'type': 'PRICE',
                'message': f"Price ${current_price:.2f} dropped below ${target_price:.2f}",
                'level': 'HIGH',
                'timestamp': datetime.now()
            }
            self.alerts.append(alert)
            return True
        return False

    def check_prediction_alert(self, confidence, signal, threshold=70):
        """Check if prediction confidence is high"""
        if confidence >= threshold:
            alert = {
                'type': 'PREDICTION',
                'message': f"{signal} signal with {confidence:.0f}% confidence",
                'level': 'MEDIUM',
                'timestamp': datetime.now()
            }
            self.alerts.append(alert)
            return True
        return False

    def check_volatility_alert(self, price_change_pct, threshold=5.0):
        """Check for unusual price movement"""
        if abs(price_change_pct) >= threshold:
            alert = {
                'type': 'VOLATILITY',
                'message': f"Large price move: {price_change_pct:+.2f}%",
                'level': 'HIGH' if abs(price_change_pct) >= 10 else 'MEDIUM',
                'timestamp': datetime.now()
            }
            self.alerts.append(alert)
            return True
        return False

    def get_active_alerts(self):
        """Get all active alerts"""
        return self.alerts

    def clear_alerts(self):
        """Clear all alerts"""
        self.alert_history.extend(self.alerts)
        self.alerts = []

    def send_notification(self, alert, method='display'):
        """
        Send alert notification (demo mode).
        In production, would use email/SMS/webhook.
        """
        if method == 'display':
            print(f"\n🔔 ALERT: {alert['message']}")
            print(f"   Level: {alert['level']}")
            print(f"   Time: {alert['timestamp'].strftime('%Y-%m-%d %H:%M:%S')}")
        elif method == 'email':
            print(f"   [DEMO] Would send email: {alert['message']}")
        elif method == 'sms':
            print(f"   [DEMO] Would send SMS: {alert['message']}")

print("\n✅ AlertSystem class created!")

print("\n💡 Example Usage:")
print("""
alert_system = AlertSystem()

# Check price alert
alert_system.check_price_alert(
    current_price=155.50,
    target_price=155.00,
    direction='above'
)

# Check prediction alert
alert_system.check_prediction_alert(
    confidence=85,
    signal='BUY',
    threshold=70
)

# Check volatility alert
alert_system.check_volatility_alert(
    price_change_pct=7.2,
    threshold=5.0
)

# Display alerts
for alert in alert_system.get_active_alerts():
    alert_system.send_notification(alert)
""")

print("\n✅ Exercise 2.3 Complete!")
print("=" * 80)
````

````python
# ==================================================
# EXERCISE 2.4: MULTI-TIMEFRAME ANALYSIS
# ==================================================

print("\n" + "=" * 80)
print("EXERCISE 2.4: Multi-Timeframe Analysis Feature")
print("=" * 80)

print("""
📖 THEORY: Multi-Timeframe Trading

Professional traders analyze multiple timeframes:

Primary Timeframes:
- Daily (1D): Short-term trading, day trading
- Weekly (1W): Swing trading, position entry/exit
- Monthly (1M): Long-term trends, portfolio allocation

Timeframe Alignment:
All timeframes bullish → Strong buy signal
All timeframes bearish → Strong sell signal
Mixed signals → Wait for alignment

Example Analysis:
Stock: NVDA

Daily (1D):
  Trend: Bullish ↗
  RSI: 65 (neutral)
  Signal: HOLD

Weekly (1W):
  Trend: Bullish ↗
  RSI: 72 (overbought)
  Signal: CAUTION

Monthly (1M):
  Trend: Bullish ↗
  RSI: 55 (neutral)
  Signal: BUY

Overall: 2/3 bullish → Moderate BUY
""")

def analyze_timeframe(df, timeframe='1D'):
    """
    Analyze a single timeframe.

    Args:
        df: Price data
        timeframe: '1D', '1W', or '1M'

    Returns:
        Dict with analysis results
    """
    # Resample to timeframe
    if timeframe == '1W':
        df_resampled = df.resample('W').agg({
            'Open': 'first',
            'High': 'max',
            'Low': 'min',
            'Close': 'last',
            'Volume': 'sum'
        })
    elif timeframe == '1M':
        df_resampled = df.resample('ME').agg({
            'Open': 'first',
            'High': 'max',
            'Low': 'min',
            'Close': 'last',
            'Volume': 'sum'
        })
    else:
        df_resampled = df

    # Calculate trend (simple moving average slope)
    ma20 = df_resampled['Close'].rolling(20).mean()
    trend = "Bullish ↗" if ma20.iloc[-1] > ma20.iloc[-5] else "Bearish ↘"

    # Calculate RSI
    delta = df_resampled['Close'].diff()
    gain = (delta.where(delta > 0, 0)).rolling(14).mean()
    loss = (-delta.where(delta < 0, 0)).rolling(14).mean()
    rs = gain / loss
    rsi = 100 - (100 / (1 + rs))
    rsi_current = rsi.iloc[-1]

    # Determine signal
    if rsi_current > 70:
        signal = "SELL (Overbought)"
    elif rsi_current < 30:
        signal = "BUY (Oversold)"
    else:
        signal = "HOLD (Neutral)"

    return {
        'timeframe': timeframe,
        'trend': trend,
        'rsi': rsi_current,
        'signal': signal,
        'last_price': df_resampled['Close'].iloc[-1]
    }

def multi_timeframe_analysis(df):
    """Analyze all timeframes and provide overall signal"""
    timeframes = ['1D', '1W', '1M']
    results = []

    for tf in timeframes:
        analysis = analyze_timeframe(df, tf)
        results.append(analysis)

    # Count bullish signals
    bullish_count = sum(1 for r in results if 'BUY' in r['signal'] or 'HOLD' in r['signal'])

    # Overall signal
    if bullish_count == 3:
        overall = "🟢 STRONG BUY - All timeframes aligned"
    elif bullish_count == 2:
        overall = "🟢 MODERATE BUY - Majority bullish"
    elif bullish_count == 1:
        overall = "🟡 MIXED - Wait for alignment"
    else:
        overall = "🔴 SELL - Bearish across timeframes"

    return results, overall

print("\n✅ Multi-timeframe analysis functions created!")

print("\n💡 Dashboard Integration:")
print("""
# Add to Streamlit dashboard

st.subheader("📊 Multi-Timeframe Analysis")

# Load data
df = load_stock_data(ticker, days=365*2)

# Analyze
results, overall = multi_timeframe_analysis(df)

# Display overall signal
st.markdown(f"### Overall Signal: {overall}")

# Display each timeframe
cols = st.columns(3)
for i, result in enumerate(results):
    with cols[i]:
        st.metric(
            label=f"{result['timeframe']} Trend",
            value=result['trend']
        )
        st.write(f"RSI: {result['rsi']:.1f}")
        st.write(f"Signal: {result['signal']}")
""")

print("\n✅ Exercise 2.4 Complete!")
print("=" * 80)
````

````python
# ==================================================
# EXERCISE 3.2: CACHING STRATEGIES
# ==================================================

print("\n" + "=" * 80)
print("EXERCISE 3.2: Implementing Caching")
print("=" * 80)

print("""
📖 THEORY: Caching in Streamlit

Streamlit Caching Decorators:

1. @st.cache_data (for data)
   • Use for: DataFrames, arrays, API responses
   • TTL: Set expiration time
   • Clear: Automatically on app restart

2. @st.cache_resource (for models)
   • Use for: ML models, database connections
   • Persists: Across multiple runs
   • Shared: Between all users

Cache Key Considerations:
- Function name + parameters = cache key
- Changing parameters = new cache entry
- Be careful with mutable objects
""")

print("\n⏱️ Creating cached functions...")

# Simulated cached data loader
cached_data_code = '''
import streamlit as st
import yfinance as yf
import pandas as pd

@st.cache_data(ttl=300)  # Cache for 5 minutes
def load_stock_data_cached(symbol, days):
    """
    Load stock data with caching.
    Same inputs = return cached data (no re-download)
    """
    end_date = datetime.now()
    start_date = end_date - timedelta(days=days)

    df = yf.download(symbol, start=start_date, end=end_date, progress=False)

    # Flatten multi-index if exists
    if isinstance(df.columns, pd.MultiIndex):
        df.columns = df.columns.get_level_values(0)

    return df

@st.cache_resource
def load_model_cached(model_path):
    """
    Load trained model with caching.
    Model loaded once, shared across all users.
    """
    from tensorflow.keras.models import load_model
    model = load_model(model_path)
    return model

@st.cache_resource
def load_scaler_cached(scaler_path):
    """
    Load scaler with caching.
    Loaded once, shared across all users.
    """
    import pickle
    with open(scaler_path, 'rb') as f:
        scaler = pickle.load(f)
    return scaler

# Usage in dashboard:
df = load_stock_data_cached('NVDA', 365)  # First call: downloads
df = load_stock_data_cached('NVDA', 365)  # Second call: cached! (instant)

model = load_model_cached('models/nvda_lstm.h5')  # Loads once
scaler = load_scaler_cached('models/nvda_scaler.pkl')  # Loads once
'''

print(cached_data_code)

print("\n📊 Caching Impact:")
print("""
Before Caching:
   Every page load → Download data → 2-3 seconds
   Every prediction → Load model → 500ms
   10 users × 3 seconds = 30 seconds load

After Caching:
   First page load → Download data → 2-3 seconds
   Next 9 loads → Return cached → <100ms
   10 users × 0.3 seconds = 3 seconds total

Performance Improvement: 10x faster! ⚡
""")

print("\n💡 Cache Management Tips:")
print("   • Set appropriate TTL (5 min for data)")
print("   • Clear cache with st.cache_data.clear()")
print("   • Monitor cache size (memory usage)")
print("   • Use cache key parameters wisely")

print("\n✅ Exercise 3.2 Complete!")
print("=" * 80)
````

````python
# ==================================================
# EXERCISE 3.3: BATCH PREDICTION OPTIMIZATION
# ==================================================

print("\n" + "=" * 80)
print("EXERCISE 3.3: Batch Prediction")
print("=" * 80)

print("""
📖 THEORY: Batch Processing for Speed

Single Prediction (Slow):
for i in range(100):
    pred = model.predict(X[i])  # 100 model calls

Time: 100 × 10ms = 1000ms (1 second)

Batch Prediction (Fast):
preds = model.predict(X[:100])  # 1 model call

Time: 1 × 50ms = 50ms

Speedup: 20x faster! ⚡

Why Batching Works:
- GPU parallelization
- Reduced overhead (loading model once)
- Vectorized operations
- Efficient memory usage
""")

def predict_single(model, X_sample):
    """Single prediction (slow)"""
    return model.predict(X_sample.reshape(1, -1), verbose=0)[0]

def predict_batch(model, X_samples):
    """Batch prediction (fast)"""
    return model.predict(X_samples, verbose=0).flatten()

print("\n✅ Batch prediction functions created!")

print("\n💡 Performance Comparison:")
comparison_code = '''
import time
import numpy as np

# Simulate 100 predictions
X_test = np.random.randn(100, 60, 35)  # 100 samples

# Method 1: Single predictions (slow)
start = time.time()
preds_single = []
for i in range(len(X_test)):
    pred = predict_single(model, X_test[i])
    preds_single.append(pred)
time_single = time.time() - start

# Method 2: Batch prediction (fast)
start = time.time()
preds_batch = predict_batch(model, X_test)
time_batch = time.time() - start

print(f"Single predictions: {time_single:.3f}s")
print(f"Batch predictions:  {time_batch:.3f}s")
print(f"Speedup: {time_single/time_batch:.1f}x faster!")
'''

print(comparison_code)

print("\n🎯 Best Practices:")
print("   • Always predict in batches when possible")
print("   • Optimal batch size: 32-128 samples")
print("   • Pre-allocate arrays (avoid appending)")
print("   • Use model.predict() over model()")

print("\n✅ Exercise 3.3 Complete!")
print("=" * 80)
````

````python
# ==================================================
# EXERCISE 3.4: MEMORY OPTIMIZATION
# ==================================================

print("\n" + "=" * 80)
print("EXERCISE 3.4: Memory Optimization")
print("=" * 80)

print("""
📖 THEORY: Reducing Memory Usage

Memory Issues in ML Applications:

1. Large DataFrames
   ⚠️ Problem: Loading 5 years × 5 stocks = huge memory
   ✅ Solution: Load only required data, use efficient dtypes

2. Multiple Models
   ⚠️ Problem: 5 models × 100MB each = 500MB
   ✅ Solution: Load on-demand, share base layers

3. Cached Data
   ⚠️ Problem: Cache grows indefinitely
   ✅ Solution: Set TTL, limit cache size

4. Feature Arrays
   ⚠️ Problem: Storing all sequences in memory
   ✅ Solution: Generator functions, lazy loading

Memory Optimization Techniques:

🎯 Data Type Optimization:
float64 → float32 (50% memory reduction)
int64 → int32 or int16 (50-75% reduction)

🎯 Selective Loading:
Load only required columns
Use date filters
Sample large datasets

🎯 Generator Functions:
Instead of: all_data = load_everything()
Use: for batch in data_generator(): ...

🎯 Garbage Collection:
Explicitly delete unused variables
Force garbage collection
Clear matplotlib figures
""")

print("\n⏱️ Creating memory-efficient functions...")

def optimize_dataframe_memory(df):
    """
    Reduce DataFrame memory usage.
    Converts to efficient dtypes.
    """
    start_mem = df.memory_usage().sum() / 1024**2

    for col in df.columns:
        col_type = df[col].dtype

        if col_type != object:
            c_min = df[col].min()
            c_max = df[col].max()

            if str(col_type)[:3] == 'int':
                if c_min > np.iinfo(np.int16).min and c_max < np.iinfo(np.int16).max:
                    df[col] = df[col].astype(np.int16)
                elif c_min > np.iinfo(np.int32).min and c_max < np.iinfo(np.int32).max:
                    df[col] = df[col].astype(np.int32)
            else:
                if c_min > np.finfo(np.float32).min and c_max < np.finfo(np.float32).max:
                    df[col] = df[col].astype(np.float32)

    end_mem = df.memory_usage().sum() / 1024**2
    reduction = (start_mem - end_mem) / start_mem * 100

    print(f"   Memory usage decreased: {start_mem:.2f}MB → {end_mem:.2f}MB ({reduction:.1f}% reduction)")

    return df

print("\n✅ Memory optimization function created!")

print("\n📊 Example Impact:")
print("""
Stock Data (5 stocks × 5 years):
Before: 50 MB (float64)
After:  25 MB (float32)
Savings: 50%

Model Weights:
Before: 200 MB (full precision)
After:  50 MB (quantized to int8)
Savings: 75%

Total Memory:
Before: 1.2 GB
After:  400 MB
Savings: 67%
""")

print("\n💡 Monitoring Memory:")
monitoring_code = '''
import psutil
import os

# Check current memory usage
process = psutil.Process(os.getpid())
mem_mb = process.memory_info().rss / 1024 / 1024
print(f"Current memory: {mem_mb:.0f} MB")

# In Streamlit dashboard
st.sidebar.metric("Memory Usage", f"{mem_mb:.0f} MB")
'''

print(monitoring_code)

print("\n✅ Exercise 3.4 Complete!")
print("=" * 80)
````
