```python
# ==================================================
# EXERCISE 1.1: TIME SERIES CONCEPTS
# ==================================================

print("\n" + "=" * 80)
print("EXERCISE 1.1: Understanding Time Series Data")
print("=" * 80)

"""
📖 THEORY: What is Time Series Data?

A time series is a sequence of data points collected at successive time intervals.
Unlike regular supervised learning, time series data has temporal ordering -
the ORDER of data points matters!

Key Characteristics:
1. TREND: Long-term increase or decrease in data
   - Example: Stock prices trending upward over years

2. SEASONALITY: Regular patterns that repeat at fixed intervals
   - Example: Retail stocks peak during holiday seasons

3. CYCLICAL PATTERNS: Long-term oscillations (not fixed period)
   - Example: Economic cycles (boom and bust)

4. IRREGULARITY/NOISE: Random fluctuations
   - Example: Daily market volatility

Why Time Series is DIFFERENT from Regular ML:
- ❌ CAN'T shuffle data (order matters!)
- ❌ CAN'T use random train-test split
- ✅ MUST use sequential split or walk-forward validation
- ✅ Past data used to predict future (autocorrelation)
- ✅ Temporal dependencies between observations

Stock Market Time Series Specifics:
- Non-stationary: Mean and variance change over time
- Random Walk: Today's price ≈ Yesterday's price + noise
- Efficient Market Hypothesis: Hard to predict consistently!
- High noise-to-signal ratio
"""

print("\n📊 Key Time Series Concepts:")
print("\n1. STATIONARITY:")
print("   - Statistical properties (mean, variance) constant over time")
print("   - Most ML models assume stationarity")
print("   - Stock prices are typically NON-stationary")
print("   - Returns (% change) are more stationary than prices")

print("\n2. AUTOCORRELATION:")
print("   - Correlation of data with its past values")
print("   - Strong autocorrelation = predictable patterns")
print("   - Used to determine how many past days to use (lag features)")

print("\n3. VOLATILITY:")
print("   - Measure of price fluctuation")
print("   - High volatility = high risk & opportunity")
print("   - Calculated using standard deviation of returns")

print("\n4. WALK-FORWARD VALIDATION:")
print("   - Train on past data, test on future data")
print("   - Mimics real-world trading scenario")
print("   - Example: Train on 2020-2023, test on 2024")

print("\n✅ Exercise 1.1 Complete!")
print("=" * 80)
```

```python
# ==================================================
# EXERCISE 1.2: STOCK MARKET DATA STRUCTURE (OHLCV)
# ==================================================

print("\n" + "=" * 80)
print("EXERCISE 1.2: Understanding Stock Market Data (OHLCV)")
print("=" * 80)

"""
📖 THEORY: OHLCV Data Format

Stock market data comes in OHLCV format:

O - OPEN: Price at market open (9:30 AM EST for US stocks)
H - HIGH: Highest price during the trading day
L - LOW: Lowest price during the trading day
C - CLOSE: Price at market close (4:00 PM EST for US stocks)
V - VOLUME: Number of shares traded

Why each matters:
- OPEN: Shows overnight sentiment and gap up/down
- HIGH/LOW: Shows intraday volatility range
- CLOSE: Most important! Used for technical indicators
- VOLUME: Confirms price movements (high volume = strong signal)

Additional Important Concepts:
- Adjusted Close: Accounts for stock splits and dividends
  Always use Adj Close for analysis!
- Market Cap: Total value of company (price × shares)
- Trading Days: Markets closed on weekends and holidays
"""

print("\n📊 Visualizing OHLCV Concept:")
print("\nExample trading day:")
print("  Open:   $150.00  (9:30 AM)")
print("  High:   $155.50  (peak during day)")
print("  Low:    $148.25  (dip during day)")
print("  Close:  $153.75  (4:00 PM)")
print("  Volume: 5.2M shares traded")

print("\n📈 What this tells us:")
print("  ✅ Bullish day (closed higher than opened)")
print("  ✅ Moderate volatility ($7.25 range)")
print("  ✅ High volume (indicates strong interest)")

print("\n✅ Exercise 1.2 Complete!")
print("=" * 80)
```

```python
# ==================================================
# EXERCISE 2.2: FETCH HISTORICAL DATA USING YAHOO FINANCE API
# ==================================================

print("\n" + "=" * 80)
print("EXERCISE 2.2: Downloading Historical Stock Data")
print("=" * 80)

"""
📖 THEORY: Yahoo Finance API (yfinance)

yfinance is a Python library that fetches data from Yahoo Finance:
- FREE and no API key required
- Real-time and historical data
- OHLCV data automatically included
- Handles stock splits and dividends (adjusted close)

Key Parameters:
- ticker: Stock symbol (e.g., 'TSLA')
- start: Start date for historical data
- end: End date for historical data
- interval: '1d' (daily), '1wk' (weekly), '1mo' (monthly)
- auto_adjust: Whether to use adjusted prices (True recommended)
"""

print("\n⏱️ Downloading historical data for all stocks...")
print("   This may take 1-2 minutes...\n")

# Dictionary to store dataframes
stock_data = {}
failed_stocks = []

for ticker, name in stocks.items():
    try:
        print(f"⏳ Fetching {ticker} ({name})...")

        # Download data
        df = yf.download(
            ticker,
            start=start_date,
            end=end_date,
            progress=False,
            auto_adjust=True  # Use adjusted prices
        )

        if len(df) > 0:
            # Add ticker column for reference
            df['Ticker'] = ticker
            stock_data[ticker] = df

            print(f"   ✅ Downloaded {len(df)} trading days")
            print(f"   📅 Range: {df.index[0].strftime('%Y-%m-%d')} to {df.index[-1].strftime('%Y-%m-%d')}")
        else:
            print(f"   ❌ No data available")
            failed_stocks.append(ticker)

    except Exception as e:
        print(f"   ❌ Error fetching data: {str(e)}")
        failed_stocks.append(ticker)

    print()

print("\n" + "=" * 80)
print("📊 Data Download Summary:")
print(f"\n✅ Successfully downloaded: {len(stock_data)} stocks")
for ticker in stock_data.keys():
    print(f"   • {ticker}: {len(stock_data[ticker])} days")

if failed_stocks:
    print(f"\n❌ Failed to download: {len(failed_stocks)} stocks")
    for ticker in failed_stocks:
        print(f"   • {ticker}")

print("\n✅ Exercise 2.2 Complete!")
print("=" * 80)
```

```python
# ==================================================
# EXERCISE 3.1: PRICE TRENDS VISUALIZATION
# ==================================================

# Import matplotlib if not already done
import matplotlib.pyplot as plt
import numpy as np

print("\n" + "=" * 80)
print("EXERCISE 3.1: Visualizing Price Trends Over Time")
print("=" * 80)

"""
📖 THEORY: Price Trends

Analyzing price trends helps us understand:
- Long-term direction (bull market vs bear market)
- Volatility (how much prices fluctuate)
- Support and resistance levels
- Overall market sentiment

We'll visualize Close prices because:
- Most important price of the day
- Used for all technical indicators
- Represents final agreed price
"""

print("\n⏱️ Creating price trend visualizations...")

# Create figure with subplots
fig, axes = plt.subplots(len(stock_data), 1, figsize=(15, 4*len(stock_data)))

if len(stock_data) == 1:
    axes = [axes]

for idx, (ticker, df) in enumerate(stock_data.items()):
    ax = axes[idx]

    # Plot close price
    ax.plot(df.index, df['Close'], linewidth=2, label=f'{ticker} Close Price', color='#2E86AB')

    # Calculate and plot moving averages
    df['MA50'] = df['Close'].rolling(window=50).mean()
    df['MA200'] = df['Close'].rolling(window=200).mean()

    ax.plot(df.index, df['MA50'], linewidth=1.5, alpha=0.7, label='50-day MA', linestyle='--', color='orange')
    ax.plot(df.index, df['MA200'], linewidth=1.5, alpha=0.7, label='200-day MA', linestyle='--', color='red')

    ax.set_title(f'{ticker} - Price Trend (5 Years)', fontsize=14, fontweight='bold')
    ax.set_xlabel('Date', fontsize=12)
    ax.set_ylabel('Price ($)', fontsize=12)
    ax.legend(loc='upper left')
    ax.grid(True, alpha=0.3)

    # Get scalar values for start and end prices
    start_price = float(df['Close'].iloc[0])
    end_price = float(df['Close'].iloc[-1])
    price_change = ((end_price - start_price) / start_price) * 100

    # Add text box with statistics
    textstr = f'Start: ${start_price:.2f}\nEnd: ${end_price:.2f}\nChange: {price_change:+.2f}%'
    props = dict(boxstyle='round', facecolor='wheat', alpha=0.5)
    ax.text(0.02, 0.98, textstr, transform=ax.transAxes, fontsize=10,
            verticalalignment='top', bbox=props)

plt.tight_layout()
plt.show()

# Print summary statistics
print("\n📊 Price Change Summary (5-year period):")
for ticker, df in stock_data.items():
    start_price = float(df['Close'].iloc[0])
    end_price = float(df['Close'].iloc[-1])
    change_pct = ((end_price - start_price) / start_price) * 100

    emoji = "📈" if change_pct > 0 else "📉"
    print(f"\n{emoji} {ticker}:")
    print(f"   Start: ${start_price:.2f}")
    print(f"   End: ${end_price:.2f}")
    print(f"   Change: {change_pct:+.2f}%")
    print(f"   Max: ${float(df['Close'].max()):.2f}")
    print(f"   Min: ${float(df['Close'].min()):.2f}")

print("\n✅ Exercise 3.1 Complete!")
print("=" * 80)
```

```python
# ==================================================
# EXERCISE 3.2: CALCULATE RETURNS AND VOLATILITY
# ==================================================

print("\n" + "=" * 80)
print("EXERCISE 3.2: Calculating Returns and Volatility")
print("=" * 80)

"""
📖 THEORY: Returns vs Prices

Why use RETURNS instead of PRICES?
1. Stationary: Returns are more stationary than prices
2. Comparable: Can compare stocks of different prices
3. Scale-free: $1 increase means different things for $10 vs $1000 stock

Types of Returns:
- Daily Return: (Today's Price - Yesterday's Price) / Yesterday's Price
- Log Return: log(Today's Price / Yesterday's Price)
  More mathematically convenient for time series

Volatility:
- Standard deviation of returns
- Measures risk and price fluctuation
- Higher volatility = higher risk & potential reward
"""

print("\n⏱️ Calculating returns and volatility for all stocks...")

returns_data = {}

for ticker, df in stock_data.items():
    # Calculate daily returns
    df['Daily_Return'] = df['Close'].pct_change() * 100  # Convert to percentage

    # Calculate log returns
    df['Log_Return'] = np.log(df['Close'] / df['Close'].shift(1)) * 100

    # Store for comparison
    returns_data[ticker] = df['Daily_Return'].dropna()

    # Calculate volatility (annualized)
    daily_volatility = df['Daily_Return'].std()
    annual_volatility = daily_volatility * np.sqrt(252)  # 252 trading days per year

    print(f"\n📊 {ticker} Statistics:")
    print(f"   Mean Daily Return: {df['Daily_Return'].mean():.4f}%")
    print(f"   Median Daily Return: {df['Daily_Return'].median():.4f}%")
    print(f"   Daily Volatility: {daily_volatility:.4f}%")
    print(f"   Annual Volatility: {annual_volatility:.2f}%")
    print(f"   Max Daily Gain: {df['Daily_Return'].max():.2f}%")
    print(f"   Max Daily Loss: {df['Daily_Return'].min():.2f}%")

# Visualize returns distribution
print("\n⏱️ Creating returns distribution plots...")

fig, axes = plt.subplots(2, 3, figsize=(15, 10))
axes = axes.flatten()

for idx, (ticker, returns) in enumerate(returns_data.items()):
    ax = axes[idx]

    # Histogram
    ax.hist(returns, bins=100, alpha=0.7, edgecolor='black', color='skyblue')
    ax.axvline(returns.mean(), color='red', linestyle='--', linewidth=2, label=f'Mean: {returns.mean():.3f}%')
    ax.axvline(0, color='black', linestyle='-', linewidth=1, alpha=0.5)

    ax.set_title(f'{ticker} - Daily Returns Distribution', fontweight='bold')
    ax.set_xlabel('Daily Return (%)')
    ax.set_ylabel('Frequency')
    ax.legend()
    ax.grid(True, alpha=0.3)

# Hide extra subplot
axes[5].axis('off')

plt.tight_layout()
plt.show()

print("\n✅ Exercise 3.2 Complete!")
print("=" * 80)
```

```python
# ==================================================
# EXERCISE 3.3: VOLUME ANALYSIS
# ==================================================

print("\n" + "=" * 80)
print("EXERCISE 3.3: Trading Volume Analysis")
print("=" * 80)

"""
📖 THEORY: Why Volume Matters

Volume is the number of shares traded in a period.
It's crucial because:
- Confirms price movements (high volume = strong signal)
- Shows market interest and liquidity
- Low volume = weak signal, price can reverse easily
- Volume spikes often precede major price moves

Volume + Price Patterns:
- ↑ Price + ↑ Volume = Strong bullish signal
- ↑ Price + ↓ Volume = Weak, may reverse
- ↓ Price + ↑ Volume = Strong bearish signal
- ↓ Price + ↓ Volume = Weak selling
"""

print("\n⏱️ Analyzing trading volume patterns...")

# Create volume analysis plots
fig, axes = plt.subplots(len(stock_data), 1, figsize=(15, 4*len(stock_data)))

if len(stock_data) == 1:
    axes = [axes]

for idx, (ticker, df) in enumerate(stock_data.items()):
    ax = axes[idx]

    # Plot volume with color based on price change
    colors = ['green' if df['Close'].iloc[i] >= df['Open'].iloc[i] else 'red'
              for i in range(len(df))]

    ax.bar(df.index, df['Volume'], color=colors, alpha=0.5, width=1)

    # Add moving average of volume
    df['Volume_MA50'] = df['Volume'].rolling(window=50).mean()
    ax.plot(df.index, df['Volume_MA50'], color='blue', linewidth=2, label='50-day Volume MA')

    ax.set_title(f'{ticker} - Trading Volume', fontsize=14, fontweight='bold')
    ax.set_xlabel('Date', fontsize=12)
    ax.set_ylabel('Volume', fontsize=12)
    ax.legend()
    ax.grid(True, alpha=0.3)

    # Calculate volume statistics
    avg_volume = float(df['Volume'].mean())
    max_volume = float(df['Volume'].max())

    textstr = f'Avg Volume: {avg_volume:,.0f}\nMax Volume: {max_volume:,.0f}'
    props = dict(boxstyle='round', facecolor='lightblue', alpha=0.5)
    ax.text(0.02, 0.98, textstr, transform=ax.transAxes, fontsize=10,
            verticalalignment='top', bbox=props)

plt.tight_layout()
plt.show()

# Print volume statistics
print("\n📊 Volume Statistics:")
for ticker, df in stock_data.items():
    print(f"\n📈 {ticker}:")
    print(f"   Average Volume: {float(df['Volume'].mean()):,.0f}")
    print(f"   Median Volume: {float(df['Volume'].median()):,.0f}")
    print(f"   Max Volume: {float(df['Volume'].max()):,.0f}")
    print(f"   Min Volume: {float(df['Volume'].min()):,.0f}")
    print(f"   Std Dev: {float(df['Volume'].std()):,.0f}")

print("\n✅ Exercise 3.3 Complete!")
print("=" * 80)
```
