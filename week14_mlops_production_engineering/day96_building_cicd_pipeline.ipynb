```python
# ==================================================
# EXERCISE 1.1: WHAT IS CI/CD?
# ==================================================

print("\n" + "=" * 80)
print("EXERCISE 1.1: What is CI/CD?")
print("=" * 80)

"""
📖 THEORY: CI/CD

CI/CD stands for Continuous Integration / Continuous Delivery (or Deployment).
It's the practice of automating the steps between writing code and
running it in production.

Without CI/CD:
  Developer writes code → manually runs tests → manually builds Docker image
  → manually pushes to registry → manually SSHs into server → manually deploys
  → hopes nothing broke

With CI/CD:
  Developer pushes code → everything above happens AUTOMATICALLY
"""

print("""
📋 CI/CD STAGES:

  ─── Continuous Integration (CI) ─────────────────────────────
  Triggered on: every push / pull request

  1. Checkout code
  2. Install dependencies
  3. Run linting (code style checks)
  4. Run unit tests
  5. Run integration tests
  6. Build Docker image (verify it builds successfully)
  7. Report results → ✅ pass or ❌ fail

  Goal: catch bugs EARLY, before they reach production
        Every PR is automatically verified before merging

  ─── Continuous Delivery (CD) ────────────────────────────────
  Triggered on: merge to main branch (after CI passes)

  1. Build final Docker image
  2. Push image to registry (Docker Hub / AWS ECR / GitHub Container Registry)
  3. Deploy to staging environment
  4. Run smoke tests on staging
  5. (Optional) Deploy to production automatically or with manual approval

  Goal: always have a deployable artifact ready
        Reduce manual steps in the release process
""")

print("""
📊 WHY CI/CD MATTERS FOR ML PROJECTS:

  Without CI/CD              With CI/CD
  ──────────────────────     ──────────────────────────────────
  "Works on my machine"      Same environment every time (Docker)
  Manual testing (skipped)   Automated tests run on every push
  Big infrequent releases    Small frequent releases (less risk)
  Hours to deploy            Minutes to deploy
  Fear of breaking prod      Confidence from automated checks
  No audit trail             Every change tracked + tested

  ML-specific benefits:
  ✅ Automatically test model loading (catches missing weights)
  ✅ Verify API endpoints return correct shapes after refactoring
  ✅ Ensure Docker image builds with correct CUDA/OpenCV versions
  ✅ Track model performance metrics across versions
""")

print("\n✅ Exercise 1.1 Complete!")
print("=" * 80)
```

```python
# ==================================================
# EXERCISE 2.1: GITHUB ACTIONS STRUCTURE
# ==================================================

print("\n" + "=" * 80)
print("EXERCISE 2.1: GitHub Actions Core Concepts")
print("=" * 80)

"""
📖 THEORY: GitHub Actions

GitHub Actions is GitHub's built-in CI/CD platform.
Workflows are defined in YAML files inside .github/workflows/.
They run on GitHub's servers (runners) — free for public repos,
limited free minutes for private repos.

Key concepts:
- Workflow   → the entire automation (one YAML file)
- Trigger    → what starts the workflow (push, PR, schedule, manual)
- Job        → a group of steps that run on one runner machine
- Step       → a single task inside a job (run a command or use an action)
- Action     → a reusable step from the GitHub Marketplace
- Runner     → the virtual machine that executes the job
- Secret     → encrypted variable stored in GitHub repo settings
"""

print("""
📋 GITHUB ACTIONS YAML STRUCTURE:

  name: Workflow name (shown in GitHub UI)

  on:                           ← TRIGGERS
    push:
      branches: [main, dev]
    pull_request:
      branches: [main]
    schedule:
      - cron: "0 2 * * *"       ← Daily at 2am UTC
    workflow_dispatch:           ← Manual trigger button in GitHub UI

  env:                          ← GLOBAL ENVIRONMENT VARIABLES
    PYTHON_VERSION: "3.10"
    REGISTRY: docker.io

  jobs:                         ← JOBS (run in parallel by default)

    job-name:                   ← Job ID (kebab-case)
      name: Human readable name
      runs-on: ubuntu-latest    ← Runner OS
      needs: [other-job]        ← Wait for other job to finish first

      strategy:                 ← Matrix: run job multiple times
        matrix:
          python-version: [3.9, 3.10, 3.11]

      steps:                    ← STEPS (run sequentially)

        - name: Step name
          uses: actions/checkout@v4     ← Use a marketplace action

        - name: Another step
          run: echo "Hello World"       ← Run a shell command

        - name: Multi-line command
          run: |
            pip install -r requirements.txt
            pytest tests/

        - name: Step with env vars
          env:
            MY_SECRET: \${{ secrets.MY_SECRET }}   ← Inject secret
          run: echo "Using secret"

        - name: Conditional step
          if: github.ref == 'refs/heads/main'      ← Only on main branch
          run: echo "On main!"
""")

print("\n✅ Exercise 2.1 Complete!")
print("=" * 80)
```

```python
# ==================================================
# EXERCISE 3.1: CI WORKFLOW - LINT AND TEST
# ==================================================

print("\n" + "=" * 80)
print("EXERCISE 3.1: CI Workflow — Lint & Test")
print("=" * 80)

ci_workflow = '''
# ============================================================
# .github/workflows/ci.yml
# Runs on every push and PR — lint, test, build verification
# ============================================================

name: CI Pipeline

on:
  push:
    branches: ["**"]             # All branches
  pull_request:
    branches: [main]

env:
  PYTHON_VERSION: "3.10"

jobs:

  # ── Job 1: Lint ───────────────────────────────────────────
  lint:
    name: Lint & Format Check
    runs-on: ubuntu-latest

    steps:
      - name: Checkout code
        uses: actions/checkout@v4

      - name: Set up Python
        uses: actions/setup-python@v5
        with:
          python-version: ${{ env.PYTHON_VERSION }}

      - name: Cache pip packages
        uses: actions/cache@v4
        with:
          path: ~/.cache/pip
          key: pip-lint-${{ hashFiles('requirements.api.txt') }}

      - name: Install lint tools
        run: pip install ruff black

      - name: Run Ruff (linter)
        run: ruff check api/ --output-format=github

      - name: Run Black (formatter check)
        run: black --check api/


  # ── Job 2: Test ───────────────────────────────────────────
  test:
    name: Run Tests
    runs-on: ubuntu-latest
    needs: lint                  # Only run if lint passes

    services:
      redis:
        image: redis:7-alpine
        ports:
          - 6379:6379
        options: >-
          --health-cmd "redis-cli ping"
          --health-interval 10s
          --health-timeout 5s
          --health-retries 5

    steps:
      - name: Checkout code
        uses: actions/checkout@v4

      - name: Set up Python
        uses: actions/setup-python@v5
        with:
          python-version: ${{ env.PYTHON_VERSION }}

      - name: Cache pip packages
        uses: actions/cache@v4
        with:
          path: ~/.cache/pip
          key: pip-test-${{ hashFiles('requirements.api.txt') }}
          restore-keys: pip-test-

      - name: Install dependencies
        run: pip install -r requirements.api.txt pytest pytest-asyncio httpx

      - name: Run tests
        env:
          REDIS_URL: redis://localhost:6379/0
          MODEL_PATH: tests/fixtures/mock_model.pt
          SECRET_KEY: test-secret-key-not-for-production
          API_KEY: test-api-key
          LOG_LEVEL: WARNING
        run: pytest tests/ -v --tb=short

      - name: Upload test results
        if: always()             # Upload even if tests fail
        uses: actions/upload-artifact@v4
        with:
          name: test-results
          path: tests/results/


  # ── Job 3: Docker Build Verification ─────────────────────
  build-verify:
    name: Verify Docker Build
    runs-on: ubuntu-latest
    needs: test                  # Only run if tests pass

    steps:
      - name: Checkout code
        uses: actions/checkout@v4

      - name: Set up Docker Buildx
        uses: docker/setup-buildx-action@v3

      - name: Build API image (no push)
        uses: docker/build-push-action@v5
        with:
          context: .
          file: Dockerfile.api
          push: false            # Build only — verify it compiles
          tags: security-api:test
          cache-from: type=gha
          cache-to: type=gha,mode=max

      - name: Build Streamlit image (no push)
        uses: docker/build-push-action@v5
        with:
          context: .
          file: Dockerfile.streamlit
          push: false
          tags: security-ui:test
          cache-from: type=gha
          cache-to: type=gha,mode=max
'''

print("📄 .github/workflows/ci.yml:")
print(ci_workflow)

print("\n✅ Exercise 3.1 Complete!")
print("=" * 80)
```

```python
# ==================================================
# EXERCISE 3.2: WRITING TESTS FOR THE API
# ==================================================

print("\n" + "=" * 80)
print("EXERCISE 3.2: Writing Tests for the FastAPI Security System")
print("=" * 80)

"""
📖 THEORY: Testing FastAPI with TestClient

FastAPI provides a TestClient (built on httpx) that lets you
make real HTTP requests to your app without running a server.

Test types:
- Unit tests      → test individual functions in isolation
- Integration tests → test API endpoints end-to-end
- Smoke tests     → verify the app starts and health check passes
"""

conftest_py = '''
# ============================================================
# tests/conftest.py — Shared Test Fixtures
# ============================================================

import pytest
from fastapi.testclient import TestClient
from unittest.mock import MagicMock, patch

from api.main import app


@pytest.fixture(scope="session")
def mock_model():
    """Mock YOLO model — avoids loading real weights in CI."""
    model = MagicMock()
    model.return_value = [MagicMock(boxes=[])]   # Empty detections
    return model


@pytest.fixture(scope="session")
def client(mock_model):
    """
    Test client with mocked model loaded into app state.
    scope="session" means the client is created ONCE per test run.
    """
    app.state.model = mock_model
    with TestClient(app) as c:
        yield c


@pytest.fixture
def auth_headers():
    """Valid API key headers for authenticated requests."""
    return {"X-API-Key": "test-api-key"}


@pytest.fixture
def admin_headers(client, auth_headers):
    """JWT token headers with admin role."""
    response = client.post("/auth/login", data={
        "username": "admin",
        "password": "testpassword"
    })
    token = response.json()["access_token"]
    return {"Authorization": f"Bearer {token}"}
'''

test_health_py = '''
# ============================================================
# tests/test_health.py — Smoke Tests
# ============================================================

def test_root_returns_200(client):
    response = client.get("/")
    assert response.status_code == 200
    assert "version" in response.json()


def test_health_check_returns_healthy(client):
    response = client.get("/health")
    assert response.status_code == 200
    data = response.json()
    assert data["status"] == "healthy"
    assert "timestamp" in data


def test_health_check_model_loaded(client):
    response = client.get("/health")
    assert response.json()["model_loaded"] is True
'''

test_detections_py = '''
# ============================================================
# tests/test_detections.py — Detection Endpoint Tests
# ============================================================

import io
import pytest
from PIL import Image


def create_test_image() -> bytes:
    """Create a small test image in memory."""
    img = Image.new("RGB", (640, 480), color=(100, 100, 100))
    buffer = io.BytesIO()
    img.save(buffer, format="JPEG")
    return buffer.getvalue()


class TestListDetections:

    def test_list_detections_requires_auth(self, client):
        """Unauthenticated request should return 401."""
        response = client.get("/api/v1/detections")
        assert response.status_code == 401

    def test_list_detections_returns_200(self, client, auth_headers):
        """Authenticated request should return 200 with valid structure."""
        response = client.get("/api/v1/detections", headers=auth_headers)
        assert response.status_code == 200
        data = response.json()
        assert "items" in data
        assert "total" in data
        assert "limit" in data
        assert "offset" in data

    def test_list_detections_pagination(self, client, auth_headers):
        """Pagination params should be respected."""
        response = client.get(
            "/api/v1/detections?limit=5&offset=0",
            headers=auth_headers
        )
        assert response.status_code == 200
        assert response.json()["limit"] == 5

    def test_list_detections_limit_max(self, client, auth_headers):
        """Limit exceeding 100 should return 400."""
        response = client.get(
            "/api/v1/detections?limit=999",
            headers=auth_headers
        )
        assert response.status_code == 400


class TestRunDetection:

    def test_run_detection_requires_auth(self, client):
        image_bytes = create_test_image()
        response = client.post(
            "/api/v1/detections",
            files={"file": ("test.jpg", image_bytes, "image/jpeg")},
        )
        assert response.status_code == 401

    def test_run_detection_returns_201(self, client, auth_headers):
        image_bytes = create_test_image()
        response = client.post(
            "/api/v1/detections",
            files={"file": ("test.jpg", image_bytes, "image/jpeg")},
            headers=auth_headers,
        )
        assert response.status_code == 201
        data = response.json()
        assert "id" in data
        assert "detections" in data
        assert "processing_time_ms" in data

    def test_run_detection_invalid_file_type(self, client, auth_headers):
        response = client.post(
            "/api/v1/detections",
            files={"file": ("test.txt", b"not an image", "text/plain")},
            headers=auth_headers,
        )
        assert response.status_code == 400

    def test_run_detection_confidence_validation(self, client, auth_headers):
        """Confidence > 1.0 should return 422."""
        image_bytes = create_test_image()
        response = client.post(
            "/api/v1/detections?confidence=1.5",
            files={"file": ("test.jpg", image_bytes, "image/jpeg")},
            headers=auth_headers,
        )
        assert response.status_code == 422
'''

print("📄 tests/conftest.py:")
print(conftest_py)
print("\n📄 tests/test_health.py:")
print(test_health_py)
print("\n📄 tests/test_detections.py:")
print(test_detections_py)

print("\n✅ Exercise 3.2 Complete!")
print("=" * 80)
```

```python
# ==================================================
# EXERCISE 4.1: CD WORKFLOW - BUILD AND PUSH
# ==================================================

print("\n" + "=" * 80)
print("EXERCISE 4.1: CD Workflow — Docker Build & Push")
print("=" * 80)

cd_workflow = '''
# ============================================================
# .github/workflows/cd.yml
# Runs on merge to main — builds, pushes Docker image, deploys
# ============================================================

name: CD Pipeline

on:
  push:
    branches: [main]             # Only on main branch merges
  workflow_dispatch:             # Manual trigger

env:
  REGISTRY: docker.io
  IMAGE_NAME_API: ${{ secrets.DOCKERHUB_USERNAME }}/security-api
  IMAGE_NAME_UI: ${{ secrets.DOCKERHUB_USERNAME }}/security-ui

jobs:

  # ── Job 1: Build & Push Docker Images ────────────────────
  push-docker:
    name: Build & Push Docker Images
    runs-on: ubuntu-latest

    outputs:
      api-tag: ${{ steps.meta-api.outputs.tags }}
      ui-tag: ${{ steps.meta-ui.outputs.tags }}

    steps:
      - name: Checkout code
        uses: actions/checkout@v4

      - name: Set up Docker Buildx
        uses: docker/setup-buildx-action@v3

      - name: Log in to Docker Hub
        uses: docker/login-action@v3
        with:
          username: ${{ secrets.DOCKERHUB_USERNAME }}
          password: ${{ secrets.DOCKERHUB_TOKEN }}  # Access token, not password!

      # ── Generate image tags ───────────────────────────────
      - name: Generate API image metadata
        id: meta-api
        uses: docker/metadata-action@v5
        with:
          images: ${{ env.IMAGE_NAME_API }}
          tags: |
            type=sha,prefix=sha-           # sha-abc1234
            type=raw,value=latest          # latest (on main only)
            type=semver,pattern={{version}} # v1.0.0 (on git tags)

      - name: Generate UI image metadata
        id: meta-ui
        uses: docker/metadata-action@v5
        with:
          images: ${{ env.IMAGE_NAME_UI }}
          tags: |
            type=sha,prefix=sha-
            type=raw,value=latest

      # ── Build and push images ─────────────────────────────
      - name: Build & Push API image
        uses: docker/build-push-action@v5
        with:
          context: .
          file: Dockerfile.api
          push: true
          tags: ${{ steps.meta-api.outputs.tags }}
          labels: ${{ steps.meta-api.outputs.labels }}
          cache-from: type=gha
          cache-to: type=gha,mode=max

      - name: Build & Push Streamlit image
        uses: docker/build-push-action@v5
        with:
          context: .
          file: Dockerfile.streamlit
          push: true
          tags: ${{ steps.meta-ui.outputs.tags }}
          labels: ${{ steps.meta-ui.outputs.labels }}
          cache-from: type=gha
          cache-to: type=gha,mode=max


  # ── Job 2: Deploy ────────────────────────────────────────
  deploy:
    name: Deploy to Server
    runs-on: ubuntu-latest
    needs: push-docker           # Wait for images to be pushed
    environment: production      # Requires manual approval in GitHub UI

    steps:
      - name: Deploy via SSH
        uses: appleboy/ssh-action@v1
        with:
          host: ${{ secrets.SERVER_HOST }}
          username: ${{ secrets.SERVER_USER }}
          key: ${{ secrets.SERVER_SSH_KEY }}
          script: |
            cd /opt/security-system

            # Pull latest images
            docker compose pull

            # Restart services with zero downtime
            docker compose up -d --no-build

            # Remove old unused images
            docker image prune -f

            # Verify deployment
            sleep 10
            curl -f http://localhost:8000/health || exit 1
            echo "✅ Deployment successful!"
'''

print("📄 .github/workflows/cd.yml:")
print(cd_workflow)

print("""
💡 KEY CD CONCEPTS:

  Image tagging strategy:
  ├── sha-abc1234  → immutable, ties image to exact commit
  ├── latest       → always points to newest main build
  └── v1.0.0       → semantic version (from git tag)

  environment: production:
  → Creates a manual approval gate in GitHub
  → Someone must approve the deployment in GitHub UI before it runs
  → Gives you a "confirm before going live" safety net

  docker compose pull:
  → Downloads the NEW image tagged :latest
  → docker compose up -d then restarts containers using the new image
  → Old containers stop, new ones start — minimal downtime
""")

print("\n✅ Exercise 4.1 Complete!")
print("=" * 80)
```
