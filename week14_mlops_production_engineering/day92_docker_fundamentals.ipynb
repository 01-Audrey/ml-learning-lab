```python
# ==================================================
# EXERCISE 1.1: WHAT IS CONTAINERIZATION?
# ==================================================

print("\n" + "=" * 80)
print("EXERCISE 1.1: What is Containerization?")
print("=" * 80)

"""
📖 THEORY: Containerization

- A container is a lightweight, standalone, executable package that includes
  everything needed to run a piece of software: code, runtime, libraries,
  environment variables, and config files.

- Containers share the host OS kernel (unlike VMs which each have their own OS)
  making them much faster to start and more resource-efficient.

- Docker is the most popular containerization platform.

Why this matters for ML:
- Eliminates "it works on my machine" — your model runs the same everywhere
- Makes deployment to cloud (AWS, GCP, Azure) straightforward
- Reproducible experiments: pin exact library versions in the image
- Easy scaling: spin up multiple containers to handle more inference requests
"""

print("""
🐳 KEY DOCKER CONCEPTS:

  IMAGE       → Blueprint/template (read-only)
                Like a class definition in Python

  CONTAINER   → Running instance of an image
                Like an object instantiated from a class

  DOCKERFILE  → Recipe/instructions to BUILD an image
                A text file with step-by-step commands

  REGISTRY    → Storage for Docker images
                Docker Hub = public registry (like GitHub for images)

  VOLUME      → Persistent storage that survives container restarts
                Map a folder on your machine → folder in the container

  NETWORK     → How containers communicate with each other
                Containers on the same network can talk by name
""")

print("⏱️  Understanding the VM vs Container difference...")

print("""
📊 VM vs Container:

  Virtual Machine:
  ┌─────────────────────────────┐
  │  App A    │  App B    │ App C│
  │  Libs     │  Libs     │ Libs │
  │  Guest OS │  Guest OS │ GOS  │
  ├─────────────────────────────┤
  │         Hypervisor          │
  │         Host OS             │
  │         Hardware            │
  └─────────────────────────────┘
  → Each VM has its own OS (heavy, slow to start, GBs each)

  Containers:
  ┌─────────────────────────────┐
  │  App A    │  App B    │ App C│
  │  Libs     │  Libs     │ Libs │
  ├─────────────────────────────┤
  │         Docker Engine       │
  │         Host OS             │
  │         Hardware            │
  └─────────────────────────────┘
  → Containers share the Host OS kernel (lightweight, MBs, starts in seconds)
""")

print("\n✅ Exercise 1.1 Complete!")
print("=" * 80)
```

```python
# ==================================================
# EXERCISE 1.2: ESSENTIAL DOCKER COMMANDS
# ==================================================

print("\n" + "=" * 80)
print("EXERCISE 1.2: Essential Docker Commands")
print("=" * 80)

"""
📖 THEORY: Docker CLI

The Docker CLI is how you interact with Docker Engine.
Commands follow the pattern: docker <object> <action>

Most common workflow:
  Build image → Run container → Check logs → Stop → Remove
"""

print("""
📋 DOCKER COMMAND CHEATSHEET:

  ─── IMAGES ──────────────────────────────────────────────
  docker build -t myapp:v1 .        # Build image from Dockerfile in current dir
  docker images                     # List all local images
  docker pull python:3.10-slim      # Download image from Docker Hub
  docker push myrepo/myapp:v1       # Upload image to registry
  docker rmi myapp:v1               # Remove an image
  docker image prune                # Remove all unused images

  ─── CONTAINERS ──────────────────────────────────────────
  docker run myapp:v1               # Create + start a container
  docker run -it myapp:v1 bash      # Interactive mode (enter the container)
  docker run -d myapp:v1            # Detached mode (runs in background)
  docker run -p 8000:8000 myapp:v1  # Map host port 8000 → container port 8000
  docker run -v ./data:/app/data myapp:v1  # Mount volume
  docker ps                         # List running containers
  docker ps -a                      # List ALL containers (including stopped)
  docker stop <container_id>        # Gracefully stop a container
  docker rm <container_id>          # Remove a stopped container
  docker logs <container_id>        # View container logs
  docker exec -it <id> bash         # Enter a running container

  ─── DOCKER COMPOSE ──────────────────────────────────────
  docker compose up                 # Start all services
  docker compose up -d              # Start in background
  docker compose down               # Stop and remove containers
  docker compose logs               # View all service logs
  docker compose build              # Rebuild images
""")

print("⏱️  Verifying Docker is installed on this machine...")

try:
    result = subprocess.run(["docker", "--version"], capture_output=True, text=True)
    if result.returncode == 0:
        print(f"✅ Docker found: {result.stdout.strip()}")
    else:
        print("⚠️  Docker not found in PATH — make sure Docker Desktop is running!")
except FileNotFoundError:
    print("⚠️  Docker not found — start Docker Desktop and ensure it's in your PATH")

print("\n✅ Exercise 1.2 Complete!")
print("=" * 80)
```

```python
# ==================================================
# EXERCISE 1.3: DOCKER IMAGE LAYERS & CACHING
# ==================================================

print("\n" + "=" * 80)
print("EXERCISE 1.3: Docker Image Layers & Caching")
print("=" * 80)

"""
📖 THEORY: Layer System

Every instruction in a Dockerfile creates a new READ-ONLY layer.
Layers are stacked to form the final image.

Why layers matter:
- Docker CACHES each layer — if a layer hasn't changed, it's reused
- This makes rebuilds MUCH faster
- Layer order matters: put things that change LEAST at the top
- Each layer adds to image size — minimize layers where possible
"""

print("""
📊 LAYER CACHING STRATEGY:

  BAD ORDER (slow rebuilds):          GOOD ORDER (fast rebuilds):
  ─────────────────────────           ────────────────────────────
  FROM python:3.10-slim               FROM python:3.10-slim

  COPY . /app          ← changes      RUN apt-get update...    ← rarely changes
  RUN pip install...   ← reruns       COPY requirements.txt .  ← rarely changes
  RUN apt-get update   ← reruns       RUN pip install...       ← only if reqs change
                                      COPY . /app              ← changes every build

  Result: pip install runs EVERY      Result: pip install only runs when
  single time you change any code     requirements.txt changes ✅

  Rule: Most stable → top, Most frequently changing → bottom
""")

print("""
⏱️  Layer size tips for ML images:

  ✅ Use slim/alpine base images:
     python:3.10-slim  (~150MB) vs python:3.10  (~900MB)

  ✅ Combine RUN commands to reduce layers:
     RUN apt-get update && apt-get install -y libgl1 && rm -rf /var/lib/apt/lists/*

  ✅ Use .dockerignore to exclude files:
     __pycache__/, *.pyc, .git/, data/, models/, venv/
     (same concept as .gitignore)

  ✅ Use multi-stage builds for production:
     Stage 1: builder (install everything, compile)
     Stage 2: runtime (copy only what's needed to run)
""")

print("\n✅ Exercise 1.3 Complete!")
print("=" * 80)
```

```python
# ==================================================
# EXERCISE 2.1: DOCKERFILE SYNTAX & INSTRUCTIONS
# ==================================================

print("\n" + "=" * 80)
print("EXERCISE 2.1: Dockerfile Syntax & Instructions")
print("=" * 80)

"""
📖 THEORY: Dockerfile Instructions

A Dockerfile is a plain text file named exactly "Dockerfile" (no extension).
Each line is an INSTRUCTION followed by arguments.
"""

print("""
📋 DOCKERFILE INSTRUCTIONS REFERENCE:

  FROM        → Base image to build on (always first)
                FROM python:3.10-slim

  WORKDIR     → Set working directory inside the container
                WORKDIR /app

  COPY        → Copy files from host → container
                COPY requirements.txt .
                COPY src/ ./src/

  ADD         → Like COPY but can unpack .tar files and fetch URLs
                (Prefer COPY unless you need ADD's extra features)

  RUN         → Execute a command during BUILD time (creates a layer)
                RUN pip install -r requirements.txt

  ENV         → Set environment variables
                ENV MODEL_PATH=/app/models
                ENV PYTHONUNBUFFERED=1

  EXPOSE      → Document which port the container listens on
                EXPOSE 8000
                (Does NOT actually publish the port — use -p at runtime)

  CMD         → Default command to run when container STARTS
                CMD ["python", "app.py"]
                (Can be overridden at runtime)

  ENTRYPOINT  → Fixed command that always runs (CMD becomes its arguments)
                ENTRYPOINT ["python"]
                CMD ["app.py"]

  ARG         → Build-time variable (not available at runtime)
                ARG MODEL_VERSION=latest

  VOLUME      → Declare a mount point for persistent data
                VOLUME ["/app/data"]
""")

dockerfile_basic = '''
# ============================================================
# Basic ML App Dockerfile
# ============================================================

# 1. Base image — slim keeps it small, 3.10 matches your dev env
FROM python:3.10-slim

# 2. Metadata (optional but professional)
LABEL maintainer="Audrey"
LABEL description="ML application container"

# 3. System dependencies (before pip — changes rarely)
RUN apt-get update && apt-get install -y \\
    libgl1-mesa-glx \\
    libglib2.0-0 \\
    && rm -rf /var/lib/apt/lists/*

# 4. Set working directory
WORKDIR /app

# 5. Copy requirements FIRST (caching optimization)
COPY requirements.txt .

# 6. Install Python dependencies
RUN pip install --no-cache-dir -r requirements.txt

# 7. Copy the rest of the application code
COPY . .

# 8. Expose the port
EXPOSE 8000

# 9. Default start command
CMD ["python", "app.py"]
'''

print("📄 Basic ML Dockerfile structure:")
print(dockerfile_basic)

print("\n✅ Exercise 2.1 Complete!")
print("=" * 80)
```

```python
# ==================================================
# EXERCISE 2.2: DOCKERFILE FOR STREAMLIT ML APP
# ==================================================

print("\n" + "=" * 80)
print("EXERCISE 2.2: Dockerfile for a Streamlit ML App")
print("=" * 80)

"""
📖 THEORY: Streamlit-specific Docker setup

Streamlit apps need a few extra things compared to plain Python apps:
- Port 8501 (Streamlit default)
- STREAMLIT_SERVER_* env vars to disable browser opening and set headless mode
- A .streamlit/config.toml or env vars for server config
"""

dockerfile_streamlit = '''
# ============================================================
# Dockerfile — Streamlit ML App
# ============================================================

FROM python:3.10-slim

# System libs needed for OpenCV and image processing
RUN apt-get update && apt-get install -y \\
    libgl1-mesa-glx \\
    libglib2.0-0 \\
    libsm6 \\
    libxext6 \\
    libxrender-dev \\
    && rm -rf /var/lib/apt/lists/*

WORKDIR /app

# Copy and install dependencies first (layer caching)
COPY requirements.txt .
RUN pip install --no-cache-dir -r requirements.txt

# Copy application code
COPY . .

# Streamlit config via environment variables
ENV STREAMLIT_SERVER_PORT=8501
ENV STREAMLIT_SERVER_HEADLESS=true
ENV STREAMLIT_SERVER_ENABLE_CORS=false
ENV PYTHONUNBUFFERED=1

EXPOSE 8501

CMD ["streamlit", "run", "app.py", "--server.address=0.0.0.0"]
'''

print("📄 Streamlit Dockerfile:")
print(dockerfile_streamlit)

dockerignore = '''
# .dockerignore — exclude these from the build context

# Python cache
__pycache__/
*.pyc
*.pyo
*.pyd
.Python

# Virtual environments
venv/
env/
.venv/

# Version control
.git/
.gitignore

# Large data/model files (mount as volumes instead)
data/
*.csv
*.pkl
*.h5
*.pt
*.pth

# Development files
*.ipynb
.ipynb_checkpoints/
tests/
docs/

# OS files
.DS_Store
Thumbs.db

# IDE
.vscode/
.idea/
'''

print("📄 .dockerignore file:")
print(dockerignore)

print("⏱️  Why .dockerignore matters:")
print("   Without it: Docker sends your ENTIRE project folder to the build context")
print("   This includes model files, datasets, .git history — can be GBs!")
print("   With it: Only the files your app actually needs get sent → faster builds")

print("\n✅ Exercise 2.2 Complete!")
print("=" * 80)
```

```python
# ==================================================
# EXERCISE 2.3: MULTI-STAGE DOCKERFILE
# ==================================================

print("\n" + "=" * 80)
print("EXERCISE 2.3: Multi-Stage Dockerfile for Production")
print("=" * 80)

"""
📖 THEORY: Multi-Stage Builds

Multi-stage builds let you use multiple FROM statements in one Dockerfile.
Each FROM starts a new build stage. You can COPY artifacts between stages.

The key benefit: your final image only contains what's needed to RUN the app,
not the tools you needed to BUILD it (compilers, test frameworks, etc.)

For ML: build stage installs everything including torch/tensorflow,
        runtime stage only keeps the inference dependencies.
"""

dockerfile_multistage = '''
# ============================================================
# Multi-Stage Dockerfile — Production ML API
# ============================================================

# ── STAGE 1: Builder ─────────────────────────────────────────
FROM python:3.10-slim AS builder

WORKDIR /build

# Install build dependencies
RUN apt-get update && apt-get install -y --no-install-recommends \\
    gcc \\
    g++ \\
    && rm -rf /var/lib/apt/lists/*

COPY requirements.txt .

# Install to a specific target directory
RUN pip install --no-cache-dir --prefix=/install -r requirements.txt


# ── STAGE 2: Runtime ─────────────────────────────────────────
FROM python:3.10-slim AS runtime

# Only install runtime system dependencies (no compilers!)
RUN apt-get update && apt-get install -y --no-install-recommends \\
    libgl1-mesa-glx \\
    libglib2.0-0 \\
    && rm -rf /var/lib/apt/lists/*

WORKDIR /app

# Copy only the installed packages from builder stage
COPY --from=builder /install /usr/local

# Copy application code
COPY . .

# Non-root user for security (best practice in production)
RUN adduser --disabled-password --gecos "" appuser
USER appuser

ENV PYTHONUNBUFFERED=1
ENV PYTHONDONTWRITEBYTECODE=1

EXPOSE 8000

CMD ["uvicorn", "main:app", "--host", "0.0.0.0", "--port", "8000"]
'''

print("📄 Multi-Stage Dockerfile:")
print(dockerfile_multistage)

print("""
📊 Size comparison (typical ML FastAPI app):

  Single-stage build:   ~2.1 GB
  Multi-stage build:    ~850 MB
  Savings:              ~60% smaller ✅

  Smaller image = faster pulls in CI/CD = faster deployments
""")

print("\n✅ Exercise 2.3 Complete!")
print("=" * 80)
```

```python
# ==================================================
# EXERCISE 3.1: DOCKER COMPOSE FUNDAMENTALS
# ==================================================

print("\n" + "=" * 80)
print("EXERCISE 3.1: Docker Compose Fundamentals")
print("=" * 80)

"""
📖 THEORY: Docker Compose

Docker Compose lets you define and run multi-container applications.
Instead of running multiple docker run commands with many flags,
you declare everything in a single docker-compose.yml file.

Perfect for ML systems that need:
- The ML app container
- A database (PostgreSQL, Redis)
- A message queue
- A monitoring service (Prometheus, Grafana)
"""

print("""
📋 docker-compose.yml structure:

  version       → Compose file format version
  services      → Each container you want to run
    build       → Where to find the Dockerfile
    image       → Use a pre-built image instead
    ports       → "host:container" port mapping
    volumes     → Mount local dirs into the container
    environment → Set env variables
    depends_on  → Start order (this service waits for another)
    networks    → Which network this service joins
    restart     → Restart policy (no / always / on-failure)

  volumes       → Named volumes (persistent storage)
  networks      → Custom networks
""")

compose_basic = '''
# ============================================================
# docker-compose.yml — Basic ML App
# ============================================================

version: "3.9"

services:

  # ── ML Application ─────────────────────────────────────────
  ml_app:
    build:
      context: .
      dockerfile: Dockerfile
    container_name: ml_app
    ports:
      - "8000:8000"
    volumes:
      - ./models:/app/models      # Model files persist
      - ./logs:/app/logs          # Logs persist
    environment:
      - MODEL_PATH=/app/models/best_model.pt
      - LOG_LEVEL=INFO
      - PYTHONUNBUFFERED=1
    depends_on:
      - redis
    restart: unless-stopped

  # ── Redis (for caching predictions) ────────────────────────
  redis:
    image: redis:7-alpine
    container_name: redis_cache
    ports:
      - "6379:6379"
    volumes:
      - redis_data:/data
    restart: unless-stopped

volumes:
  redis_data:
'''

print("📄 Basic docker-compose.yml:")
print(compose_basic)

print("\n✅ Exercise 3.1 Complete!")
print("=" * 80)
```

```python
# ==================================================
# EXERCISE 3.2: DOCKER COMPOSE FOR SECURITY SYSTEM
# ==================================================

print("\n" + "=" * 80)
print("EXERCISE 3.2: Docker Compose Architecture for Security System")
print("=" * 80)

"""
📖 THEORY: Planning the Security System Docker setup

Your AI Security & Surveillance System (Weeks 4-6) has:
- Streamlit frontend UI
- FastAPI backend (detection + tracking API)
- TextAI/Streamlit interface for alerts
- Model files (YOLO weights)
- Video input/output files

We'll design the full compose setup here as a plan for Day 93.
"""

print("""
📊 Security System — Container Architecture:

  ┌─────────────────────────────────────────────────────┐
  │                   Docker Network                     │
  │                                                      │
  │  ┌────────────────┐     ┌────────────────────────┐  │
  │  │  streamlit_ui  │────▶│    fastapi_backend     │  │
  │  │  (port 8501)   │     │    (port 8000)         │  │
  │  └────────────────┘     └────────────┬───────────┘  │
  │                                      │               │
  │                         ┌────────────▼───────────┐  │
  │                         │    redis_cache          │  │
  │                         │    (port 6379)          │  │
  │                         └────────────────────────┘  │
  │                                                      │
  │  Shared Volumes:                                     │
  │  ./models  → /app/models   (YOLO weights)           │
  │  ./videos  → /app/videos   (input videos)           │
  │  ./results → /app/results  (output frames)          │
  │  ./logs    → /app/logs     (system logs)            │
  └─────────────────────────────────────────────────────┘
""")

compose_security = '''
# ============================================================
# docker-compose.yml — AI Security System
# (Prepared on Day 92, implemented on Day 93)
# ============================================================

version: "3.9"

services:

  # ── FastAPI Backend (Detection + Tracking) ────────────────
  fastapi_backend:
    build:
      context: .
      dockerfile: Dockerfile.api
    container_name: security_api
    ports:
      - "8000:8000"
    volumes:
      - ./models:/app/models
      - ./videos:/app/videos
      - ./results:/app/results
      - ./logs:/app/logs
    environment:
      - MODEL_PATH=/app/models/best.pt
      - REDIS_URL=redis://redis_cache:6379
      - LOG_LEVEL=INFO
    depends_on:
      - redis_cache
    restart: unless-stopped
    deploy:
      resources:
        limits:
          memory: 4G        # YOLO needs memory

  # ── Streamlit UI (Frontend) ───────────────────────────────
  streamlit_ui:
    build:
      context: .
      dockerfile: Dockerfile.streamlit
    container_name: security_ui
    ports:
      - "8501:8501"
    volumes:
      - ./results:/app/results
    environment:
      - API_URL=http://fastapi_backend:8000
      - STREAMLIT_SERVER_HEADLESS=true
    depends_on:
      - fastapi_backend
    restart: unless-stopped

  # ── Redis Cache ───────────────────────────────────────────
  redis_cache:
    image: redis:7-alpine
    container_name: security_redis
    ports:
      - "6379:6379"
    volumes:
      - redis_data:/data
    restart: unless-stopped

volumes:
  redis_data:

networks:
  default:
    name: security_network
'''

print("📄 Security System docker-compose.yml (plan for Day 93):")
print(compose_security)

print("\n✅ Exercise 3.2 Complete!")
print("=" * 80)
```

```python
# ==================================================
# EXERCISE 3.3: USEFUL DOCKER PATTERNS FOR ML
# ==================================================

print("\n" + "=" * 80)
print("EXERCISE 3.3: Useful Docker Patterns for ML Projects")
print("=" * 80)

"""
📖 THEORY: ML-Specific Docker Patterns

A few patterns that come up repeatedly when Dockerizing ML projects.
"""

print("""
⏱️  Pattern 1: Health Checks
─────────────────────────────
Add to Dockerfile or compose to auto-restart unhealthy containers:

  HEALTHCHECK --interval=30s --timeout=10s --retries=3 \\
    CMD curl -f http://localhost:8000/health || exit 1

In docker-compose.yml:
  healthcheck:
    test: ["CMD", "curl", "-f", "http://localhost:8000/health"]
    interval: 30s
    timeout: 10s
    retries: 3
    start_period: 40s    # Give the model time to load


⏱️  Pattern 2: Environment Files
──────────────────────────────────
Keep secrets out of docker-compose.yml with .env files:

  # .env  (git-ignored!)
  MODEL_PATH=/app/models/best.pt
  API_KEY=your_secret_key

  # docker-compose.yml
  env_file:
    - .env


⏱️  Pattern 3: GPU Support
────────────────────────────
For CUDA-enabled containers (NVIDIA):

  deploy:
    resources:
      reservations:
        devices:
          - driver: nvidia
            count: 1
            capabilities: [gpu]

  Base image: FROM nvidia/cuda:11.8-cudnn8-runtime-ubuntu22.04


⏱️  Pattern 4: Model Volume Mount
───────────────────────────────────
Never bake large model weights into the image — they can be 500MB+.
Mount them as volumes instead:

  volumes:
    - ./models:/app/models:ro    # :ro = read-only (safer)

  This keeps your image small and lets you swap models without rebuilding.
""")

print("\n✅ Exercise 3.3 Complete!")
print("=" * 80)
```
