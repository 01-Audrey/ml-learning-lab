````python
# ==================================================
# EXERCISE 1.2: LOAD ALL WEEK 5 COMPONENTS
# ==================================================

print("\n" + "=" * 80)
print("EXERCISE 1.2: Load All Week 5 Components")
print("=" * 80)

"""
📖 THEORY: Component Loading

Load all components from previous days:
- Face detector (Day 29)
- Face embedder (Day 30)
- Face matcher (Day 30)
- Tracker (Day 31)
- Alert system (Day 32-33)
- Safety system (Day 34)
"""

print("\n⏱️ Loading all components...\n")

# ==================================================
# 1. FACE DETECTION (DAY 29)
# ==================================================
print("1️⃣ Loading face detection model...")

models_dir = Path("models")
prototxt_path = models_dir / "deploy.prototxt"
caffemodel_path = models_dir / "res10_300x300_ssd_iter_140000.caffemodel"

if prototxt_path.exists() and caffemodel_path.exists():
    face_net = cv2.dnn.readNetFromCaffe(
        str(prototxt_path),
        str(caffemodel_path)
    )
    CONFIDENCE_THRESHOLD = 0.5
    print("   ✅ Face detector loaded")
else:
    print("   ❌ Face detection models not found")
    face_net = None

# Load eye detector for alignment
eye_cascade_path = models_dir / "haarcascade_eye.xml"
if eye_cascade_path.exists():
    eye_cascade = cv2.CascadeClassifier(str(eye_cascade_path))
    print("   ✅ Eye detector loaded")
else:
    print("   ⚠️  Eye detector not found (alignment may be limited)")
    eye_cascade = None

# Face alignment function
def align_face(image, face_box):
    """Align face using eye detection."""
    (startX, startY, endX, endY) = face_box
    
    face_roi = image[startY:endY, startX:endX]
    if face_roi.shape[0] == 0 or face_roi.shape[1] == 0:
        return None
    
    if eye_cascade is not None:
        gray_face = cv2.cvtColor(face_roi, cv2.COLOR_BGR2GRAY)
        eyes = eye_cascade.detectMultiScale(gray_face, scaleFactor=1.1, minNeighbors=5, minSize=(20, 20))
        
        if len(eyes) >= 2:
            eyes = sorted(eyes, key=lambda x: x[0])
            left_eye = eyes[0]
            right_eye = eyes[1]
            
            left_eye_center = (startX + left_eye[0] + left_eye[2] // 2, startY + left_eye[1] + left_eye[3] // 2)
            right_eye_center = (startX + right_eye[0] + right_eye[2] // 2, startY + right_eye[1] + right_eye[3] // 2)
            
            dY = right_eye_center[1] - left_eye_center[1]
            dX = right_eye_center[0] - left_eye_center[0]
            angle = np.degrees(np.arctan2(dY, dX))
            
            eyes_center = (float((left_eye_center[0] + right_eye_center[0]) / 2),
                          float((left_eye_center[1] + right_eye_center[1]) / 2))
            
            M = cv2.getRotationMatrix2D(eyes_center, angle, 1.0)
            (h, w) = image.shape[:2]
            aligned = cv2.warpAffine(image, M, (w, h), flags=cv2.INTER_CUBIC)
            
            face_width = endX - startX
            face_height = endY - startY
            padding = int(0.3 * max(face_width, face_height))
            
            crop_startX = max(0, startX - padding)
            crop_startY = max(0, startY - padding)
            crop_endX = min(w, endX + padding)
            crop_endY = min(h, endY + padding)
            
            face_crop = aligned[crop_startY:crop_endY, crop_startX:crop_endX]
            
            if face_crop.shape[0] > 0 and face_crop.shape[1] > 0:
                return cv2.resize(face_crop, (160, 160))
    
    # Fallback
    face_width = endX - startX
    face_height = endY - startY
    padding = int(0.2 * max(face_width, face_height))
    
    crop_startX = max(0, startX - padding)
    crop_startY = max(0, startY - padding)
    crop_endX = min(image.shape[1], endX + padding)
    crop_endY = min(image.shape[0], endY + padding)
    
    face_crop = image[crop_startY:crop_endY, crop_startX:crop_endX]
    
    if face_crop.shape[0] > 0 and face_crop.shape[1] > 0:
        return cv2.resize(face_crop, (160, 160))
    
    return None

print("   ✅ Face alignment function loaded")

# ==================================================
# 2. FACE EMBEDDER (DAY 30)
# ==================================================
print("\n2️⃣ Loading face embedding model...")

class FaceEmbedder(nn.Module):
    def __init__(self, embedding_dim=512):
        super(FaceEmbedder, self).__init__()
        resnet = models.resnet18(pretrained=True)
        self.features = nn.Sequential(*list(resnet.children())[:-1])
        self.embedding = nn.Linear(512, embedding_dim)
    
    def forward(self, x):
        x = self.features(x)
        x = x.view(x.size(0), -1)
        x = self.embedding(x)
        x = nn.functional.normalize(x, p=2, dim=1)
        return x

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
embedding_model = FaceEmbedder(embedding_dim=512).to(device)
embedding_model.eval()

preprocess = transforms.Compose([
    transforms.ToPILImage(),
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
    transforms.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225])
])

def generate_embedding(face_image):
    """Generate 512-dim embedding."""
    try:
        face_rgb = cv2.cvtColor(face_image, cv2.COLOR_BGR2RGB)
        face_tensor = preprocess(face_rgb).unsqueeze(0).to(device)
        
        with torch.no_grad():
            embedding = embedding_model(face_tensor)
        
        return embedding.cpu().numpy().flatten()
    except:
        return None

print("   ✅ Face embedder loaded")

# ==================================================
# 3. FACE MATCHER & DATABASE (DAY 30)
# ==================================================
print("\n3️⃣ Loading face matcher and database...")

class FaceMatcher:
    """Face matcher from Day 30."""
    def __init__(self, similarity_threshold=0.6, use_average=True):
        self.similarity_threshold = similarity_threshold
        self.use_average = use_average
        self.database = {}
    
    def add_person(self, person_id, embeddings):
        if person_id not in self.database:
            self.database[person_id] = []
        if isinstance(embeddings, list):
            self.database[person_id].extend(embeddings)
        else:
            self.database[person_id].append(embeddings)
    
    def match_face(self, query_embedding):
        if len(self.database) == 0:
            return {'person_id': 'unknown', 'similarity': 0.0, 'is_match': False, 'all_scores': {}}
        
        person_scores = {}
        for person_id, person_embeddings in self.database.items():
            similarities = [np.dot(query_embedding, emb) for emb in person_embeddings]
            person_scores[person_id] = np.mean(similarities) if self.use_average else np.max(similarities)
        
        best_person = max(person_scores, key=person_scores.get)
        best_similarity = person_scores[best_person]
        is_match = best_similarity >= self.similarity_threshold
        
        return {
            'person_id': best_person if is_match else 'unknown',
            'similarity': best_similarity,
            'is_match': is_match,
            'all_scores': person_scores
        }

class FaceDatabase:
    """Face database from Day 30."""
    def __init__(self, database_dir="face_database"):
        self.database_dir = Path(database_dir)
        self.database_file = self.database_dir / "database.json"
        self.database = self._load_database()
    
    def _load_database(self):
        if self.database_file.exists():
            with open(self.database_file, 'r') as f:
                return json.load(f)
        return {"persons": {}, "stats": {}}
    
    def get_all_persons(self):
        return self.database["persons"]

# Load database and matcher
db = FaceDatabase("face_database")
matcher = FaceMatcher(similarity_threshold=0.6)

# Load embeddings into matcher
for person_id, person_data in db.get_all_persons().items():
    person_embeddings = []
    for face_entry in person_data['faces']:
        if face_entry.get('embedding') is not None:
            embedding = np.array(face_entry['embedding'])
            person_embeddings.append(embedding)
    
    if len(person_embeddings) > 0:
        matcher.add_person(person_id, person_embeddings)

print(f"   ✅ Face matcher loaded ({len(matcher.database)} persons)")

# ==================================================
# 4. CENTROID TRACKER (DAY 31)
# ==================================================
print("\n4️⃣ Loading centroid tracker...")

class CentroidTracker:
    """Centroid tracker from Day 31."""
    def __init__(self, max_disappeared=30, max_distance=50):
        self.nextObjectID = 0
        self.objects = OrderedDict()
        self.disappeared = OrderedDict()
        self.max_disappeared = max_disappeared
        self.max_distance = max_distance
        self.bboxes = OrderedDict()
    
    def register(self, centroid, bbox):
        self.objects[self.nextObjectID] = centroid
        self.bboxes[self.nextObjectID] = bbox
        self.disappeared[self.nextObjectID] = 0
        self.nextObjectID += 1
    
    def deregister(self, objectID):
        del self.objects[objectID]
        del self.bboxes[objectID]
        del self.disappeared[objectID]
    
    def update(self, detections):
        if len(detections) == 0:
            for objectID in list(self.disappeared.keys()):
                self.disappeared[objectID] += 1
                if self.disappeared[objectID] > self.max_disappeared:
                    self.deregister(objectID)
            return self.get_objects()
        
        input_centroids = np.zeros((len(detections), 2), dtype="int")
        input_bboxes = []
        
        for i, (startX, startY, endX, endY) in enumerate(detections):
            cX = int((startX + endX) / 2.0)
            cY = int((startY + endY) / 2.0)
            input_centroids[i] = (cX, cY)
            input_bboxes.append((startX, startY, endX, endY))
        
        if len(self.objects) == 0:
            for i in range(len(input_centroids)):
                self.register(input_centroids[i], input_bboxes[i])
        else:
            objectIDs = list(self.objects.keys())
            object_centroids = list(self.objects.values())
            
            D = dist.cdist(np.array(object_centroids), input_centroids)
            rows = D.min(axis=1).argsort()
            cols = D.argmin(axis=1)[rows]
            
            used_rows = set()
            used_cols = set()
            
            for (row, col) in zip(rows, cols):
                if row in used_rows or col in used_cols:
                    continue
                if D[row, col] > self.max_distance:
                    continue
                
                objectID = objectIDs[row]
                self.objects[objectID] = input_centroids[col]
                self.bboxes[objectID] = input_bboxes[col]
                self.disappeared[objectID] = 0
                
                used_rows.add(row)
                used_cols.add(col)
            
            unused_rows = set(range(D.shape[0])) - used_rows
            for row in unused_rows:
                objectID = objectIDs[row]
                self.disappeared[objectID] += 1
                if self.disappeared[objectID] > self.max_disappeared:
                    self.deregister(objectID)
            
            unused_cols = set(range(D.shape[1])) - used_cols
            for col in unused_cols:
                self.register(input_centroids[col], input_bboxes[col])
        
        return self.get_objects()
    
    def get_objects(self):
        result = OrderedDict()
        for objectID in self.objects.keys():
            result[objectID] = (self.objects[objectID], self.bboxes[objectID])
        return result

print("   ✅ Centroid tracker loaded")

print("\n✅ All components loaded successfully!")
print(f"\n📊 System Status:")
print(f"   Face detection: {'✅' if face_net else '❌'}")
print(f"   Face alignment: {'✅' if eye_cascade else '⚠️'}")
print(f"   Face embedder: ✅")
print(f"   Face matcher: ✅ ({len(matcher.database)} persons)")
print(f"   Centroid tracker: ✅")

print("\n✅ Exercise 1.2 Complete!")
print("=" * 80)
````

````python
# ==================================================
# EXERCISE 1.3: INTEGRATED PIPELINE CLASS
# ==================================================

print("\n" + "=" * 80)
print("EXERCISE 1.3: Integrated Pipeline Class")
print("=" * 80)

"""
📖 THEORY: Complete Integration

Bring all components together in single pipeline:
- Face detection
- Tracking
- Recognition
- Alert triggering
- Performance monitoring
"""

class IntegratedSecurityPipeline:
    """
    Complete integrated security & safety pipeline.
    """
    
    def __init__(self, face_net, matcher, alert_enabled=False):
        """
        Initialize integrated pipeline.
        
        Args:
            face_net: Face detection network
            matcher: FaceMatcher instance
            alert_enabled: Enable alert generation
        """
        self.face_net = face_net
        self.matcher = matcher
        self.alert_enabled = alert_enabled
        
        # Initialize tracker
        self.tracker = CentroidTracker(max_disappeared=30, max_distance=50)
        
        # Recognition tracking
        self.track_to_person = {}
        self.unknown_counter = 0
        self.recognition_cache = {}
        self.recognition_interval = 10
        self.frame_count = 0
        
        # Performance tracking
        self.fps_deque = deque(maxlen=30)
        self.detection_times = []
        self.recognition_times = []
        
        # Statistics
        self.stats = {
            'total_frames': 0,
            'faces_detected': 0,
            'known_persons': 0,
            'unknown_persons': 0,
            'unique_tracks': set()
        }
    
    def process_frame(self, frame):
        """
        Process single frame through complete pipeline.
        
        Args:
            frame: Input BGR frame
        
        Returns:
            annotated_frame, results_dict
        """
        start_time = time.time()
        
        self.frame_count += 1
        self.stats['total_frames'] += 1
        
        results = {
            'faces': [],
            'tracks': [],
            'fps': 0,
            'detection_time': 0,
            'recognition_time': 0
        }
        
        # Step 1: Face Detection
        detection_start = time.time()
        face_boxes = self._detect_faces(frame)
        detection_time = time.time() - detection_start
        self.detection_times.append(detection_time)
        results['detection_time'] = detection_time * 1000  # ms
        
        self.stats['faces_detected'] += len(face_boxes)
        
        # Step 2: Tracking
        tracked_objects = self.tracker.update(face_boxes)
        self.stats['unique_tracks'].update(tracked_objects.keys())
        
        # Step 3: Recognition (every N frames)
        recognition_start = time.time()
        run_recognition = (self.frame_count % self.recognition_interval == 0)
        
        for track_id, (centroid, bbox) in tracked_objects.items():
            person_id = "unknown"
            person_name = "Unknown"
            similarity = 0.0
            
            if run_recognition or track_id not in self.track_to_person:
                aligned_face = align_face(frame, bbox)
                
                if aligned_face is not None:
                    embedding = generate_embedding(aligned_face)
                    
                    if embedding is not None:
                        match_result = self.matcher.match_face(embedding)
                        
                        if match_result['is_match']:
                            person_id = match_result['person_id']
                            similarity = match_result['similarity']
                            
                            person_data = db.get_all_persons().get(person_id)
                            person_name = person_data['name'] if person_data else "Unknown"
                            
                            self.stats['known_persons'] += 1
                        else:
                            if track_id not in self.track_to_person:
                                self.unknown_counter += 1
                                person_id = f"unknown_{self.unknown_counter}"
                                self.stats['unknown_persons'] += 1
                            else:
                                person_id = self.track_to_person[track_id]
                            
                            person_name = f"Unknown #{person_id.split('_')[1]}"
                            similarity = match_result['similarity']
                        
                        self.track_to_person[track_id] = person_id
                        self.recognition_cache[track_id] = (person_id, similarity, 0)
            
            elif track_id in self.recognition_cache:
                person_id, similarity, frames_since = self.recognition_cache[track_id]
                self.recognition_cache[track_id] = (person_id, similarity, frames_since + 1)
                
                if person_id.startswith("unknown_"):
                    person_name = f"Unknown #{person_id.split('_')[1]}"
                else:
                    person_data = db.get_all_persons().get(person_id)
                    person_name = person_data['name'] if person_data else "Unknown"
            
            # Store result
            results['faces'].append({
                'track_id': track_id,
                'person_id': person_id,
                'person_name': person_name,
                'similarity': similarity,
                'bbox': bbox,
                'centroid': centroid
            })
        
        recognition_time = time.time() - recognition_start
        self.recognition_times.append(recognition_time)
        results['recognition_time'] = recognition_time * 1000  # ms
        
        # Calculate FPS
        frame_time = time.time() - start_time
        fps = 1.0 / frame_time if frame_time > 0 else 0
        self.fps_deque.append(fps)
        results['fps'] = np.mean(self.fps_deque)
        
        # Annotate frame
        annotated_frame = self._annotate_frame(frame.copy(), results)
        
        return annotated_frame, results
    
    def _detect_faces(self, frame):
        """Detect faces in frame."""
        (h, w) = frame.shape[:2]
        blob = cv2.dnn.blobFromImage(
            cv2.resize(frame, (300, 300)),
            1.0,
            (300, 300),
            (104.0, 177.0, 123.0)
        )
        
        self.face_net.setInput(blob)
        detections = self.face_net.forward()
        
        face_boxes = []
        for i in range(detections.shape[2]):
            confidence = detections[0, 0, i, 2]
            
            if confidence > CONFIDENCE_THRESHOLD:
                box = detections[0, 0, i, 3:7] * np.array([w, h, w, h])
                (startX, startY, endX, endY) = box.astype("int")
                
                startX = max(0, startX)
                startY = max(0, startY)
                endX = min(w, endX)
                endY = min(h, endY)
                
                face_boxes.append((startX, startY, endX, endY))
        
        return face_boxes
    
    def _annotate_frame(self, frame, results):
        """Annotate frame with detections and info."""
        for face in results['faces']:
            bbox = face['bbox']
            person_name = face['person_name']
            similarity = face['similarity']
            
            # Draw bounding box
            color = (0, 255, 0) if not person_name.startswith("Unknown") else (0, 0, 255)
            cv2.rectangle(frame, (bbox[0], bbox[1]), (bbox[2], bbox[3]), color, 2)
            
            # Draw label
            label = f"{person_name} ({similarity:.2f})"
            cv2.putText(frame, label, (bbox[0], bbox[1] - 10),
                       cv2.FONT_HERSHEY_SIMPLEX, 0.6, color, 2)
        
        # Draw FPS and stats
        fps_text = f"FPS: {results['fps']:.1f}"
        cv2.putText(frame, fps_text, (10, 30),
                   cv2.FONT_HERSHEY_SIMPLEX, 1, (0, 255, 0), 2)
        
        people_text = f"People: {len(results['faces'])}"
        cv2.putText(frame, people_text, (10, 70),
                   cv2.FONT_HERSHEY_SIMPLEX, 1, (0, 255, 0), 2)
        
        return frame
    
    def get_statistics(self):
        """Get pipeline statistics."""
        return {
            'total_frames': self.stats['total_frames'],
            'faces_detected': self.stats['faces_detected'],
            'known_persons': self.stats['known_persons'],
            'unknown_persons': self.stats['unknown_persons'],
            'unique_tracks': len(self.stats['unique_tracks']),
            'avg_fps': np.mean(self.fps_deque) if self.fps_deque else 0,
            'avg_detection_time_ms': np.mean(self.detection_times) * 1000 if self.detection_times else 0,
            'avg_recognition_time_ms': np.mean(self.recognition_times) * 1000 if self.recognition_times else 0
        }

print("\n✅ IntegratedSecurityPipeline class implemented!")
print("\n📊 Features:")
print("   - Complete pipeline: Detection → Tracking → Recognition")
print("   - Performance monitoring (FPS, timings)")
print("   - Frame annotation")
print("   - Statistics tracking")
print("   - Real-time processing")

print("\n✅ Exercise 1.3 Complete!")
print("=" * 80)
````

````python
# ==================================================
# EXERCISE 2.2: BENCHMARK SUITE
# ==================================================

print("\n" + "=" * 80)
print("EXERCISE 2.2: Benchmark Suite")
print("=" * 80)

"""
📖 THEORY: Systematic Benchmarking

Create benchmark suite to measure:
- FPS under load
- Component timing
- Accuracy metrics
"""

class PerformanceBenchmark:
    """
    Performance benchmarking suite.
    """
    
    def __init__(self, pipeline):
        """
        Initialize benchmark.
        
        Args:
            pipeline: IntegratedSecurityPipeline instance
        """
        self.pipeline = pipeline
        self.results = {
            'fps_tests': [],
            'latency_tests': [],
            'accuracy_tests': []
        }
    
    def benchmark_fps(self, frame, num_frames=100):
        """
        Benchmark FPS over multiple frames.
        
        Args:
            frame: Test frame
            num_frames: Number of frames to process
        
        Returns:
            dict with FPS statistics
        """
        print(f"\n⏱️  Benchmarking FPS over {num_frames} frames...")
        
        fps_values = []
        start_time = time.time()
        
        for i in range(num_frames):
            frame_start = time.time()
            _, results = self.pipeline.process_frame(frame)
            frame_time = time.time() - frame_start
            
            fps = 1.0 / frame_time if frame_time > 0 else 0
            fps_values.append(fps)
            
            if (i + 1) % 20 == 0:
                print(f"   Progress: {i+1}/{num_frames} frames")
        
        total_time = time.time() - start_time
        
        benchmark_result = {
            'num_frames': num_frames,
            'total_time': total_time,
            'avg_fps': np.mean(fps_values),
            'min_fps': np.min(fps_values),
            'max_fps': np.max(fps_values),
            'std_fps': np.std(fps_values),
            'median_fps': np.median(fps_values)
        }
        
        self.results['fps_tests'].append(benchmark_result)
        
        return benchmark_result
    
    def benchmark_component_timing(self, frame, num_iterations=50):
        """
        Benchmark individual component timings.
        
        Args:
            frame: Test frame
            num_iterations: Number of iterations
        
        Returns:
            dict with component timings
        """
        print(f"\n⏱️  Benchmarking component timings over {num_iterations} iterations...")
        
        detection_times = []
        recognition_times = []
        total_times = []
        
        for i in range(num_iterations):
            start = time.time()
            _, results = self.pipeline.process_frame(frame)
            total_time = time.time() - start
            
            detection_times.append(results['detection_time'])
            recognition_times.append(results['recognition_time'])
            total_times.append(total_time * 1000)  # ms
        
        timing_result = {
            'num_iterations': num_iterations,
            'detection': {
                'avg_ms': np.mean(detection_times),
                'min_ms': np.min(detection_times),
                'max_ms': np.max(detection_times),
                'std_ms': np.std(detection_times)
            },
            'recognition': {
                'avg_ms': np.mean(recognition_times),
                'min_ms': np.min(recognition_times),
                'max_ms': np.max(recognition_times),
                'std_ms': np.std(recognition_times)
            },
            'total': {
                'avg_ms': np.mean(total_times),
                'min_ms': np.min(total_times),
                'max_ms': np.max(total_times),
                'std_ms': np.std(total_times)
            }
        }
        
        self.results['latency_tests'].append(timing_result)
        
        return timing_result
    
    def generate_report(self):
        """Generate benchmark report."""
        report = []
        report.append("\n" + "=" * 80)
        report.append("📊 PERFORMANCE BENCHMARK REPORT")
        report.append("=" * 80)
        
        # FPS Tests
        if self.results['fps_tests']:
            report.append("\n🚀 FPS BENCHMARKS:")
            for i, test in enumerate(self.results['fps_tests'], 1):
                report.append(f"\n   Test {i}:")
                report.append(f"      Frames processed: {test['num_frames']}")
                report.append(f"      Total time: {test['total_time']:.2f}s")
                report.append(f"      Average FPS: {test['avg_fps']:.1f}")
                report.append(f"      Min FPS: {test['min_fps']:.1f}")
                report.append(f"      Max FPS: {test['max_fps']:.1f}")
                report.append(f"      Std Dev: {test['std_fps']:.1f}")
                report.append(f"      Median FPS: {test['median_fps']:.1f}")
        
        # Latency Tests
        if self.results['latency_tests']:
            report.append("\n⏱️  LATENCY BENCHMARKS:")
            for i, test in enumerate(self.results['latency_tests'], 1):
                report.append(f"\n   Test {i}:")
                report.append(f"      Iterations: {test['num_iterations']}")
                report.append(f"\n      Detection:")
                report.append(f"         Avg: {test['detection']['avg_ms']:.1f} ms")
                report.append(f"         Min: {test['detection']['min_ms']:.1f} ms")
                report.append(f"         Max: {test['detection']['max_ms']:.1f} ms")
                report.append(f"\n      Recognition:")
                report.append(f"         Avg: {test['recognition']['avg_ms']:.1f} ms")
                report.append(f"         Min: {test['recognition']['min_ms']:.1f} ms")
                report.append(f"         Max: {test['recognition']['max_ms']:.1f} ms")
                report.append(f"\n      Total Pipeline:")
                report.append(f"         Avg: {test['total']['avg_ms']:.1f} ms")
                report.append(f"         Min: {test['total']['min_ms']:.1f} ms")
                report.append(f"         Max: {test['total']['max_ms']:.1f} ms")
        
        report.append("\n" + "=" * 80)
        
        return "\n".join(report)

print("\n✅ PerformanceBenchmark class implemented!")
print("\n📊 Features:")
print("   - FPS benchmarking")
print("   - Component timing analysis")
print("   - Statistical metrics (avg, min, max, std)")
print("   - Comprehensive reporting")

print("\n✅ Exercise 2.2 Complete!")
print("=" * 80)
````
