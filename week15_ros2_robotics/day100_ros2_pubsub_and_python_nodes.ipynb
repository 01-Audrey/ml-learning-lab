```python
# ==================================================
# EXERCISE 1.1: PUB/SUB DEEP DIVE
# ==================================================

print("\n" + "=" * 80)
print("EXERCISE 1.1: Publisher/Subscriber Pattern Deep Dive")
print("=" * 80)

"""
📖 THEORY: Publisher/Subscriber

The pub/sub pattern is the backbone of ROS2 communication.
It's an asynchronous, decoupled messaging pattern where:
- Publishers send data WITHOUT knowing who receives it
- Subscribers receive data WITHOUT knowing who sent it
- Topics are the named channels that connect them

This is different from direct function calls — there's no
tight coupling between the sender and receiver.
"""

print("""
📋 PUB/SUB vs REQUEST/RESPONSE:

  Pub/Sub (Topics):                Request/Response (Services):
  ──────────────────────────────   ────────────────────────────────
  Continuous data stream           One-time call
  Fire and forget                  Wait for response
  Many-to-many                     One-to-one
  Asynchronous                     Synchronous
  No feedback                      Gets result back

  Use topics for:                  Use services for:
  • Sensor readings (camera, lidar)• Commands (spawn, reset, teleport)
  • Robot state (pose, velocity)   • Configuration changes
  • Continuous control commands    • Queries (get current state)
  • Detection results              • Initialization actions


📋 TOPIC LIFECYCLE:
─────────────────────────────────────────────────────────────

  1. Publisher node created
     → Declares: "I will publish geometry_msgs/Twist on /cmd_vel"

  2. Subscriber node created
     → Declares: "I want to receive geometry_msgs/Twist from /cmd_vel"

  3. ROS2 DDS (middleware) matches them automatically
     → No central broker needed (unlike MQTT or Redis pub/sub)
     → Works across machines on the same network!

  4. Publisher sends messages at its own rate
     → Subscriber's callback fires for each message received
     → If publisher is faster than subscriber: messages may be dropped
       (controlled by QoS - Quality of Service settings)

  5. Either node can stop at any time
     → Other nodes continue without crashing
     → Decoupling = resilience


📋 QoS (QUALITY OF SERVICE):
─────────────────────────────────────────────────────────────
  QoS controls how messages are delivered. Key settings:

  queue_size (depth):
    How many messages to buffer if subscriber is slow
    10 = keep last 10 messages (good default)
    1  = only keep latest (good for real-time sensor data)

  Reliability:
    RELIABLE   = guarantee delivery (like TCP) — slower
    BEST_EFFORT = drop if busy (like UDP) — faster, for sensors

  History:
    KEEP_LAST  = keep last N messages (most common)
    KEEP_ALL   = keep everything (careful: uses lots of memory)

  For TurtleSim: queue_size=10, RELIABLE is fine.
  For a real camera at 60fps: queue_size=1, BEST_EFFORT is better.
""")

print("\n✅ Exercise 1.1 Complete!")
print("=" * 80)
```

```python
# ==================================================
# EXERCISE 2.1: YOUR FIRST PUBLISHER NODE
# ==================================================

print("\n" + "=" * 80)
print("EXERCISE 2.1: First Publisher Node — Turtle Mover")
print("=" * 80)

"""
📖 THEORY: ROS2 Python Node Structure

Every ROS2 Python node follows this structure:
1. Import rclpy and message types
2. Create a class that inherits from Node
3. In __init__: create publishers, subscribers, timers
4. Define callbacks for timers and subscribers
5. In main: init rclpy, create node, spin, shutdown
"""

turtle_mover_py = '''
# ============================================================
# turtle_mover.py — Publisher Node
# Publishes velocity commands to move the turtle
#
# Save to: ~/ros2_ws/src/turtle_mover.py
# Run with: python3 turtle_mover.py
# ============================================================

import rclpy
from rclpy.node import Node
from geometry_msgs.msg import Twist
import math


class TurtleMover(Node):
    """
    Publisher node that moves the turtle in different patterns.
    Publishes geometry_msgs/Twist to /turtle1/cmd_vel
    """

    def __init__(self):
        # Initialize the node with a name
        super().__init__("turtle_mover")

        # Create publisher
        # Arguments: message_type, topic_name, queue_size
        self.publisher = self.create_publisher(
            Twist,
            "/turtle1/cmd_vel",
            10,
        )

        # Create a timer — calls self.move_callback every 0.1 seconds (10 Hz)
        self.timer = self.create_timer(0.1, self.move_callback)

        # State variables
        self.step = 0
        self.pattern = "circle"   # "circle", "square", "figure8"

        self.get_logger().info(
            f"TurtleMover started — pattern: {self.pattern}"
        )

    def move_callback(self):
        """Called every 0.1 seconds by the timer."""
        msg = Twist()

        if self.pattern == "circle":
            msg = self.circle_pattern()
        elif self.pattern == "square":
            msg = self.square_pattern()
        elif self.pattern == "figure8":
            msg = self.figure8_pattern()

        self.publisher.publish(msg)
        self.step += 1

    def circle_pattern(self) -> Twist:
        """Move in a continuous circle."""
        msg = Twist()
        msg.linear.x = 2.0
        msg.angular.z = 1.0
        return msg

    def square_pattern(self) -> Twist:
        """
        Move in a square: forward for 2 seconds, turn for 1 second.
        Timer runs at 10Hz → 20 steps = 2 seconds, 10 steps = 1 second
        """
        msg = Twist()
        phase = self.step % 30      # 30 steps = one side + one turn

        if phase < 20:              # 0-19: move forward (2 seconds)
            msg.linear.x = 2.0
            msg.angular.z = 0.0
        else:                       # 20-29: turn left (1 second)
            msg.linear.x = 0.0
            msg.angular.z = math.pi / 2   # 90 degrees per second

        return msg

    def figure8_pattern(self) -> Twist:
        """Move in a figure-8 using alternating circles."""
        msg = Twist()
        phase = self.step % 200     # 200 steps = full figure-8

        msg.linear.x = 2.0
        if phase < 100:
            msg.angular.z = 1.0    # Left circle
        else:
            msg.angular.z = -1.0   # Right circle

        return msg

    def stop(self):
        """Publish zero velocity to stop the turtle."""
        msg = Twist()               # All zeros by default
        self.publisher.publish(msg)
        self.get_logger().info("Turtle stopped")


def main(args=None):
    rclpy.init(args=args)           # Initialize ROS2

    node = TurtleMover()

    try:
        rclpy.spin(node)            # Keep node running, process callbacks
    except KeyboardInterrupt:
        node.stop()                 # Stop turtle on Ctrl+C
    finally:
        node.destroy_node()
        rclpy.shutdown()


if __name__ == "__main__":
    main()
'''

print("📄 turtle_mover.py:")
print(turtle_mover_py)

print("""
📋 HOW TO RUN:
─────────────────────────────────────────────────────────────
  Terminal 1: ros2 run turtlesim turtlesim_node
  Terminal 2: python3 ~/ros2_ws/src/turtle_mover.py

  To change pattern, edit self.pattern = "square" or "figure8"
  Press Ctrl+C in Terminal 2 to stop

📋 KEY PARTS EXPLAINED:

  super().__init__("turtle_mover")
    → Registers this node with ROS2 under the name "turtle_mover"
    → ros2 node list will now show /turtle_mover

  self.create_publisher(Twist, "/turtle1/cmd_vel", 10)
    → Twist = message type
    → "/turtle1/cmd_vel" = topic name
    → 10 = queue size (buffer last 10 messages)

  self.create_timer(0.1, self.move_callback)
    → Call move_callback every 0.1 seconds (10 Hz)
    → Timer is how you publish at a regular rate

  rclpy.spin(node)
    → Keeps the node alive and processes timer/subscriber callbacks
    → Without spin(), the node exits immediately
""")

print("\n✅ Exercise 2.1 Complete!")
print("=" * 80)
```

```python
# ==================================================
# EXERCISE 2.2: YOUR FIRST SUBSCRIBER NODE
# ==================================================

print("\n" + "=" * 80)
print("EXERCISE 2.2: First Subscriber Node — Pose Listener")
print("=" * 80)

"""
📖 THEORY: Subscribers

A subscriber registers a callback function that fires
automatically whenever a new message arrives on the topic.
The callback receives the message as its argument.
"""

pose_listener_py = '''
# ============================================================
# pose_listener.py — Subscriber Node
# Subscribes to turtle pose and prints position info
#
# Save to: ~/ros2_ws/src/pose_listener.py
# Run with: python3 pose_listener.py
# ============================================================

import rclpy
from rclpy.node import Node
from turtlesim.msg import Pose
import math


class PoseListener(Node):
    """
    Subscriber node that listens to turtle position.
    Subscribes to turtlesim/Pose on /turtle1/pose
    """

    def __init__(self):
        super().__init__("pose_listener")

        # Create subscriber
        # Arguments: message_type, topic_name, callback, queue_size
        self.subscription = self.create_subscription(
            Pose,
            "/turtle1/pose",
            self.pose_callback,
            10,
        )

        # Track stats
        self.message_count = 0
        self.last_x = None
        self.last_y = None
        self.total_distance = 0.0

        self.get_logger().info("PoseListener started — watching turtle position")

    def pose_callback(self, msg: Pose):
        """
        Called automatically every time a new Pose message arrives.
        TurtleSim publishes pose at ~62 Hz.
        """
        self.message_count += 1

        # Calculate distance traveled since last message
        if self.last_x is not None:
            dx = msg.x - self.last_x
            dy = msg.y - self.last_y
            distance = math.sqrt(dx**2 + dy**2)
            self.total_distance += distance

        self.last_x = msg.x
        self.last_y = msg.y

        # Convert theta to degrees for readability
        heading_deg = math.degrees(msg.theta)

        # Log every 30th message (avoid flooding the terminal at 62Hz)
        if self.message_count % 30 == 0:
            self.get_logger().info(
                f"Position: ({msg.x:.2f}, {msg.y:.2f}) | "
                f"Heading: {heading_deg:.1f}° | "
                f"Speed: {msg.linear_velocity:.2f} | "
                f"Distance traveled: {self.total_distance:.2f}"
            )

        # Boundary warning — TurtleSim canvas is ~11.08 x 11.08
        BOUNDARY_MARGIN = 1.5
        if (msg.x < BOUNDARY_MARGIN or msg.x > 11.08 - BOUNDARY_MARGIN or
                msg.y < BOUNDARY_MARGIN or msg.y > 11.08 - BOUNDARY_MARGIN):
            self.get_logger().warn(
                f"⚠️  Turtle near boundary! Position: ({msg.x:.2f}, {msg.y:.2f})"
            )


def main(args=None):
    rclpy.init(args=args)

    node = PoseListener()

    try:
        rclpy.spin(node)
    except KeyboardInterrupt:
        node.get_logger().info(
            f"Shutting down. Total messages received: {node.message_count}"
        )
    finally:
        node.destroy_node()
        rclpy.shutdown()


if __name__ == "__main__":
    main()
'''

print("📄 pose_listener.py:")
print(pose_listener_py)

print("""
📋 HOW TO RUN (3 terminals):
─────────────────────────────────────────────────────────────
  Terminal 1: ros2 run turtlesim turtlesim_node
  Terminal 2: python3 ~/ros2_ws/src/turtle_mover.py
  Terminal 3: python3 ~/ros2_ws/src/pose_listener.py

  Terminal 3 will show live position updates as the turtle moves.

📋 KEY PARTS EXPLAINED:

  self.create_subscription(Pose, "/turtle1/pose", self.pose_callback, 10)
    → Pose = message type we expect
    → "/turtle1/pose" = topic to subscribe to
    → self.pose_callback = function called on each new message
    → 10 = queue size

  def pose_callback(self, msg: Pose):
    → msg is automatically populated with the received Pose data
    → msg.x, msg.y, msg.theta, msg.linear_velocity, msg.angular_velocity

  self.message_count % 30 == 0
    → Throttle logging — pose publishes at ~62Hz
    → Printing every message would flood the terminal
    → Every 30th message = ~2 updates per second in the log
""")

print("\n✅ Exercise 2.2 Complete!")
print("=" * 80)
```

```python
# ==================================================
# EXERCISE 2.3: COMBINED NODE — PUBLISHER + SUBSCRIBER
# ==================================================

print("\n" + "=" * 80)
print("EXERCISE 2.3: Combined Node — Publisher + Subscriber")
print("=" * 80)

"""
📖 THEORY: Single Node with Both

A single node can be both a publisher AND a subscriber.
This is very common — most robot nodes receive sensor data
(subscriber) and send commands (publisher) in the same node.

Example: A controller node that:
- Subscribes to /turtle1/pose (where is the turtle?)
- Publishes to /turtle1/cmd_vel (move the turtle)
"""

smart_mover_py = '''
# ============================================================
# smart_mover.py — Publisher + Subscriber Combined
# Reads turtle pose and makes decisions based on position
#
# Save to: ~/ros2_ws/src/smart_mover.py
# Run with: python3 smart_mover.py
# ============================================================

import rclpy
from rclpy.node import Node
from geometry_msgs.msg import Twist
from turtlesim.msg import Pose
import math


class SmartMover(Node):
    """
    Combined publisher + subscriber node.
    Moves the turtle forward but turns away from boundaries.
    This is a simplified version of what Day 104 will build fully.
    """

    CANVAS_MIN = 1.0
    CANVAS_MAX = 10.08
    BOUNDARY_MARGIN = 2.0      # Start turning when within 2 units of wall
    FORWARD_SPEED = 2.0
    TURN_SPEED = 1.5

    def __init__(self):
        super().__init__("smart_mover")

        # Publisher — sends velocity commands
        self.cmd_publisher = self.create_publisher(
            Twist,
            "/turtle1/cmd_vel",
            10,
        )

        # Subscriber — reads turtle position
        self.pose_subscription = self.create_subscription(
            Pose,
            "/turtle1/pose",
            self.pose_callback,
            10,
        )

        # Timer — runs control loop at 10 Hz
        self.timer = self.create_timer(0.1, self.control_loop)

        # State
        self.current_pose = None
        self.near_boundary = False

        self.get_logger().info("SmartMover started — boundary-aware movement")

    def pose_callback(self, msg: Pose):
        """Update current pose whenever a new message arrives."""
        self.current_pose = msg

        # Check boundary proximity
        near_x = (msg.x < self.CANVAS_MIN + self.BOUNDARY_MARGIN or
                  msg.x > self.CANVAS_MAX - self.BOUNDARY_MARGIN)
        near_y = (msg.y < self.CANVAS_MIN + self.BOUNDARY_MARGIN or
                  msg.y > self.CANVAS_MAX - self.BOUNDARY_MARGIN)

        if (near_x or near_y) and not self.near_boundary:
            self.get_logger().warn(
                f"⚠️  Approaching boundary at ({msg.x:.1f}, {msg.y:.1f})"
            )

        self.near_boundary = near_x or near_y

    def control_loop(self):
        """
        Main control logic — runs at 10 Hz via timer.
        Decision: move forward normally, or turn away from boundary.
        """
        if self.current_pose is None:
            return                  # Wait until we have pose data

        msg = Twist()

        if self.near_boundary:
            # Turn away from boundary — stop forward, rotate
            msg.linear.x = 0.0
            msg.angular.z = self.TURN_SPEED
            self.get_logger().info("🔄 Turning away from boundary")
        else:
            # Safe — move forward
            msg.linear.x = self.FORWARD_SPEED
            msg.angular.z = 0.0

        self.cmd_publisher.publish(msg)


def main(args=None):
    rclpy.init(args=args)
    node = SmartMover()

    try:
        rclpy.spin(node)
    except KeyboardInterrupt:
        # Stop turtle on exit
        stop_msg = Twist()
        node.cmd_publisher.publish(stop_msg)
        node.get_logger().info("SmartMover stopped")
    finally:
        node.destroy_node()
        rclpy.shutdown()


if __name__ == "__main__":
    main()
'''

print("📄 smart_mover.py:")
print(smart_mover_py)

print("""
📋 HOW TO RUN:
─────────────────────────────────────────────────────────────
  Terminal 1: ros2 run turtlesim turtlesim_node
  Terminal 2: python3 ~/ros2_ws/src/smart_mover.py

  Watch the turtle move forward and turn away from walls
  automatically — no keyboard needed!

📋 KEY PATTERN — Subscriber + Timer:

  pose_callback() → called at ~62 Hz (whenever pose arrives)
    → Updates self.current_pose and self.near_boundary

  control_loop()  → called at 10 Hz (timer)
    → Reads self.current_pose and self.near_boundary
    → Makes decision and publishes command

  This separation is important:
  → Subscriber updates state as fast as data arrives
  → Control loop runs at a fixed, predictable rate
  → Never do heavy computation inside a subscriber callback!
""")

print("\n✅ Exercise 2.3 Complete!")
print("=" * 80)
```
