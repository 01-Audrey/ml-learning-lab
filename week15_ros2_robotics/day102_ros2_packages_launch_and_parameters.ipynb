```python
# ==================================================
# EXERCISE 1.1: WHAT IS A ROS2 PACKAGE?
# ==================================================

print("\n" + "=" * 80)
print("EXERCISE 1.1: What is a ROS2 Package?")
print("=" * 80)

"""
📖 THEORY: ROS2 Packages

A package is the fundamental unit of organization in ROS2.
It's a directory containing:
- Python nodes (executable scripts)
- Launch files (startup configurations)
- Configuration files (parameters, YAML)
- Resource files (models, maps, images)
- Metadata (package.xml, setup.py)

Why packages matter:
- Shareable: publish to GitHub, others can install with apt/pip
- Buildable: colcon build compiles and installs everything
- Discoverable: ros2 run <package> <node> finds your nodes
- Namespaced: avoids conflicts with other packages
"""

print("""
📋 PACKAGE STRUCTURE (Python):
─────────────────────────────────────────────────────────────

  my_package/
  ├── my_package/               ← Python module (same name as package)
  │   ├── __init__.py           ← Makes it a Python package
  │   ├── turtle_mover.py       ← Node 1
  │   ├── pose_listener.py      ← Node 2
  │   └── smart_mover.py        ← Node 3
  │
  ├── launch/                   ← Launch files
  │   └── turtlesim.launch.py
  │
  ├── config/                   ← Parameter files
  │   └── params.yaml
  │
  ├── resource/                 ← Package marker (required)
  │   └── my_package
  │
  ├── test/                     ← Unit tests
  │   └── test_nodes.py
  │
  ├── package.xml               ← Package metadata & dependencies
  ├── setup.py                  ← Python build config
  └── setup.cfg                 ← Python package config

📋 TWO TYPES OF ROS2 PACKAGES:
─────────────────────────────────────────────────────────────
  ament_python   → Pure Python nodes (what we use)
  ament_cmake    → C++ nodes or mixed Python/C++

  For ML/robotics with Python: always use ament_python.
""")

print("\n✅ Exercise 1.1 Complete!")
print("=" * 80)
```

```python
# ==================================================
# EXERCISE 1.2: CREATING THE PACKAGE
# ==================================================

print("\n" + "=" * 80)
print("EXERCISE 1.2: Creating the TurtleSim Controller Package")
print("=" * 80)

"""
📖 THEORY: ros2 pkg create

The ros2 pkg create command generates the full package
structure with all required files automatically.
"""

print("""
📋 CREATE THE PACKAGE:
─────────────────────────────────────────────────────────────
  # Navigate to workspace source directory
  cd ~/ros2_ws/src

  # Create package
  ros2 pkg create \\
    --build-type ament_python \\
    --dependencies rclpy geometry_msgs turtlesim std_srvs \\
    turtle_controller

  # This generates:
  # turtle_controller/
  # ├── turtle_controller/
  # │   └── __init__.py
  # ├── resource/
  # │   └── turtle_controller
  # ├── test/
  # │   ├── test_copyright.py
  # │   └── test_pep257.py
  # ├── package.xml
  # ├── setup.cfg
  # └── setup.py

  # Create additional directories we need
  cd turtle_controller
  mkdir launch config

  # Verify structure
  ls -la
""")

package_xml = '''
<!-- ============================================================
     package.xml — Package metadata
     Located at: ~/ros2_ws/src/turtle_controller/package.xml
     ============================================================ -->

<?xml version="1.0"?>
<?xml-model href="http://download.ros.org/schema/package_format3.xsd"
  schematypens="http://www.w3.org/2001/XMLSchema"?>
<package format="3">
  <name>turtle_controller</name>
  <version>0.1.0</version>
  <description>
    TurtleSim Controller package — Week 15 of ML Journey.
    Demonstrates ROS2 pub/sub, services, actions, and launch files.
  </description>
  <maintainer email="audrey@example.com">Audrey</maintainer>
  <license>MIT</license>

  <depend>rclpy</depend>
  <depend>geometry_msgs</depend>
  <depend>turtlesim</depend>
  <depend>std_srvs</depend>

  <test_depend>ament_copyright</test_depend>
  <test_depend>ament_flake8</test_depend>
  <test_depend>ament_pep257</test_depend>
  <test_depend>python3-pytest</test_depend>

  <export>
    <build_type>ament_python</build_type>
  </export>
</package>
'''

setup_py = '''
# ============================================================
# setup.py — Python build configuration
# Located at: ~/ros2_ws/src/turtle_controller/setup.py
# ============================================================

from setuptools import setup, find_packages
import os
from glob import glob

package_name = "turtle_controller"

setup(
    name=package_name,
    version="0.1.0",
    packages=find_packages(exclude=["test"]),
    data_files=[
        # Required: register package with ROS2
        ("share/ament_index/resource_index/packages",
         ["resource/" + package_name]),
        ("share/" + package_name, ["package.xml"]),

        # Install launch files
        (os.path.join("share", package_name, "launch"),
         glob("launch/*.launch.py")),

        # Install config files
        (os.path.join("share", package_name, "config"),
         glob("config/*.yaml")),
    ],
    install_requires=["setuptools"],
    zip_safe=True,
    maintainer="Audrey",
    maintainer_email="audrey@example.com",
    description="TurtleSim Controller — Week 15 ML Journey",
    license="MIT",
    entry_points={
        "console_scripts": [
            # Format: "executable_name = package.module:main_function"
            "turtle_mover = turtle_controller.turtle_mover:main",
            "pose_listener = turtle_controller.pose_listener:main",
            "smart_mover = turtle_controller.smart_mover:main",
            "turtle_commander_server = turtle_controller.turtle_commander_server:main",
            "rotate_client = turtle_controller.rotate_client:main",
        ],
    },
)
'''

print("📄 package.xml:")
print(package_xml)
print("\n📄 setup.py:")
print(setup_py)

print("""
📋 BUILD THE PACKAGE:
─────────────────────────────────────────────────────────────
  # Copy your scripts from Day 100-101 into the package
  cp ~/ros2_ws/src/turtle_mover.py \\
     ~/ros2_ws/src/turtle_controller/turtle_controller/
  cp ~/ros2_ws/src/pose_listener.py \\
     ~/ros2_ws/src/turtle_controller/turtle_controller/
  cp ~/ros2_ws/src/smart_mover.py \\
     ~/ros2_ws/src/turtle_controller/turtle_controller/
  cp ~/ros2_ws/src/turtle_commander_server.py \\
     ~/ros2_ws/src/turtle_controller/turtle_controller/
  cp ~/ros2_ws/src/rotate_client.py \\
     ~/ros2_ws/src/turtle_controller/turtle_controller/

  # Build the workspace
  cd ~/ros2_ws
  colcon build --packages-select turtle_controller

  # Source the new install
  source install/setup.bash

  # Now you can run nodes with ros2 run instead of python3:
  ros2 run turtle_controller turtle_mover
  ros2 run turtle_controller pose_listener
  ros2 run turtle_controller smart_mover
""")

print("\n✅ Exercise 1.2 Complete!")
print("=" * 80)
```

```python
# ==================================================
# EXERCISE 2.1: WHAT ARE LAUNCH FILES?
# ==================================================

print("\n" + "=" * 80)
print("EXERCISE 2.1: What are Launch Files?")
print("=" * 80)

"""
📖 THEORY: Launch Files

A launch file starts multiple nodes with a single command.
Without launch files you need separate terminals for each node.

With a launch file:
  ros2 launch turtle_controller turtlesim.launch.py

Starts: turtlesim + smart_mover + pose_listener + commander_server
        all at once, all configured, all with correct parameters.

Launch files in ROS2 Humble are Python files (not XML like ROS1).
"""

print("""
📋 LAUNCH FILE COMPONENTS:
─────────────────────────────────────────────────────────────
  from launch import LaunchDescription
    → The return type — describes what to launch

  Node(...)
    → Launch a single node

  IncludeLaunchDescription(...)
    → Include another launch file

  DeclareLaunchArgument(...)
    → Define CLI arguments (overridable at launch time)

  LaunchConfiguration(...)
    → Read the value of a launch argument

  ExecuteProcess(...)
    → Run any shell command

📋 BASIC LAUNCH FILE STRUCTURE:
─────────────────────────────────────────────────────────────
  from launch import LaunchDescription
  from launch_ros.actions import Node

  def generate_launch_description():
      return LaunchDescription([
          Node(...),
          Node(...),
      ])
""")

print("\n✅ Exercise 2.1 Complete!")
print("=" * 80)
```

```python
# ==================================================
# EXERCISE 2.2: WRITING LAUNCH FILES
# ==================================================

print("\n" + "=" * 80)
print("EXERCISE 2.2: Writing Launch Files")
print("=" * 80)

basic_launch_py = '''
# ============================================================
# launch/turtlesim_basic.launch.py — Basic Launch File
# Starts TurtleSim + SmartMover + PoseListener
#
# Save to: ~/ros2_ws/src/turtle_controller/launch/
# Run with: ros2 launch turtle_controller turtlesim_basic.launch.py
# ============================================================

from launch import LaunchDescription
from launch_ros.actions import Node


def generate_launch_description():
    """
    Launch TurtleSim + SmartMover + PoseListener together.
    All nodes start simultaneously in the same terminal.
    """
    return LaunchDescription([

        # ── TurtleSim Simulator ───────────────────────────────
        Node(
            package="turtlesim",
            executable="turtlesim_node",
            name="turtlesim",
            output="screen",
        ),

        # ── Smart Mover (boundary-aware controller) ───────────
        Node(
            package="turtle_controller",
            executable="smart_mover",
            name="smart_mover",
            output="screen",
        ),

        # ── Pose Listener (position monitor) ─────────────────
        Node(
            package="turtle_controller",
            executable="pose_listener",
            name="pose_listener",
            output="screen",
        ),

    ])
'''

advanced_launch_py = '''
# ============================================================
# launch/turtlesim_full.launch.py — Full Launch File
# With arguments, parameters, and namespace support
#
# Run with:
#   ros2 launch turtle_controller turtlesim_full.launch.py
#   ros2 launch turtle_controller turtlesim_full.launch.py \\
#     movement_pattern:=square boundary_margin:=3.0
# ============================================================

from launch import LaunchDescription
from launch.actions import DeclareLaunchArgument, LogInfo
from launch.substitutions import LaunchConfiguration
from launch_ros.actions import Node
import os
from ament_index_python.packages import get_package_share_directory


def generate_launch_description():

    # ── Declare Launch Arguments (overridable from CLI) ───────
    movement_pattern_arg = DeclareLaunchArgument(
        "movement_pattern",
        default_value="circle",
        description="Movement pattern: circle, square, figure8",
    )

    boundary_margin_arg = DeclareLaunchArgument(
        "boundary_margin",
        default_value="2.0",
        description="How close to wall before turning (units)",
    )

    forward_speed_arg = DeclareLaunchArgument(
        "forward_speed",
        default_value="2.0",
        description="Forward speed of the turtle",
    )

    # ── Read launch argument values ───────────────────────────
    movement_pattern = LaunchConfiguration("movement_pattern")
    boundary_margin  = LaunchConfiguration("boundary_margin")
    forward_speed    = LaunchConfiguration("forward_speed")

    # ── Config file path ──────────────────────────────────────
    pkg_dir = get_package_share_directory("turtle_controller")
    config_file = os.path.join(pkg_dir, "config", "params.yaml")

    return LaunchDescription([

        # Declare arguments first
        movement_pattern_arg,
        boundary_margin_arg,
        forward_speed_arg,

        # Log what we're launching
        LogInfo(msg=["Launching TurtleSim Controller | ",
                     "Pattern: ", movement_pattern]),

        # ── TurtleSim ─────────────────────────────────────────
        Node(
            package="turtlesim",
            executable="turtlesim_node",
            name="turtlesim",
            output="screen",
        ),

        # ── Smart Mover with parameters ───────────────────────
        Node(
            package="turtle_controller",
            executable="smart_mover",
            name="smart_mover",
            output="screen",
            parameters=[
                {
                    "boundary_margin": boundary_margin,
                    "forward_speed": forward_speed,
                },
                config_file,   # Also load from YAML (YAML overrides above)
            ],
        ),

        # ── Pose Listener ─────────────────────────────────────
        Node(
            package="turtle_controller",
            executable="pose_listener",
            name="pose_listener",
            output="screen",
            parameters=[config_file],
        ),

        # ── Commander Server ──────────────────────────────────
        Node(
            package="turtle_controller",
            executable="turtle_commander_server",
            name="turtle_commander_server",
            output="screen",
        ),

    ])
'''

print("📄 launch/turtlesim_basic.launch.py:")
print(basic_launch_py)
print("\n📄 launch/turtlesim_full.launch.py:")
print(advanced_launch_py)

print("""
📋 HOW TO RUN:
─────────────────────────────────────────────────────────────
  # After colcon build + source install/setup.bash:

  # Basic launch (default settings)
  ros2 launch turtle_controller turtlesim_basic.launch.py

  # Full launch with custom arguments
  ros2 launch turtle_controller turtlesim_full.launch.py \\
    movement_pattern:=square boundary_margin:=3.0

  # List available launch arguments
  ros2 launch turtle_controller turtlesim_full.launch.py --show-args

📋 KEY LAUNCH FILE CONCEPTS:

  output="screen"
    → Print node logs to the terminal (vs "log" = save to file only)

  parameters=[{"key": value}]
    → Pass parameters directly to the node

  parameters=[config_file]
    → Load parameters from a YAML file

  LaunchConfiguration("arg_name")
    → Read the value of a declared argument
    → Can be passed directly into parameters=[]
""")

print("\n✅ Exercise 2.2 Complete!")
print("=" * 80)
```

```python
# ==================================================
# EXERCISE 3.1: ROS2 PARAMETERS
# ==================================================

print("\n" + "=" * 80)
print("EXERCISE 3.1: ROS2 Parameters")
print("=" * 80)

"""
📖 THEORY: Parameters

Parameters make nodes configurable without code changes.
Instead of hardcoding values like FORWARD_SPEED = 2.0,
you declare them as parameters that can be set:
- At launch time (via launch file or CLI)
- At runtime (via ros2 param set)
- From a YAML config file

This is equivalent to environment variables in Docker (Week 14).
"""

params_yaml = '''
# ============================================================
# config/params.yaml — Parameter configuration file
# Located at: ~/ros2_ws/src/turtle_controller/config/params.yaml
# ============================================================

# Parameters are namespaced by node name
smart_mover:
  ros__parameters:
    forward_speed: 2.0
    turn_speed: 1.5
    boundary_margin: 2.0
    canvas_min: 1.0
    canvas_max: 10.08
    log_interval: 30         # Log every Nth pose message

pose_listener:
  ros__parameters:
    log_interval: 30
    boundary_margin: 2.0

turtle_commander_server:
  ros__parameters:
    spin_speed: 2.0
    home_x: 5.544
    home_y: 5.544
'''

smart_mover_with_params_py = '''
# ============================================================
# turtle_controller/smart_mover.py — Updated with Parameters
# ============================================================

import rclpy
from rclpy.node import Node
from geometry_msgs.msg import Twist
from turtlesim.msg import Pose
import math


class SmartMover(Node):

    def __init__(self):
        super().__init__("smart_mover")

        # ── Declare Parameters (name, default_value) ──────────
        self.declare_parameter("forward_speed", 2.0)
        self.declare_parameter("turn_speed", 1.5)
        self.declare_parameter("boundary_margin", 2.0)
        self.declare_parameter("canvas_min", 1.0)
        self.declare_parameter("canvas_max", 10.08)

        # ── Read Parameter Values ──────────────────────────────
        self.forward_speed    = self.get_parameter("forward_speed").value
        self.turn_speed       = self.get_parameter("turn_speed").value
        self.boundary_margin  = self.get_parameter("boundary_margin").value
        self.canvas_min       = self.get_parameter("canvas_min").value
        self.canvas_max       = self.get_parameter("canvas_max").value

        self.get_logger().info(
            f"SmartMover started | "
            f"speed={self.forward_speed} | "
            f"margin={self.boundary_margin}"
        )

        # Publisher + Subscriber + Timer (same as before)
        self.cmd_publisher = self.create_publisher(
            Twist, "/turtle1/cmd_vel", 10
        )
        self.pose_subscription = self.create_subscription(
            Pose, "/turtle1/pose", self.pose_callback, 10
        )
        self.timer = self.create_timer(0.1, self.control_loop)

        self.current_pose = None
        self.near_boundary = False

    def pose_callback(self, msg: Pose):
        self.current_pose = msg
        near_x = (msg.x < self.canvas_min + self.boundary_margin or
                  msg.x > self.canvas_max - self.boundary_margin)
        near_y = (msg.y < self.canvas_min + self.boundary_margin or
                  msg.y > self.canvas_max - self.boundary_margin)
        self.near_boundary = near_x or near_y

    def control_loop(self):
        if self.current_pose is None:
            return

        msg = Twist()
        if self.near_boundary:
            msg.linear.x = 0.0
            msg.angular.z = self.turn_speed      # ← from parameter
        else:
            msg.linear.x = self.forward_speed    # ← from parameter
            msg.angular.z = 0.0

        self.cmd_publisher.publish(msg)


def main(args=None):
    rclpy.init(args=args)
    node = SmartMover()
    try:
        rclpy.spin(node)
    except KeyboardInterrupt:
        stop_msg = Twist()
        node.cmd_publisher.publish(stop_msg)
    finally:
        node.destroy_node()
        rclpy.shutdown()


if __name__ == "__main__":
    main()
'''

print("📄 config/params.yaml:")
print(params_yaml)
print("\n📄 turtle_controller/smart_mover.py (with parameters):")
print(smart_mover_with_params_py)

print("""
📋 USING PARAMETERS FROM CLI:
─────────────────────────────────────────────────────────────
  # List all parameters for a running node
  ros2 param list /smart_mover

  # Get a specific parameter value
  ros2 param get /smart_mover forward_speed

  # Set a parameter at runtime (live update!)
  ros2 param set /smart_mover forward_speed 3.0
  ros2 param set /smart_mover boundary_margin 3.5

  # Dump all parameters to YAML
  ros2 param dump /smart_mover

  # Load parameters from file at runtime
  ros2 param load /smart_mover config/params.yaml

📋 PARAMETERS vs ENVIRONMENT VARIABLES (Week 14 comparison):

  Docker ENV vars:                ROS2 Parameters:
  ─────────────────────────       ─────────────────────────────
  Set at container start          Set at node start or runtime
  String only                     Typed (int, float, bool, list)
  Can\'t change without restart   Can change at runtime!
  os.environ.get("KEY", default)  self.get_parameter("key").value
  .env file                       params.yaml file
""")

print("\n✅ Exercise 3.1 Complete!")
print("=" * 80)
```
