```python
# ==================================================
# EXERCISE 1.1: WHAT IS ROS2?
# ==================================================

print("\n" + "=" * 80)
print("EXERCISE 1.1: What is ROS2?")
print("=" * 80)

"""
📖 THEORY: ROS2 (Robot Operating System 2)

ROS2 is NOT an operating system — it's a middleware framework
for building robot software. It provides:
- A communication layer between software components
- Standard message types for sensors and actuators
- Tools for visualization, debugging, and simulation
- A package ecosystem with thousands of ready-made components

ROS1 vs ROS2:
- ROS1: older, Linux-only, no real-time support, single master node
- ROS2: cross-platform, real-time capable, decentralized, DDS-based
         → ROS2 is the present and future of robotics

Who uses ROS2:
- Boston Dynamics (Spot robot)
- NASA (space robotics)
- Autonomous vehicles (Autoware stack)
- Warehouse robots (Amazon Robotics research)
- Drone systems (PX4 integration)
"""

print("""
📋 ROS2 CORE CONCEPTS:

  ─── Node ────────────────────────────────────────────────────
  A single executable process that does ONE thing.
  Think of it like a microservice in software engineering.

  Examples:
  ├── camera_node       → reads camera, publishes images
  ├── detector_node     → subscribes to images, publishes detections
  ├── controller_node   → subscribes to detections, publishes commands
  └── motor_node        → subscribes to commands, drives motors

  ─── Topic ───────────────────────────────────────────────────
  A named channel for streaming data between nodes.
  Publishers send data, Subscribers receive data.
  Many-to-many: multiple publishers → multiple subscribers.

  Examples:
  ├── /camera/image_raw     → camera images (sensor_msgs/Image)
  ├── /detections           → detected objects (custom msg)
  ├── /cmd_vel              → velocity commands (geometry_msgs/Twist)
  └── /odom                 → robot position (nav_msgs/Odometry)

  ─── Message ─────────────────────────────────────────────────
  The data structure sent over a topic.
  Like typed JSON — strongly typed, language-agnostic.

  ─── Service ─────────────────────────────────────────────────
  Request/Response communication (like an API call).
  Synchronous — caller waits for response.

  ─── Action ──────────────────────────────────────────────────
  Goal/Feedback/Result communication for long-running tasks.
  Asynchronous — caller gets progress updates, can cancel.

  ─── Package ─────────────────────────────────────────────────
  A collection of nodes, configs, launch files, and messages.
  Like a Python package but for ROS2 components.

  ─── Workspace ───────────────────────────────────────────────
  A directory where you build and install ROS2 packages.
  Standard name: ~/ros2_ws
""")

print("""
📊 ROS2 ARCHITECTURE vs WEEK 14 API ARCHITECTURE:

  Week 14 (FastAPI):              Week 15 (ROS2):
  ─────────────────────           ────────────────────────────
  HTTP endpoints                  Topics (publish/subscribe)
  Request/Response                Services (request/response)
  Background tasks                Actions (goal/feedback/result)
  Middleware                      Node lifecycle management
  Docker containers               ROS2 nodes
  Redis pub/sub                   ROS2 DDS middleware

  The concepts map almost 1:1 — you already understand the
  architecture patterns from Week 14! ✅
""")

print("\n✅ Exercise 1.1 Complete!")
print("=" * 80)
```

```python
# ==================================================
# EXERCISE 1.2: ROS2 INSTALLATION (WSL2)
# ==================================================

print("\n" + "=" * 80)
print("EXERCISE 1.2: ROS2 Humble Installation on WSL2")
print("=" * 80)

"""
📖 THEORY: ROS2 Humble

ROS2 Humble Hawksbill is the current LTS (Long Term Support) release.
LTS = supported until May 2027 — safe to build projects on.

Installation target: WSL2 Ubuntu 22.04 (Jammy)
ROS2 Humble requires Ubuntu 22.04 — perfect match.
"""

print("""
📋 STEP 1: Open WSL2 Ubuntu Terminal
─────────────────────────────────────────────────────────────
In Windows, open PowerShell or Windows Terminal and run:
  wsl

Or search "Ubuntu" in Start menu.

Verify you're on Ubuntu 22.04:
  lsb_release -a

Expected output:
  Description: Ubuntu 22.04.x LTS


📋 STEP 2: Set Locale
─────────────────────────────────────────────────────────────
  sudo locale-gen en_US en_US.UTF-8
  sudo update-locale LC_ALL=en_US.UTF-8 LANG=en_US.UTF-8
  export LANG=en_US.UTF-8


📋 STEP 3: Add ROS2 Repository
─────────────────────────────────────────────────────────────
  # Add universe repository
  sudo apt install software-properties-common -y
  sudo add-apt-repository universe

  # Add ROS2 GPG key
  sudo apt update && sudo apt install curl -y
  sudo curl -sSL https://raw.githubusercontent.com/ros/rosdistro/master/ros.key \\
    -o /usr/share/keyrings/ros-archive-keyring.gpg

  # Add ROS2 repository to sources
  echo "deb [arch=$(dpkg --print-architecture) \\
    signed-by=/usr/share/keyrings/ros-archive-keyring.gpg] \\
    http://packages.ros.org/ros2/ubuntu \\
    $(. /etc/os-release && echo $UBUNTU_CODENAME) main" | \\
    sudo tee /etc/apt/sources.list.d/ros2.list > /dev/null


📋 STEP 4: Install ROS2 Humble
─────────────────────────────────────────────────────────────
  sudo apt update
  sudo apt upgrade -y

  # Desktop install (includes ROS, RViz, demos, tutorials)
  sudo apt install ros-humble-desktop -y

  # Development tools
  sudo apt install ros-dev-tools -y

  ⏱️  This takes 5-15 minutes depending on internet speed.


📋 STEP 5: Configure Environment
─────────────────────────────────────────────────────────────
  # Add ROS2 to your shell startup (so it auto-sources)
  echo "source /opt/ros/humble/setup.bash" >> ~/.bashrc
  source ~/.bashrc

  # Verify installation
  ros2 --version

  Expected output:
    ros2 humble


📋 STEP 6: Create Your Workspace
─────────────────────────────────────────────────────────────
  mkdir -p ~/ros2_ws/src
  cd ~/ros2_ws
  colcon build
  echo "source ~/ros2_ws/install/setup.bash" >> ~/.bashrc
  source ~/.bashrc

  Your workspace structure:
  ~/ros2_ws/
  ├── src/          ← your packages go here
  ├── build/        ← colcon build output (auto-generated)
  ├── install/      ← installed packages (auto-generated)
  └── log/          ← build logs (auto-generated)
""")

print("\n✅ Exercise 1.2 Complete!")
print("=" * 80)
```

```python
# ==================================================
# EXERCISE 2.1: LAUNCHING TURTLESIM
# ==================================================

print("\n" + "=" * 80)
print("EXERCISE 2.1: Launching TurtleSim")
print("=" * 80)

"""
📖 THEORY: TurtleSim

TurtleSim is ROS2's "Hello World" — a simple 2D turtle simulator.
It has:
- A turtle that moves around a canvas
- Topics for velocity commands and pose (position)
- Services for teleporting, spawning, resetting
- Perfect for learning ROS2 concepts without real hardware
"""

print("""
📋 LAUNCH TURTLESIM:
─────────────────────────────────────────────────────────────
Open TWO WSL2 terminal windows side by side.

Terminal 1 — Launch TurtleSim node:
  ros2 run turtlesim turtlesim_node

  → A blue window with a turtle appears ✅
  → Terminal shows: [INFO] Starting turtlesim with node name /turtlesim

Terminal 2 — Launch keyboard teleop:
  ros2 run turtlesim turtle_teleop_key

  → Use arrow keys to move the turtle
  → The turtle draws lines as it moves

📋 WHAT JUST HAPPENED:
─────────────────────────────────────────────────────────────
  Two nodes are now running:

  /turtlesim         ← the simulator node
    Subscribes to: /turtle1/cmd_vel  (receives movement commands)
    Publishes to:  /turtle1/pose     (broadcasts turtle position)

  /teleop_turtle     ← the keyboard control node
    Publishes to:  /turtle1/cmd_vel  (sends movement commands)

  Data flow:
  [keyboard input] → teleop_turtle → /turtle1/cmd_vel → turtlesim → [turtle moves]
                                                    ↓
                                           /turtle1/pose → [any subscriber]

📋 ros2 run SYNTAX:
─────────────────────────────────────────────────────────────
  ros2 run <package_name> <executable_name>

  ros2 run turtlesim turtlesim_node
           ↑                ↑
           package          executable inside that package
""")

print("\n✅ Exercise 2.1 Complete!")
print("=" * 80)
```

```python
# ==================================================
# EXERCISE 2.2: ROS2 CLI COMMANDS
# ==================================================

print("\n" + "=" * 80)
print("EXERCISE 2.2: Essential ROS2 CLI Commands")
print("=" * 80)

"""
📖 THEORY: ROS2 CLI

ROS2 has a comprehensive command-line interface for inspecting
and interacting with a running ROS2 system.
Everything follows: ros2 <verb> <noun> [options]
"""

print("""
📋 NODE COMMANDS:
─────────────────────────────────────────────────────────────
  ros2 node list
    → List all running nodes
    Output: /turtlesim
            /teleop_turtle

  ros2 node info /turtlesim
    → Detailed info about a node
    Output: Subscribers, Publishers, Services, Actions

  ros2 node info /teleop_turtle
    → See what /teleop_turtle publishes


📋 TOPIC COMMANDS:
─────────────────────────────────────────────────────────────
  ros2 topic list
    → List all active topics
    Output: /turtle1/cmd_vel
            /turtle1/pose
            /rosout
            /parameter_events

  ros2 topic list -t
    → List topics WITH their message types
    Output: /turtle1/cmd_vel [geometry_msgs/msg/Twist]
            /turtle1/pose    [turtlesim/msg/Pose]

  ros2 topic echo /turtle1/pose
    → Watch live data on a topic (Ctrl+C to stop)
    Output (updates as turtle moves):
      x: 5.544444561004639
      y: 5.544444561004639
      theta: 0.0
      linear_velocity: 0.0
      angular_velocity: 0.0

  ros2 topic info /turtle1/cmd_vel
    → See publisher/subscriber count for a topic
    Output: Type: geometry_msgs/msg/Twist
            Publisher count: 1
            Subscription count: 1

  ros2 topic hz /turtle1/pose
    → Measure publish rate (messages per second)
    Output: average rate: 62.5 Hz


📋 MESSAGE COMMANDS:
─────────────────────────────────────────────────────────────
  ros2 interface show geometry_msgs/msg/Twist
    → Show the fields of a message type
    Output:
      Vector3 linear
        float64 x
        float64 y
        float64 z
      Vector3 angular
        float64 x
        float64 y
        float64 z

  ros2 interface show turtlesim/msg/Pose
    → Show TurtleSim pose message structure
    Output:
      float32 x
      float32 y
      float32 theta
      float32 linear_velocity
      float32 angular_velocity


📋 PUBLISH DIRECTLY FROM CLI (no code needed!):
─────────────────────────────────────────────────────────────
  # Move turtle forward (one message)
  ros2 topic pub --once /turtle1/cmd_vel geometry_msgs/msg/Twist \\
    "{linear: {x: 2.0, y: 0.0, z: 0.0}, angular: {x: 0.0, y: 0.0, z: 0.0}}"

  # Move turtle in a circle (continuous)
  ros2 topic pub /turtle1/cmd_vel geometry_msgs/msg/Twist \\
    "{linear: {x: 1.0}, angular: {z: 1.0}}"

  # Stop turtle
  ros2 topic pub --once /turtle1/cmd_vel geometry_msgs/msg/Twist \\
    "{linear: {x: 0.0}, angular: {z: 0.0}}"


📋 SERVICE COMMANDS:
─────────────────────────────────────────────────────────────
  ros2 service list
    → List all available services
    Output: /turtle1/teleport_absolute
            /turtle1/teleport_relative
            /turtle1/set_pen
            /turtlesim/reset
            /spawn
            /kill

  ros2 service call /turtlesim/reset std_srvs/srv/Empty
    → Reset turtlesim (clears the canvas, resets turtle position)

  ros2 service call /turtle1/teleport_absolute \\
    turtlesim/srv/TeleportAbsolute "{x: 5.0, y: 5.0, theta: 0.0}"
    → Teleport turtle to center of canvas

  ros2 service call /spawn turtlesim/srv/Spawn \\
    "{x: 3.0, y: 3.0, theta: 0.0, name: 'turtle2'}"
    → Spawn a second turtle!
""")

print("\n✅ Exercise 2.2 Complete!")
print("=" * 80)
```

```python
# ==================================================
# EXERCISE 2.3: ROS2 GRAPH VISUALIZATION
# ==================================================

print("\n" + "=" * 80)
print("EXERCISE 2.3: Visualizing the ROS2 Graph")
print("=" * 80)

"""
📖 THEORY: rqt_graph

rqt_graph is a GUI tool that shows the node/topic graph visually.
It's the best way to understand how data flows in a ROS2 system.
"""

print("""
📋 LAUNCH RQT_GRAPH:
─────────────────────────────────────────────────────────────
  # In a new terminal (with TurtleSim running):
  rqt_graph

  → A window shows nodes as ovals and topics as arrows

  With TurtleSim + teleop running, you'll see:
  [/teleop_turtle] ──/turtle1/cmd_vel──▶ [/turtlesim]
                                               ↓
                                      /turtle1/pose
                                          (published)

📋 OTHER USEFUL RQT TOOLS:
─────────────────────────────────────────────────────────────
  rqt_plot
    → Plot topic data over time (great for visualizing pose)
    ros2 run rqt_plot rqt_plot /turtle1/pose/x:y

  rqt
    → Full RQT dashboard (includes all tools above)
    → Open Plugins menu to add panels

  rviz2
    → 3D visualization tool (used for real robot data)
    → Won't be useful until Day 103 (sensor data)
    ros2 run rviz2 rviz2

📋 RECORD AND PLAY BACK ROS2 DATA:
─────────────────────────────────────────────────────────────
  # Record all topics to a bag file
  ros2 bag record -a -o my_session

  # Play back the recording
  ros2 bag play my_session

  # Record specific topics only
  ros2 bag record /turtle1/pose /turtle1/cmd_vel -o turtle_session

  → Bag files are like video recordings of your ROS2 system
  → Great for debugging — replay sensor data without the real robot
""")

print("\n✅ Exercise 2.3 Complete!")
print("=" * 80)
```
